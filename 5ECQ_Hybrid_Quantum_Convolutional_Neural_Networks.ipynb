# **Introducción**

<p align="justify" style="text-align: justify; width: 100%;">

La <b>clasificación de imágenes</b> busca <b>asignar una etiqueta</b> a una imagen a partir de la información contenida en sus píxeles. Algunos ejemplos incluyen:

</p>

<ul>
  <li><b>Radiografías:</b> Normal o Neumonía.</li>
  <li><b>Animales:</b> Gato, perro o ave.</li>
  <li><b>Dígitos manuscritos:</b> 0, 1, 2, ..., 9.</li>
</ul>

Screenshot 2026-08-07 005023.png

<p align="center">
  <small><b>Figura 1.</b> Ejemplo de <b>clasificación de imágenes</b>. <i>Tomado de [Fig. 1].</i></small>
</p>

<p align="justify" style="text-align: justify; width: 100%;">

Si buscamos utilizar un <b>circuito cuántico para clasificar imágenes</b>, podemos partir de la información más elemental que las compone: sus <b>píxeles</b>. Los valores de cada píxel pueden utilizarse para <b>codificar información en los qubits</b> y construir así la entrada del circuito cuántico.

</p>

Screenshot 2026-08-07 010629.png

<p align="center">
  <small><b>Figura 2.</b> Codificación del <b>valor de un píxel</b> en el estado de un <b>qubit</b> mediante una rotación angular. <i>Ilustración del qubit de [Fig. 2].</i></small>
</p>

<p align="justify" style="text-align: justify; width: 100%;">

Con esta estrategia podemos pensar en <b>utilizar un VQC para clasificar imágenes</b>, codificando la información de sus píxeles en los qubits.

Sin embargo, aparece una limitación importante: al asignar <b>un qubit por cada píxel</b>, el número de qubits necesarios aumenta rápidamente con el tamaño de la imagen.

</p>

Screenshot 2026-08-07 011234.png

<p align="center">
  <small><b>Figura 3.</b> Codificación <b>píxel a qubit</b> de una imagen de <b>4×4</b> mediante un <b>VQC de 16 qubits</b>.</small>
</p>

<p align="justify" style="text-align: justify; width: 100%;">

Ahora bien, ¿qué sucede si <b>aumentamos la resolución de la imagen</b>? Al mantener la asignación de <b>un qubit por píxel</b>, una imagen de <b>8×8</b> requeriría <b>64 qubits</b>, mientras que una de <b>16×16</b> necesitaría <b>256 qubits</b>.

</p>

Screenshot 2026-08-07 011437.png

<p align="center">
  <small><b>Figura 4.</b> Crecimiento del número de <b>qubits</b> al aumentar la resolución: <b>(a)</b> 8×8 → 64 qubits y <b>(b)</b> 16×16 → 256 qubits, utilizando misma estrategia.</small>
</p>

<p align="justify" style="text-align: justify; width: 100%;">

Esta relación evidencia que la estrategia de <b>un qubit por píxel no es escalable</b>. Incluso una imagen de <b>16×16 píxeles</b> requiere <b>256 qubits</b>, mientras que aplicaciones reales, como el <b>análisis de imágenes médicas</b>, suelen trabajar con resoluciones considerablemente mayores.

</p>

# **1. Redes Neuronales Convolucionales (CNN)**

<p align="justify" style="text-align: justify; width: 100%;">

Una <b>Red Neuronal Convolucional (CNN)</b> es un modelo de <i>Deep Learning</i> que emplea <b>filtros entrenables</b> para extraer características de una imagen y utilizarlas en tareas de <b>clasificación, detección o reconocimiento</b>.

</p>

<ul>
  <li><b>Convoluciones:</b> extraen patrones locales.</li>
  <li><b>Pooling:</b> reduce la dimensión espacial.</li>
</ul>

Screenshot 2026-08-06 103158.png

<p align="center">
  <small><b>Figura 5.</b> Arquitectura general de una red neuronal convolucional para clasificación de imágenes. <i>Tomado de [Fig. 5].</i></small>
</p>

<p align="justify" style="text-align: justify; width: 100%;">

Los <b>filtros convolucionales constituyen el mecanismo principal para extraer características</b> en una CNN. Esta idea servirá como base para introducir posteriormente la <b>quanvolución, donde los filtros clásicos son reemplazados por circuitos cuánticos</b>.

</p>

# **2. Quanvoluciones**

<p align="justify" style="text-align: justify; width: 100%;">

La <b>quanvolución procesa localmente parches de una imagen mediante circuitos cuánticos</b>. Cada parche <b>se codifica en un estado cuántico</b>, <b>se transforma con un circuito parametrizado</b> y <b>se mide para generar nuevos mapas de características</b>, de manera análoga a un <b>filtro convolucional clásico</b>.

</p>

Screenshot 2026-08-06 132430.png

<p align="center">
  <small><b>Figura 7.</b> Procesamiento quanvolucional aplicado a parches locales de <b>2×2</b>. Cada parche se <b>codifica</b>, se <b>procesa mediante un circuito cuántico parametrizado</b> y se <b>mide para generar mapas de características</b>. </i></small>
</p>

<p align="justify" style="text-align: justify; width: 100%;">

De esta forma, en lugar de utilizar <b>un qubit por cada píxel de la imagen completa</b>, únicamente necesitamos <b>4 qubits</b> para procesar parches de <b>2×2</b>, reutilizando el <b>mismo circuito cuántico</b> a lo largo de toda la imagen.

</p>

## **3.2 Arquitectura del modelo híbrido cuántico-clásico**

<p align="justify" style="text-align: justify; width: 100%;">
<b>Inicialización del dispositivo cuántico:</b>
</p>

```python
dispositivo_cuantico_SGJ = qml.device("default.qubit.torch", wires=NUMERO_QUBITS_SGJ, shots=None, torch_device="cuda" if DISPOSITIVO_SGJ.type == "cuda" else "cpu")
```

<p align="justify" style="text-align: justify; width: 100%;">
<b>Construcción del circuito cuántico:</b>
</p>

```python
@qml.qnode(dispositivo_cuantico_SGJ, interface="torch", diff_method="backprop")
def circuito_cuantico_SGJ(angulos_parches_SGJ, pesos_cuanticos_SGJ):
    """Circuito quanvolucional batched de SGJ-HQCNN."""

    # Codificación de los cuatro píxeles
    for i in range(NUMERO_QUBITS_SGJ):
        qml.RY(angulos_parches_SGJ[:, i], wires=i)

    # Ansatz variacional
    for capa in range(NUMERO_CAPAS_CUANTICAS_SGJ):
        for qubit in range(NUMERO_QUBITS_SGJ):
            qml.RY(pesos_cuanticos_SGJ[capa, qubit, 0], wires=qubit)
            qml.RZ(pesos_cuanticos_SGJ[capa, qubit, 1], wires=qubit)
        qml.CNOT(wires=[0, 1])
        qml.CNOT(wires=[1, 2])
        qml.CNOT(wires=[2, 3])
        qml.CNOT(wires=[3, 0])

    return tuple(qml.expval(qml.PauliZ(i)) for i in range(NUMERO_QUBITS_SGJ))
```

<p align="justify" style="text-align: justify; width: 100%;">
<b>Implementación de la capa quanvolucional:</b>
</p>

```python
class GonzalezJuarezQuanvolucion(nn.Module):
    """Capa quanvolucional SGJ. Entrada: (B,1,28,28) → Salida: (B,4,14,14)"""

    def __init__(self):
        super().__init__()
        self.pesos_cuanticos_SGJ = nn.Parameter(0.1 * torch.randn(NUMERO_CAPAS_CUANTICAS_SGJ, NUMERO_QUBITS_SGJ, 2, device=DISPOSITIVO_SGJ))
        self.register_buffer("firma_numerica_SGJ", torch.tensor([19,5,2,1,19,20,9,1,14], dtype=torch.int16), persistent=True)

    def forward(self, imagenes_lote_SGJ):
        B = imagenes_lote_SGJ.shape[0]

        # (B,1,28,28) → (B,4,196) y aplanar a (B×196,4)
        parches = F.unfold(imagenes_lote_SGJ, kernel_size=2, stride=2)
        angulos = torch.pi * parches.transpose(1,2).reshape(-1, 4)

        # Procesar en bloques para evitar OOM
        salidas_bloques = []
        for inicio in range(0, len(angulos), TAMANO_BLOQUE_PARCHES_SGJ):
            bloque = angulos[inicio:inicio+TAMANO_BLOQUE_PARCHES_SGJ]
            salida = torch.stack(circuito_cuantico_SGJ(bloque, self.pesos_cuanticos_SGJ), dim=1).float()
            salidas_bloques.append(salida)

        # (B×196,4) → (B,4,14,14)
        return torch.cat(salidas_bloques, dim=0).view(B, 196, 4).permute(0,2,1).reshape(B, 4, 14, 14)
```

<p align="justify" style="text-align: justify; width: 100%;">
<b>Implementación del modelo híbrido cuántico-clásico:</b>
</p>

```python
class GonzalezJuarezHQCNN(nn.Module):

    def __init__(self):
        super().__init__()
        self.capa_quanvolucional = GonzalezJuarezQuanvolucion()
        self.clasificador_clasico = nn.Sequential(
            nn.Conv2d(4, 16, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2),
            nn.Flatten(),
            nn.Linear(16 * 7 * 7, 64),
            nn.ReLU(),
            nn.Dropout(0.20),
            nn.Linear(64, 1)
        )

    def forward(self, imagenes_lote_SGJ):
        mapas = self.capa_quanvolucional(imagenes_lote_SGJ)
        return self.clasificador_clasico(mapas)
```

<p align="justify" style="text-align: justify; width: 100%;">
<b>Inicialización del modelo:</b>
</p>

```python
modelo_hibrido_SGJ = GonzalezJuarezHQCNN().to(DISPOSITIVO_SGJ)

# Contar parámetros
param_cuanticos = sum(p.numel() for p in modelo_hibrido_SGJ.capa_quanvolucional.parameters())
param_clasicos = sum(p.numel() for p in modelo_hibrido_SGJ.clasificador_clasico.parameters())
param_totales = param_cuanticos + param_clasicos

print("ARQUITECTURA HQCNN")
print("28×28 → Quanvolución → 4×14×14 → CNN → 16×7×7 → MLP → 1")
print("\nParámetros")
print(f"Cuánticos : {param_cuanticos:,}")
print(f"Clásicos  : {param_clasicos:,}")
print(f"Totales   : {param_totales:,}")
```

## **3.3 Configuración del entrenamiento**

<p align="justify" style="text-align: justify; width: 100%;">
<b>Ponderación de clases:</b>
</p>

```python
# Preparar pesos para balance de clases
y_ent_np = y_ent_t.numpy().reshape(-1)
clase_normal = int(np.sum(y_ent_np == 0))
clase_neumonia = int(np.sum(y_ent_np == 1))
peso_positivo = torch.tensor([clase_normal / clase_neumonia], dtype=torch.float32, device=DISPOSITIVO_SGJ)
```

<p align="justify" style="text-align: justify; width: 100%;">
<b>Función de pérdida y optimizador:</b>
</p>

```python
# Función de pérdida y optimizador
funcion_perdida_SGJ = nn.BCEWithLogitsLoss(pos_weight=peso_positivo)
optimizador_SGJ = torch.optim.Adam([
    {"params": modelo_hibrido_SGJ.capa_quanvolucional.parameters(), "lr": TASA_APRENDIZAJE_CUANTICA_SGJ},
    {"params": modelo_hibrido_SGJ.clasificador_clasico.parameters(), "lr": TASA_APRENDIZAJE_CLASICA_SGJ, "weight_decay": DECAIMIENTO_PESOS_SGJ}
])
```

<p align="justify" style="text-align: justify; width: 100%;">
<b>Cálculo de métricas de evaluación:</b>
</p>

```python
# Métricas
def calcular_metricas(y_real, probs):
    preds = (probs >= UMBRAL_CLASIFICACION_SGJ).astype(int)
    return {
        "accuracy": accuracy_score(y_real, preds),
        "accuracy_balanceada": balanced_accuracy_score(y_real, preds),
        "roc_auc": roc_auc_score(y_real, probs)
    }
```

<p align="justify" style="text-align: justify; width: 100%;">
<b>Definición de una época de entrenamiento:</b>
</p>

```python
# Función para una época
def ejecutar_epoca(cargador, entrenar=True):
    modelo_hibrido_SGJ.train(entrenar)
    perdida_total = 0.0
    y_todas, probs_todas = [], []

    for imgs, lbls in cargador:
        imgs = imgs.to(DISPOSITIVO_SGJ)
        lbls = lbls.to(DISPOSITIVO_SGJ)

        if entrenar:
            optimizador_SGJ.zero_grad(set_to_none=True)

        with torch.set_grad_enabled(entrenar):
            logits = modelo_hibrido_SGJ(imgs)
            loss = funcion_perdida_SGJ(logits, lbls)
            if entrenar:
                loss.backward()
                optimizador_SGJ.step()

        probs = torch.sigmoid(logits)
        perdida_total += loss.item() * len(imgs)
        y_todas.extend(lbls.detach().cpu().numpy().reshape(-1))
        probs_todas.extend(probs.detach().cpu().numpy().reshape(-1))

    y_todas = np.asarray(y_todas)
    probs_todas = np.asarray(probs_todas)

    return {
        "perdida": perdida_total / len(cargador.dataset),
        **calcular_metricas(y_todas, probs_todas),
        "etiquetas": y_todas,
        "probabilidades": probs_todas
    }
```

<p align="justify" style="text-align: justify; width: 100%;">
<b>Función de visualización de los mapas cuánticos:</b>
</p>

```python
# Visualización de mapas cuánticos
@torch.no_grad()
def mostrar_mapas_cuanticos(titulo):
    """Visualiza la imagen original y sus 4 mapas cuánticos."""
    modelo_hibrido_SGJ.eval()

    ref = imagen_referencia_SGJ.unsqueeze(0).to(DISPOSITIVO_SGJ)
    mapas = modelo_hibrido_SGJ.capa_quanvolucional(ref)[0].detach().cpu().numpy()

    fig, axes = plt.subplots(1, 5, figsize=(14, 3), constrained_layout=True)

    # Imagen original
    axes[0].imshow(imagen_referencia_SGJ[0], cmap="gray")
    axes[0].set_title("Imagen original\n" + ("Neumonía" if etiqueta_referencia_SGJ == 1 else "Normal"))
    axes[0].axis("off")

    # Mapas cuánticos
    for i in range(4):
        axes[i + 1].imshow(mapas[i], cmap="gray", vmin=-1, vmax=1)
        axes[i + 1].set_title(rf"Mapa $\langle Z_{i}\rangle$")
        axes[i + 1].axis("off")

    fig.suptitle(titulo, fontsize=13)
    fig.text(0.99, 0.01, "SGJ-HQCNN · Sebastián González Juárez · UNAM 2026",
             ha="right", va="bottom", fontsize=8, alpha=0.65)

    display(fig)
    plt.close(fig)
```

<p align="justify" style="text-align: justify; width: 100%;">
<b>Inicialización de variables de entrenamiento:</b>
</p>

```python
# Inicializar historial y mejores valores
historial_entrenamiento_SGJ = []

mejor_estado_modelo_SGJ = None
mejor_epoca_SGJ = None
mejor_perdida_validacion_SGJ = float("inf")
```

## **3.4 Entrenamiento del Modelo**

```python
inicio_entrenamiento_SGJ = time.time()

# Bucle de entrenamiento
for epoca_SGJ in range(1, NUMERO_EPOCAS_SGJ + 1):
    inicio_epoca_SGJ = time.time()

    # Entrenamiento y validación
    resultado_entrenamiento_SGJ = ejecutar_epoca(cargador_entrenamiento_SGJ, entrenar=True)
    resultado_validacion_SGJ = ejecutar_epoca(cargador_validacion_SGJ, entrenar=False)
    tiempo_epoca_SGJ = time.time() - inicio_epoca_SGJ

    # Guardar historial
    historial_entrenamiento_SGJ.append({
        "epoca": epoca_SGJ,
        "perdida_entrenamiento": resultado_entrenamiento_SGJ["perdida"],
        "accuracy_entrenamiento": resultado_entrenamiento_SGJ["accuracy"],
        "perdida_validacion": resultado_validacion_SGJ["perdida"],
        "accuracy_validacion": resultado_validacion_SGJ["accuracy"],
        "tiempo_segundos": tiempo_epoca_SGJ
    })

    # Seleccionar mejor modelo por pérdida de validación
    modelo_mejoro_SGJ = resultado_validacion_SGJ["perdida"] < mejor_perdida_validacion_SGJ
    if modelo_mejoro_SGJ:
        mejor_perdida_validacion_SGJ = float(resultado_validacion_SGJ["perdida"])
        mejor_epoca_SGJ = int(epoca_SGJ)
        mejor_estado_modelo_SGJ = copy.deepcopy(modelo_hibrido_SGJ.state_dict())

    # Mostrar resultados de la época
    print(f"\nÉpoca {epoca_SGJ:02d}/{NUMERO_EPOCAS_SGJ} | tiempo={tiempo_epoca_SGJ:.1f}s{'  ← mejor' if modelo_mejoro_SGJ else ''}")
    print(f"  Entrenamiento | pérdida={resultado_entrenamiento_SGJ['perdida']:.6f} | accuracy={resultado_entrenamiento_SGJ['accuracy']:.4f}")
    print(f"  Validación    | pérdida={resultado_validacion_SGJ['perdida']:.6f} | accuracy={resultado_validacion_SGJ['accuracy']:.4f}")

    # Mostrar mapas cuánticos después de cada época
    mostrar_mapas_cuanticos(f"Evolución de los mapas cuánticos — época {epoca_SGJ:02d}")

    # Liberar referencias temporales y memoria
    del resultado_entrenamiento_SGJ, resultado_validacion_SGJ
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

# Restaurar el mejor modelo
if mejor_estado_modelo_SGJ is None:
    raise RuntimeError("No se pudo conservar ningún modelo durante el entrenamiento.")
modelo_hibrido_SGJ.load_state_dict(mejor_estado_modelo_SGJ)
modelo_hibrido_SGJ.eval()

tiempo_total_entrenamiento_SGJ = time.time() - inicio_entrenamiento_SGJ

# Mostrar mapas del modelo restaurado
mostrar_mapas_cuanticos(f"Mapas cuánticos del modelo seleccionado — época {mejor_epoca_SGJ:02d}")

# Resumen
print(f"\nMejor época: {mejor_epoca_SGJ}")
print(f"Mejor pérdida de validación: {mejor_perdida_validacion_SGJ:.6f}")
print(f"Tiempo total: {tiempo_total_entrenamiento_SGJ / 60:.2f} min")
print(f"Tiempo medio por época: {np.mean([fila_SGJ['tiempo_segundos'] for fila_SGJ in historial_entrenamiento_SGJ]):.2f} s")
```

## **3.5 Desempeño durante el entrenamiento**

<p align="justify" style="text-align: justify; width: 100%;">
<b>Historial del entrenamiento:</b>
</p>

```python
# DataFrame con el historial de entrenamiento
historial_df = pd.DataFrame(historial_entrenamiento_SGJ)
display(historial_df.round({"perdida_entrenamiento": 4, "accuracy_entrenamiento": 4, "perdida_validacion": 4, "accuracy_validacion": 4, "tiempo_segundos": 1}))
```

<p align="justify" style="text-align: justify; width: 100%;">
<b>Mejor época del entrenamiento (Elegida a partir de la menor pérdida de validación):</b>
</p>

```python
# Mostrar la mejor época
mejor_fila = historial_df[historial_df["epoca"] == mejor_epoca_SGJ].copy()
mejor_fila = mejor_fila.rename(columns={
    "epoca": "Época",
    "perdida_entrenamiento": "Pérdida entrenamiento",
    "accuracy_entrenamiento": "Accuracy entrenamiento",
    "perdida_validacion": "Pérdida validación",
    "accuracy_validacion": "Accuracy validación",
    "tiempo_segundos": "Tiempo [s]"
})

print("Modelo seleccionado:")
display(mejor_fila.round({"Pérdida entrenamiento": 4, "Accuracy entrenamiento": 4,
                          "Pérdida validación": 4, "Accuracy validación": 4, "Tiempo [s]": 1}))
```
