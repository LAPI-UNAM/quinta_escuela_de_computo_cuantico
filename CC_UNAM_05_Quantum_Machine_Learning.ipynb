## 1.1. Crecimiento Exponencial de los Datos


En las últimas décadas hemos presenciado una explosión en la disponibilidad de datos y, con ello, un creciente interés en las aplicaciones estadísticas y de *Machine Learning* que permiten aprovecharlos. Esta demanda se debe a un hecho importante: **estos métodos han logrado resultados que antes parecían difíciles de alcanzar**.

El *Machine Learning* ha generado nuevas perspectivas y métodos para analizar y utilizar los datos que producimos diariamente. Sus aplicaciones abarcan múltiples campos, entre ellos las ciencias sociales, los negocios, la biología, la medicina y el procesamiento de imágenes.


La cantidad de datos generados, capturados, copiados y consumidos en el mundo ha crecido con rapidez. Este fenómeno es relevante para el *Machine Learning*: disponer de más datos abre nuevas posibilidades, pero también exige mejores métodos para almacenarlos, procesarlos y extraer información útil.


**Una estimación histórica de IDC:** en el informe *The Digitization of the World: From Edge to Core*, publicado en noviembre de 2018, IDC estimó que la *Global DataSphere* pasaría de 33 zettabytes (ZB) en 2018 a 175 ZB en 2025. Esto equivale a una tasa de crecimiento anual compuesta cercana al 27 %. Es importante interpretar los 175 ZB como una **proyección realizada en 2018**, no como una medición definitiva de 2025.

Las estimaciones se revisan conforme cambian la tecnología y la metodología. IDC mantiene su colección de [pronósticos oficiales](https://www.idc.com/research/forecasts.jsp); su informe [*Worldwide IDC Global DataSphere Forecast, 2025-2029*](https://my.idc.com/getdoc.jsp?containerId=US53363625) corresponde a una edición más reciente. Para consultar el origen de la estimación histórica puede revisarse el [informe de IDC patrocinado por Seagate](https://www.seagate.com/files/www-content/our-story/trends/files/idc-seagate-dataage-whitepaper.pdf).

# 2. Machine Learning

El *Machine Learning* (ML), es un subcampo de las ciencias de la computación que desarrolla algoritmos capaces de construir modelos a partir de una colección de ejemplos. Estos ejemplos pueden provenir de la naturaleza, ser definidos por personas o ser generados por otros algoritmos.

El ML también puede describirse como el proceso de resolver un problema práctico mediante la recopilación de datos y la construcción algorítmica de un modelo estadístico basado en ellos. Después, el modelo se utiliza para hacer predicciones, clasificar información o apoyar la toma de decisiones relacionadas con ese problema.


<div align="center"><img width="650px" style="display:block; margin-left: auto; margin-right: auto;" src="https://images.claudiazm.xyz/ecc26/MLFlow.png"></div>

Un modelo busca representar una situación real de una forma que pueda manipularse con mayor facilidad. Para ello utiliza variables que describen las características relevantes del sistema bajo estudio.


### ¿Una máquina aprende?

Es importante ser claros: las máquinas **no** "aprenden" del mismo modo que una persona. Un algoritmo de aprendizaje automático ajusta una función matemática a partir de una colección de entradas, llamadas *datos de entrenamiento*, y de un objetivo. Idealmente, la función también produce resultados adecuados para entradas nuevas, siempre que provengan de la misma distribución estadística o de una suficientemente parecida a la de los datos de entrenamiento.

## 2.1. Resumen de Machine Learning

Hablemos de *Machine Learning*. Es probable que ya interactúes con sistemas de ML todos los días.

Una de sus aplicaciones más conocidas es la **clasificación**: se dispone de datos asociados con categorías y se busca decidir a cuál pertenece una entrada nueva. Por ejemplo, un filtro de correo recibe un mensaje y estima si es *spam* o si probablemente es un correo deseado.

Otra aplicación es el **clustering** (agrupamiento). En las compras por internet o en las plataformas de *streaming*, una empresa puede identificar grupos de usuarios con comportamientos parecidos. Las preferencias de personas con patrones similares pueden ayudar a generar recomendaciones, aunque los sistemas reales suelen combinar esta idea con otras técnicas.

Los modelos de ML también pueden generar imágenes. Algunos producen rostros ficticios con una apariencia muy realista. En [thispersondoesnotexist.com](https://thispersondoesnotexist.com/), por ejemplo, se genera el rostro de una persona que no existe. Estos ejemplos tienen algo en común: construyen un **modelo que identifica patrones en los datos** y los utiliza para producir una salida. Esa es la idea esencial del *Machine Learning*.


<div align="center"><img width="750px" style="display:block; margin-left: auto; margin-right: auto;" src="https://images.claudiazm.xyz/ecc26/MLExamples.png"></div>

La estructura general de una tarea de ML consiste en utilizar datos para ajustar los parámetros de un modelo y obtener un resultado, como una categoría, un valor numérico o una acción.

Por lo general, se presentan cuatro familias principales de ML, aunque existen enfoques adicionales y combinaciones entre ellas:

- **Supervised learning** (aprendizaje supervisado)
- **Unsupervised learning** (aprendizaje no supervisado)
- **Generative learning** (aprendizaje generativo)
- **Reinforcement learning** (aprendizaje por refuerzo)


<div align="center"><img width="500px" style="display:block; margin-left: auto; margin-right: auto;" src="https://images.claudiazm.xyz/ecc26/MLFamilies.png"></div>

En el **supervised learning** se utilizan datos con etiquetas conocidas. Por ejemplo, una persona puede revisar un conjunto de correos electrónicos e indicar cuáles son *spam*. A partir de esos ejemplos se ajusta un modelo que aprende a separar las clases. Después, el modelo puede estimar la etiqueta de un correo que no formaba parte del conjunto de entrenamiento.


En el **unsupervised learning** los datos no tienen etiquetas asignadas. Aquí aparecen tareas como el agrupamiento en clústeres, donde se buscan características compartidas entre los puntos para organizarlos en grupos. Existen numerosos modelos para hacerlo. En una plataforma de video o una tienda en línea, por ejemplo, el resultado podría ser una segmentación de usuarios con comportamientos parecidos.


En el **generative learning** el objetivo no es clasificar o agrupar datos, sino aprender la estructura subyacente de un conjunto de información para crear contenido completamente nuevo. Por ejemplo, un modelo alimentado con miles de ilustraciones puede aprender las pautas visuales y el estilo del arte para generar una imagen inédita a partir de una descripción en texto. Al capturar las reglas y patrones de los datos de entrenamiento, el sistema es capaz de producir texto, imágenes, audio o código que imitan el dominio original.

En el **reinforcement learning** un agente interactúa con un entorno y recibe recompensas o penalizaciones. Pensemos en el videojuego *Super Mario*: el agente controla a Mario y puede recibir una recompensa al completar el nivel o una penalización al caer por un precipicio o chocar con un *Goomba*. El modelo, que en problemas complejos puede ser una red neuronal, aprende una estrategia que busca maximizar la recompensa acumulada.

## 2.2. ¿Cómo "aprende" un modelo de ML?

<div align="center"><img width="900px" style="display:block; margin-left: auto; margin-right: auto;" src="https://images.claudiazm.xyz/ecc26/GradientDescentLearning.gif"></div>

$$J(\vec{w}, b) = \frac{1}{m} \sum_{i=1}^{m} L(\hat{y}^{(i)}, y^{(i)})$$

A $J$ se le conoce como **función de costo** o **función objetivo**. Durante el entrenamiento buscamos minimizarla, pues resume de manera matemática qué tan alejadas están las predicciones del modelo de los valores esperados en el conjunto de datos.


Una **función de pérdida** (*loss function*) $L$ compara un valor objetivo $y^{(i)}$ con la predicción correspondiente $\hat{y}^{(i)}$. La función de costo reúne las pérdidas de varios ejemplos y permite evaluar qué tan bien se ajusta el modelo a los datos de entrenamiento. El objetivo del entrenamiento es reducirla, sin olvidar que el modelo también debe funcionar con datos nuevos.

Una función de pérdida común en problemas de regresión es el error cuadrático medio, MSE (*Mean Squared Error*), que calcula el promedio de las diferencias al cuadrado entre los valores objetivo y las predicciones:


$$MSE = \frac{1}{m} \sum_{i=1}^{m} (\hat{y}^{(i)} - y^{(i)})^2$$

La minimización de la función de costo $J(\vec{w}, b)$ se realiza mediante algoritmos de optimización. Uno de los más conocidos es el **[descenso de gradiente](https://en.wikipedia.org/wiki/Gradient_descent)**.

Este algoritmo ajusta los parámetros del modelo, $\vec{w}$ y $b$, de manera iterativa para reducir la función de costo. Para ello calcula el gradiente con respecto a cada parámetro y actualiza los parámetros en la dirección opuesta. En el caso de un peso $w$ y un sesgo $b$ (*bias*), las actualizaciones se escriben como:

$$ w_{t+1} = w_t - \eta \frac{\partial J}{\partial w} $$

$$ b_{t+1} = b_t - \eta \frac{\partial J}{\partial b} $$

donde $\eta$ es la *tasa de aprendizaje*, que controla el tamaño de cada paso de actualización.

## 2.3. Modelo de Clasificación

La **clasificación** consiste en predecir a qué categoría pertenece una entrada. Estas categorías también se denominan clases o etiquetas. Un modelo de clasificación aproxima una función que relaciona las variables de entrada con una salida discreta.

Por ejemplo, al contestar estas preguntas estamos resolviendo problemas de clasificación:

- ¿Este correo electrónico es *spam* o no?
- ¿Este producto es un libro, una película o una prenda de vestir?
- ¿Esta película es una comedia romántica, un documental o una película de acción?

## 2.4. Modelo de Regresión

La **regresión** se refiere a problemas de modelado predictivo en los que se estima un valor numérico a partir de un conjunto de entradas. El modelo busca aproximar la relación entre las variables de entrada y la salida. Después puede utilizar esa relación para predecir el resultado de entradas nuevas.

Por ejemplo, al contestar estas preguntas estamos resolviendo problemas de regresión:

- ¿Cuál será la temperatura en Cancún mañana?
- Para este producto, ¿cuántas unidades se venderán?
- ¿A qué precio se venderá esta casa?

## 2.5. Redes Neuronales

Desde comienzos de la década de 2010, muchos avances importantes en inteligencia artificial, como el reconocimiento de voz, la traducción automática, la visión por computadora y la generación de contenido, se han apoyado en el *Deep Learning* o aprendizaje profundo.

El *Deep Learning* es una categoría dentro de las **Redes Neuronales Artificiales**. Estas redes tuvieron una inspiración histórica en algunas ideas sobre las neuronas biológicas, pero son modelos matemáticos muy simplificados y no reproducen el funcionamiento completo del cerebro.

Cada unidad, también llamada neurona artificial (o nodo), combina las entradas que recibe y aplica una función de activación para producir una salida. Las salidas pasan a la siguiente capa, que las utiliza como nuevas entradas. El proceso continúa hasta llegar a la capa final, que produce el resultado del modelo.

Si lo escribimos mediante funciones, obtenemos una serie de funciones anidadas, llamada [composición de funciones](https://es.wikipedia.org/wiki/Funci%C3%B3n_compuesta). Por ejemplo, una red neuronal de 3 capas que devuelve el vector $\vec{y}$ puede expresarse como:

$$\vec{y} = f_{NN}\left(\vec{x}\right)=f_3\left(f_2\left(f_1\left(\vec{x}\right)\right)\right)$$

Una red neuronal consta de unidades (neuronas) organizadas en capas que transforman un vector de entrada en una salida. Cada unidad combina sus entradas, aplica una función, normalmente no lineal, y pasa el resultado a la siguiente capa. En una red *[feed-forward](https://es.wikipedia.org/wiki/Red_neuronal_prealimentada)*, la información avanza hacia las capas posteriores y no regresa a las anteriores. Las conexiones tienen pesos que se ajustan durante el entrenamiento para adaptar la red al problema. Este proceso puede apreciarse en la siguiente imagen.


<div align="center"><img width="800px" style="display:block; margin-left: auto; margin-right: auto;" src="https://images.claudiazm.xyz/ecc26/MLFeedforwardNN.png"></div>

Las redes neuronales son una forma de hacer *Machine Learning*. Una computadora ajusta el modelo mediante el análisis de ejemplos de entrenamiento. En el aprendizaje supervisado, estos ejemplos pueden etiquetarse manualmente. Por ejemplo, un sistema de reconocimiento de objetos puede recibir miles de imágenes etiquetadas como automóviles, casas o tazas de café y buscar patrones visuales relacionados con cada etiqueta. Los sistemas modernos también pueden aprender a partir de datos sin etiquetas manuales mediante técnicas autosupervisadas, entre otros enfoques.

### Neurona: perceptrón

<div align="center"><img width="300px" style="display:block; margin-left: auto; margin-right: auto;" src="https://images.claudiazm.xyz/ecc26/MLPerceptron.png"></div>

Un **perceptrón** es uno de los modelos más sencillos de neurona artificial. Está inspirado de manera general en la idea de recibir varias señales y producir una salida.

El perceptrón recibe múltiples entradas que representan características de un dato. Estas entradas suelen denotarse como $x_1, x_2, \ldots, x_n$. A cada entrada $x_i$ se le asigna un peso $w_i$, que determina su contribución a la salida.

El perceptrón calcula una suma ponderada de sus entradas:

$$ z = \sum_{i=1}^{n} w_i x_i + b $$

donde $b$ es el sesgo (*bias*), un valor que permite desplazar el umbral de activación.

Los pesos $w_i$ y el sesgo $b$ son los **parámetros** del perceptrón. En una red neuronal existe un conjunto de parámetros para cada neurona.

La suma ponderada pasa por una *función de activación*. En las redes neuronales se utilizan funciones como la [sigmoide](https://en.wikipedia.org/wiki/Sigmoid_function), la [tangente hiperbólica](https://mathworld.wolfram.com/HyperbolicTangent.html) (*tanh*) o [ReLU](https://en.wikipedia.org/wiki/Rectified_linear_unit) (*Rectified Linear Unit*). El perceptrón clásico utiliza una [función escalón](https://en.wikipedia.org/wiki/Step_function):

$$ y = f(z) = \begin{cases}
1 & \text{si } z \geq 0 \\
0 & \text{si } z < 0
\end{cases} $$

El perceptrón clásico se entrena con una regla que modifica los pesos y el sesgo cuando comete un error de clasificación. En redes con funciones de activación diferenciables, los parámetros suelen ajustarse mediante descenso de gradiente y propagación hacia atrás (*backpropagation*).

### Ejemplo: ChatGPT

**ChatGPT** utiliza *Deep Learning*. Es un producto desarrollado por OpenAI que emplea modelos de la familia GPT (*Generative Pre-trained Transformer*), basados en la arquitectura **Transformer**.

La arquitectura [Transformer](https://arxiv.org/abs/1706.03762), presentada por Vaswani et al. en 2017, utiliza mecanismos de **atención** para relacionar elementos de una secuencia. Esta arquitectura resultó especialmente efectiva para el procesamiento del lenguaje natural y desplazó en gran medida a las redes recurrentes y convolucionales en este campo. Posteriormente se extendió a imágenes, audio y otros tipos de datos.

Los modelos GPT contienen múltiples capas de transformadores, lo que les permite aprender representaciones complejas del texto. Como referencia histórica, [GPT-3](https://arxiv.org/abs/2005.14165), presentado en 2020, tenía 175 mil millones de parámetros y 96 capas.

OpenAI no publicó el número de parámetros ni todos los detalles de la arquitectura de [GPT-4](https://cdn.openai.com/papers/gpt-4.pdf). Por ello, las cifras de 1.76 billones de parámetros y 120 capas que circularon en fuentes no oficiales se consideran **estimaciones no verificadas**. Lo mismo ocurre con la afirmación de que GPT-4 empleaba una arquitectura de [mezcla de expertos o *Mixture of Experts*](https://arxiv.org/abs/1701.06538); la técnica existe y está bien documentada, pero OpenAI no confirmó públicamente que esa fuera la arquitectura de GPT-4. Cabe destacar que el número de parámetros no determina por sí solo la calidad de un modelo, pero sí ofrece una buena estimación de su escala y capacidad computacional.

ChatGPT es un producto que cambia con el tiempo y puede integrar diferentes modelos y herramientas. En julio de 2026, OpenAI presentó oficialmente la familia [GPT-5.6](https://openai.com/es-ES/index/gpt-5-6/). OpenAI tampoco ha publicado una cantidad oficial de parámetros para esta familia.

Los modelos de lenguaje se preentrenan con grandes colecciones de texto para aprender regularidades y predecir continuaciones. Después pueden pasar por etapas adicionales de entrenamiento y evaluación, que incluyen ejemplos supervisados y retroalimentación humana. Este proceso permite generar texto coherente y útil, pero no garantiza que cada respuesta sea correcta; sus resultados deben verificarse cuando la precisión sea importante.

## 2.6. Ejemplo: Red Neuronal para Clasificación de Datos

La primera parte del código genera un conjunto aleatorio de puntos bidimensionales distribuido en $\texttt{K}$ ramas. Por ejemplo, cuando $\texttt{K=3}$, el conjunto de datos puede tener el siguiente aspecto:


<div align="center"><img width="400px" style="display:block; margin-left: auto; margin-right: auto;" src="https://images.claudiazm.xyz/ecc26/MLExampleNN.png"></div>

Para cada punto $\mathbf{x} = (x_1, x_2)$, la etiqueta indica a qué rama pertenece. En este ejemplo, $y$ puede tomar los valores $0$, $1$ o $2$.

Nuestro objetivo es implementar una red neuronal capaz de clasificar las ramas.

La salida de la red se comparará con las etiquetas mediante la llamada *[codificación one-hot](https://medium.com/@michaeldelsole/what-is-one-hot-encoding-and-how-to-do-it-f0ae272f1179)*. Para una etiqueta $\texttt{y=k}$, la codificación correspondiente es un vector de dimensión $\texttt{K}$ cuyas entradas son cero, excepto la entrada $\texttt{k}$, que vale $1$. Cuando $\texttt{K=3}$, las codificaciones son:

\begin{equation*}
0 \rightarrow \begin{bmatrix} 1 \\ 0 \\ 0 \end{bmatrix}, \qquad
1 \rightarrow \begin{bmatrix} 0 \\ 1 \\ 0 \end{bmatrix}, \qquad
2 \rightarrow \begin{bmatrix} 0 \\ 0 \\ 1 \end{bmatrix}.
\end{equation*}

El código define primero la estructura de la red neuronal y después utiliza el conjunto de datos para entrenarla.


```python
import matplotlib.pyplot as plt
import numpy as np

N = 50 # número de puntos por rama
K = 3  # número de ramas

N_train = N * K                              # número total de puntos en el conjunto de entrenamiento
x_train = np.zeros((N_train,2))              # matriz que contiene los puntos de datos bidimensionales
y_train = np.zeros(N_train, dtype='uint8')   # etiquetas (no en representación one-hot)

mag_noise = 0.2  # controla cuánto ruido se agrega a los datos
dTheta    = 4    # diferencia en theta en cada rama

### Generación de datos
for j in range(K):
  ix = range(N*j, N*(j+1))
  r  = np.linspace(0.01, 1, N) # radio
  th = np.linspace(j*(2*np.pi)/K, j*(2*np.pi)/K + dTheta, N) + np.random.randn(N)*mag_noise # theta
  x_train[ix] = np.c_[r*np.cos(th), r*np.sin(th)]
  y_train[ix] = j

### Graficar el conjunto de datos:
fig = plt.figure(1, figsize=(5,5))
plt.scatter(x_train[:, 0], x_train[:, 1], c=y_train, s=40)#, cmap=plt.cm.Spectral)
plt.xlim([-1, 1])
plt.ylim([-1, 1])
plt.xlabel(r'$x_1$')
plt.ylabel(r'$x_2$')
plt.show()
```

### Definir la arquitectura de red y los hiperparámetros de entrenamiento

```python
# El siguiente código necesita Torch:
#      pip install torch

%matplotlib inline
from IPython import display

import time
import torch

### Feedforward NN sin capa oculta:
class FeedforwardNN(torch.nn.Module):
    def __init__(self, input_size, output_size):
        super(FeedforwardNN, self).__init__()

        # tamaños de capa:
        self.input_size = input_size
        self.output_size = output_size

        # funciones utilizadas dentro del Feedforward NN:
        self.linear1 = torch.nn.Linear(self.input_size, self.output_size)
        self.relu    = torch.nn.ReLU() 
        self.sigmoid = torch.nn.Sigmoid()
        self.softmax = torch.nn.Softmax()      
    def forward(self, x):
        # capa 1:
        linear1_out = self.linear1(x)
        a1 = self.sigmoid(linear1_out)

        # Salida de la red:
        aL = a1

        return aL

### Feedforward NN con una capa oculta:
class FeedforwardNN_withHidden(torch.nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super(FeedforwardNN_withHidden, self).__init__()

        # tamaños de capa:
        self.input_size = input_size
        self.hidden_size  = hidden_size
        self.output_size = output_size

        # funciones utilizadas dentro del Feedforward NN:
        self.linear1 = torch.nn.Linear(self.input_size, self.hidden_size)
        self.linear2 = torch.nn.Linear(self.hidden_size, self.output_size)
        self.relu    = torch.nn.ReLU() 
        self.sigmoid = torch.nn.Sigmoid()
        self.softmax = torch.nn.Softmax()      
    def forward(self, x):
        # capa 1:
        linear1_out = self.linear1(x)
        a1 = self.sigmoid(linear1_out)

        # capa 2:
        linear2_out = self.linear2(a1) 
        a2 = self.softmax(linear2_out) 

        # Salida de la red:
        aL = a2

        return aL

input_size = 2
hidden_size = 8
output_size = K
model = FeedforwardNN_withHidden(input_size, hidden_size, output_size)

### Almacenar los datos de entrada como un tensor de PyTorch
x_train = torch.tensor(x_train, dtype = torch.float)

### Codificación one-hot
y_onehot = np.zeros((y_train.size, K))
y_onehot[np.arange(y_train.size), y_train] = 1
y_onehot = torch.tensor(y_onehot, dtype = torch.float)

### Utilizar la propagación hacia atrás para minimizar la función de costo utilizando el algoritmo de descenso de gradiente:
learning_rate = 3
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)     # SGD = stochastic gradient descent

### Función de costo:
cost_func = torch.nn.MSELoss()

N_epochs = 10000 # número de veces para ejecutar el descenso de gradiente
```

### Entrenamiento

```python
epoch_list    = []
cost_training = []
acc_training  = []

## Función para graficar:
def updatePlot():

    ### Generar coordenadas que cubran todo el plano:
    padding = 0.1
    spacing = 0.02
    
    x1_min, x1_max = x_train[:, 0].min().item() - padding, x_train[:, 0].max().item() + padding
    x2_min, x2_max = x_train[:, 1].min().item() - padding, x_train[:, 1].max().item() + padding
    x1_grid, x2_grid = np.meshgrid(np.arange(x1_min, x1_max, spacing), np.arange(x2_min, x2_max, spacing))

    torch_input = torch.tensor(np.c_[x1_grid.ravel(), x2_grid.ravel()], dtype = torch.float)
    NN_output = model(torch_input)
    predicted_class = np.argmax(NN_output.detach().numpy(), axis=1)

    ### Graficar el clasificador:
    plt.subplot(121)
    plt.contourf(x1_grid, x2_grid, predicted_class.reshape(x1_grid.shape), K, alpha=0.8)
    plt.scatter(x_train[:, 0], x_train[:, 1], c=y_train, s=40)
    plt.xlim(x1_grid.min(), x1_grid.max())
    plt.ylim(x2_grid.min(), x2_grid.max())
    plt.xlabel(r'$x_1$')
    plt.ylabel(r'$x_2$')

    ### Graficar la función de costo durante el entrenamiento:
    plt.subplot(222)
    plt.plot(epoch_list, cost_training, 'o-')
    plt.xlabel('')
    plt.ylabel('Costo de entrenamiento')

    ### Graficar la precisión del entrenamiento:
    plt.subplot(224)
    plt.plot(epoch_list, acc_training, 'o-')
    plt.xlabel('Epoca')
    plt.ylabel('Precisión de entrenamiento')
    

### Entrenar durante varias épocas: ###
for epoch in range(N_epochs):

    optimizer.zero_grad()                   # establecer los gradientes en cero (necesario ya que PyTorch acumule los gradientes)
    NN_output = model(x_train)              # Salida de la red neuronal
    cost = cost_func(NN_output, y_onehot)
    cost.backward()                         # calcula los gradientes
    optimizer.step()                        # actualizando los parametros

    ### Actualizar la gráfica e imprimir los resultados cada 500 épocas:
    if epoch % 500 == 0:
        predicted_class = np.argmax(NN_output.detach().numpy(), axis=1)
        accuracy = np.mean(predicted_class == y_train)
        
        epoch_list.append(epoch)
        cost_training.append(cost.detach().numpy())
        acc_training.append(accuracy)
        
        ### Actualizar la gráfica del clasificador resultante:
        plt.close(2)
        fig = plt.figure(2, figsize=(10,5))
        #fig.subplots_adjust(hspace=.3, wspace=.3)
        plt.clf()
        updatePlot()
        display.display(fig)
        print("Iteración %d:\n  Costo de entrenamiento %f\n  Precisión de entrenamiento %f\n" % (epoch, cost, accuracy) )
        display.clear_output(wait=True)
        time.sleep(0.2) # para reducir la velocidad de las actualizaciones de la gráfica

print("Costo final de entrenamiento %f\nPrecisión final de entrenamiento %f\n" % (cost, accuracy) )
```

La idea básica del **Quantum Machine Learning** (QML) es combinar conceptos de *Machine Learning* y *Computación Cuántica*:

- **Machine Learning:** aprender patrones a partir de datos de entrenamiento.
- **Computación Cuántica:** utilizar estados, operaciones y mediciones cuánticas para procesar información.

De acuerdo con la naturaleza de los datos y del procesamiento, podemos organizar el ML y el QML en las cuatro categorías de la siguiente figura:

También conviene distinguir entre dos etapas tecnológicas. Una computadora cuántica tolerante a fallas utiliza corrección de errores y combina varios qubits físicos para codificar qubits lógicos más confiables. Aunque ya existen demostraciones de qubits lógicos y corrección de errores, todavía no contamos con computadoras cuánticas tolerantes a fallas de gran escala capaces de ejecutar las aplicaciones completas que se estudian teóricamente.

Los procesadores disponibles actualmente suelen describirse como dispositivos de corto plazo o *near-term*. Están afectados por ruido, errores y un número limitado de operaciones confiables, pero permiten estudiar de manera experimental los componentes de los algoritmos. Entre los enfoques de QML investigados para estos dispositivos se encuentran los *kernels* cuánticos ([Havlíček et al.](https://arxiv.org/abs/1804.11326)), las redes generativas antagónicas cuánticas o *quantum GANs* ([Zoufal et al.](https://arxiv.org/abs/1904.00043)) y las máquinas cuánticas de Boltzmann ([Zoufal et al.](https://arxiv.org/abs/2006.06004)).

El objetivo del QML es combinar recursos clásicos y cuánticos para resolver ciertas tareas mejor que los métodos puramente clásicos. Esto es una meta de investigación, no una ventaja garantizada. Aun así, ya podemos implementar modelos pequeños, estudiar sus limitaciones y entender qué componentes podrían resultar útiles en hardware futuro.

## 3.2. Algoritmos Variacionales

Los **algoritmos variacionales** son métodos híbridos que combinan circuitos cuánticos parametrizados con optimización clásica. Se utilizan en problemas de optimización, simulación y machine learning. Sus circuitos pueden mantenerse relativamente cortos, por lo que son una de las familias más estudiadas para hardware de corto plazo, aunque el ruido todavía puede limitar sus resultados.

- La computadora cuántica prepara estados y realiza mediciones
- La computadora clásica calcula una función objetivo y actualiza los parámetros del circuito a partir de los resultados


Primero se define una función de costo relacionada con la tarea. Por ejemplo, en química cuántica puede representar la energía estimada de una molécula; en clasificación puede medir la diferencia entre las etiquetas y las predicciones del modelo. El objetivo es encontrar parámetros que minimicen o maximicen esa función, según el problema.


En cada iteración se ejecuta el circuito cuántico con los parámetros actuales y se estiman las cantidades necesarias mediante mediciones. La computadora clásica utiliza esos resultados para proponer nuevos parámetros. El ciclo se repite hasta cumplir un criterio de parada, como alcanzar un número máximo de iteraciones o dejar de observar una mejora significativa.


Entre los primeros artículos que propusieron **circuitos variacionales como modelos supervisados de QML** se encuentran [Farhi y Neven (2018)](https://arxiv.org/abs/1802.06002) y [Schuld et al. (2018)](https://arxiv.org/abs/1804.00633).


<div align="center"><img width="800px" style="display:block; margin-left: auto; margin-right: auto;" src="https://images.claudiazm.xyz/ecc26/QMLVariationalAlgorithms.png"></div>

### Tipos comunes de algoritmos variacionales

1. **Variational Quantum Eigensolver (VQE):**
   - Se utiliza principalmente para aproximar el estado de menor energía de un sistema cuántico, como ocurre en problemas de química cuántica.
   - Un circuito parametrizado prepara un estado y se mide su energía. Después, un optimizador clásico ajusta los parámetros para reducirla.

2. **Quantum Approximate Optimization Algorithm (QAOA):**
   - Está diseñado para construir aproximaciones a problemas de optimización combinatoria.
   - Alterna operadores relacionados con el problema y con la mezcla de soluciones. Un optimizador clásico ajusta sus parámetros.

3. **Quantum Neural Networks (QNNs):**
   - Utilizan circuitos cuánticos parametrizados como modelos entrenables.
   - Sus parámetros se ajustan para aprender patrones y realizar tareas como clasificación o regresión.


### Ventajas y desafíos

**Ventajas:**

- Permiten explorar modelos híbridos con circuitos relativamente cortos.
- Pueden utilizar superposición, interferencia y entrelazamiento para construir transformaciones difíciles de representar con circuitos clásicos sencillos.
- Ya pueden ejecutarse en simuladores y, a pequeña escala, en hardware cuántico actual.

**Desafíos:**

- No existe una ventaja cuántica general demostrada para todas estas aplicaciones.
- El ruido y el número finito de mediciones afectan las estimaciones.
- La optimización puede ser costosa y presentar gradientes muy pequeños, conocidos como *barren plateaus*.
- Un circuito más profundo o expresivo no necesariamente se entrena mejor.


<a id="3.2.1"></a>
### 3.2.1. Feature maps

Un *feature map* o mapa de características codifica una muestra clásica $\vec{x}$ en un estado cuántico. Si se parte de $n$ qubits en $\lvert 0\rangle$, se escribe

$$
\lvert\psi_{\mathrm{in}}(\vec{x})\rangle=U_F(\vec{x})\lvert 0\rangle^{\otimes n}
$$

La codificación forma parte del modelo, determina qué relaciones entre los datos serán fáciles o difíciles de representar.

| Codificación | Idea principal | Consideración importante |
|---|---|---|
| **Por ángulos o rotaciones** | Cada característica controla un ángulo de una compuerta como $R_x$, $R_y$ o $R_z$. | Es sencilla y suele usar aproximadamente un qubit por característica; los datos deben escalarse para controlar la periodicidad. |
| **Por amplitudes** | Las componentes de un vector normalizado se almacenan en las amplitudes del estado. | Puede representar $2^n$ componentes con $n$ qubits, pero preparar un estado arbitrario puede requerir un circuito costoso. |
| **En la base computacional** | Los datos binarios o discretos se representan como $\lvert b_1b_2\cdots b_n\rangle$. | Es fácil de preparar, pero los datos continuos deben discretizarse o binarizarse. |
| **Con entrelazamiento, como el mapa ZZ** | Operaciones de varios qubits introducen términos que dependen de pares de características. | Permite representar correlaciones desde la codificación, a cambio de mayor profundidad. |
| **Data re-uploading** | Los mismos datos se codifican varias veces, intercalados con bloques entrenables. | Aumenta la flexibilidad sin añadir qubits, pero aumenta la profundidad y el número de operaciones. |

### 3.2.2. Ansätze

El *ansatz* es la parte entrenable del circuito. Después de que el *feature map* prepara el estado de entrada, el ansatz aplica una secuencia de capas parametrizadas:

$$
U_A(\boldsymbol{\theta})=U_L(\boldsymbol{\theta}_L)\cdots U_2(\boldsymbol{\theta}_2)U_1(\boldsymbol{\theta}_1)
$$

Los parámetros $\boldsymbol{\theta}$ son los valores que modifica el optimizador clásico.

| Familia de ansatz | Estructura básica | Uso o ventaja principal |
|---|---|---|
| **Two-local** | Alterna rotaciones de un qubit y compuertas entre pares de qubits. | Es un punto de partida modular y fácil de modificar. |
| **Hardware-efficient** | Utiliza compuertas y conexiones cercanas a las nativas del dispositivo. | Reduce la profundidad adicional producida por la transpilación. |
| **Strongly entangling layers** | Emplea rotaciones generales y cambia el alcance del entrelazamiento entre capas. | Ofrece una familia flexible de interacciones locales y no locales. |
| **Estructurados o inspirados en el problema** | Su conectividad refleja una estructura física, jerárquica o matemática, como un árbol tensorial. | Puede incorporar conocimiento previo y evitar parámetros innecesarios. |

Un ansatz más expresivo puede representar más transformaciones, pero también puede ser más profundo, sensible al ruido o difícil de entrenar. Por ello se busca un equilibrio entre expresividad, número de parámetros, conectividad y entrenabilidad.

### 3.2.3. Funciones de pérdida

El circuito produce mediciones o valores esperados, pero estos todavía no constituyen una función de pérdida. Primero se aplica una regla clásica para convertir la lectura cuántica en una predicción $\hat y$. Para un lote $B$, el objetivo puede escribirse como

$$
\mathcal{L}_B(\boldsymbol{\theta})=
\frac{1}{\lvert B\rvert}\sum_{j\in B}
c\!\left(y^j,\hat y(\vec{x}^{\,j},\boldsymbol{\theta})\right)
$$

La función $c$ depende de la tarea:

- **Clasificación binaria:** entropía cruzada binaria o pérdida *hinge*.
- **Clasificación multiclase:** entropía cruzada sobre un vector de probabilidades.
- **Regresión:** error cuadrático medio, error absoluto medio o pérdida de Huber.
- **Aprendizaje de distribuciones:** divergencias como Kullback-Leibler.
- **Simulación física:** puede optimizarse directamente un valor esperado, como la energía en VQE.

No existe una pérdida universalmente mejor. Debe ser compatible con el significado de la salida y con la tarea que se desea resolver.

### 3.2.4. Gradientes y optimizadores

El optimizador clásico propone nuevos valores de $\boldsymbol{\theta}$. Hay dos rutas principales:

- **Métodos basados en gradientes:** descenso de gradiente, Adam y métodos cuasi-Newton como L-BFGS-B. El gradiente puede obtenerse mediante diferencias finitas, diferenciación en simuladores o reglas cuánticas como *parameter shift*.
- **Métodos sin gradientes explícitos:** COBYLA, Nelder-Mead y otros métodos que comparan valores de la función objetivo. SPSA estima una dirección de cambio con pocas evaluaciones y suele considerarse especialmente cuando hay ruido de muestreo.

Para muchas rotaciones generadas por matrices de Pauli, la derivada de un valor esperado $f$ puede calcularse exactamente con la regla *parameter shift*:

$$
\frac{\partial f}{\partial\theta_m}
=\frac{f(\theta_m+\pi/2)-f(\theta_m-\pi/2)}{2}
$$

Esta regla evita elegir un paso numérico, pero requiere ejecutar el circuito con parámetros desplazados. En hardware, cada evaluación se estima con un número finito de *shots*, por lo que el gradiente también contiene ruido. La elección del optimizador depende del número de parámetros, la estabilidad del gradiente y el presupuesto de ejecuciones.

<p style="text-align:center;"><small>El optimizador siempre pertenece al lado clásico del ciclo; lo que cambia es la información que utiliza para actualizar los parámetros.</small></p>

## 3.3. Ejemplo sencillo de Algoritmo Variacional


Construiremos un algoritmo variacional sencillo capaz de encontrar un parámetro que define el estado de un qubit.

Primero buscaremos el ángulo de una rotación alrededor del eje $Y$ que transforma el estado inicial $\lvert 0\rangle$ en $\lvert 1\rangle$.


<div align="center"><img width="300px" style="display:block; margin-left: auto; margin-right: auto;" src="https://images.claudiazm.xyz/ecc26/QMLExampleVarAlgo.png"></div>

```python
from qiskit import QuantumCircuit
import qiskit.quantum_info as qi
from qiskit.visualization import plot_bloch_multivector

import numpy as np
from math import radians, degrees
from scipy.optimize import minimize


def costo_edo_1(proba_edo_1):
    return 1 - proba_edo_1                      # diferencia de probabilidad de que sea 1


def fn_objetivo(angle, fn_costo, shots=2000):

    qc = QuantumCircuit(1)

    qc.ry(radians(angle[0]), 0)                 # rotación en Y por una cantidad de 'angle' grados

    sv = qi.Statevector(qc)
    counts = sv.sample_counts(shots=shots)
    
    proba_edo_1 = (counts['1'] if '1' in counts else 0)/shots    # probabilidad de que sea el estado 1

    cost = fn_costo(proba_edo_1)

    print('Intento: ángulo =', angle, 'produce un costo promedio =', cost)

    return cost
```

Definimos un valor inicial para el parámetro que buscamos.


```python
initial_guess = [90]
```

Utilizamos la función `minimize` de SciPy para encontrar el valor del ángulo


```python
result = minimize(fn_objetivo, initial_guess, args=(costo_edo_1,), method='Powell')
```

```python
print('\nÁngulo optimizado en grados:', [a % 360 for a in result.x])
```

Mostramos en la esfera de Bloch el estado obtenido con el parámetro encontrado:


```python
qc = QuantumCircuit(1)
qc.ry(radians(result.x[0]), 0)

sv = qi.Statevector(qc)
plot_bloch_multivector(sv)
```

Ahora encontremos el ángulo necesario para terminar en el estado $\lvert -i\rangle$

$$
\lvert -i\rangle=\frac{\lvert 0\rangle-i\lvert 1\rangle}{\sqrt{2}}
$$

Este estado tiene las mismas probabilidades de medición en la base computacional que $\lvert +\rangle$, $\lvert -\rangle$ y $\lvert +i\rangle$. Por ello, no basta con pedir una probabilidad de $1/2$ para el resultado `1`. En este caso utilizaremos la fidelidad con el estado objetivo.


```python
def fn_objetivo_menos_i(angle):
    qc = QuantumCircuit(1)
    qc.rx(radians(angle[0]), 0)

    estado = qi.Statevector(qc)
    estado_objetivo = qi.Statevector([1 / np.sqrt(2), -1j / np.sqrt(2)])
    costo = 1 - qi.state_fidelity(estado, estado_objetivo)

    print('Intento: ángulo =', angle, 'produce un costo =', costo)
    return costo


initial_guess = [0]
result_menos_i = minimize(fn_objetivo_menos_i, initial_guess, method='Powell')

print('\nÁngulo optimizado en grados:', [a % 360 for a in result_menos_i.x])
```

```python
qc = QuantumCircuit(1)
qc.rx(radians(result_menos_i.x[0]), 0)

sv = qi.Statevector(qc)
plot_bloch_multivector(sv)
```

## 4.1. Ejemplo: Regresión


Prepararemos un conjunto pequeño de datos de regresión para aproximar una función seno con ruido. Conservaremos el ejemplo de un solo qubit para que sea posible observar con claridad el papel del *feature map*, el ansatz y el regresor.


```python
num_samples = 20
eps = 0.2
lb, ub = -np.pi, np.pi

rng = np.random.default_rng(4567890)

X_ = np.linspace(lb, ub, num=50).reshape(50, 1)


def f(x):
    return np.sin(x)


X = (ub - lb) * rng.random((num_samples, 1)) + lb
y = f(X[:, 0]) + eps * (2 * rng.random(num_samples) - 1)

plt.plot(X_, f(X_), "r--", label="Función seno")
plt.plot(X, y, "bo", label="Datos con ruido")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

### Entrenamiento con QML

Entrenaremos un regresor cuántico variacional (VQR). El VQR es el regresor más simple disponible en Qiskit Machine Learning y es un buen punto de partida para los recién llegados al QML que tienen experiencia en ML clásico.

Pero antes de entrenar un modelo, examinemos qué comprende la clase `VQR`. Dos de sus elementos centrales son el mapa de características y el ansatz.

Nuestros datos son clásicos, lo que significa que consisten en un conjunto de bits, no qubits. Necesitamos una forma de codificar los datos como qubits. Este proceso es crucial si queremos obtener un modelo cuántico eficaz. Por lo general, nos referimos a este mapeo como codificación de datos, incrustación de datos o carga de datos y esta es la función del **mapa de características** (feature map).

Una vez que se cargan los datos, debemos aplicar inmediatamente un circuito cuántico parametrizado. Este circuito es un análogo directo de las capas en las redes neuronales clásicas. Tiene un conjunto de parámetros o pesos ajustables. Los pesos se optimizan de modo que minimicen una función objetivo. Esta función objetivo caracteriza la distancia entre las predicciones y los datos etiquetados conocidos. Un circuito cuántico parametrizado (Variational Quantum Circuit, VQC) también se denomina estado de prueba parametrizado, **forma variacional** o **ansatz**. Quizás, este último sea el término más utilizado.

<div align="center"><img width="550px" style="display:block; margin-left: auto; margin-right: auto;" src="https://images.claudiazm.xyz/ecc26/QMLqnnGeneral.png"></div>

### Regresión con un `EstimatorQNN`


Un `EstimatorQNN` devuelve valores esperados. Con el observable predeterminado de un qubit, su salida se encuentra en $[-1,1]$, un intervalo adecuado para aproximar la función seno. También es posible construir modelos multidimensionales o utilizar un `SamplerQNN` cuando la salida se define mediante probabilidades.


```python
# Construir un feature map sencillo
param_x = Parameter("x")
feature_map_reg = QuantumCircuit(1, name="fm")
feature_map_reg.ry(param_x, 0)

# Construir un ansatz sencillo
param_y = Parameter("y")
ansatz_reg = QuantumCircuit(1, name="vf")
ansatz_reg.ry(param_y, 0)

# Componer el circuito completo
regression_circuit = QuantumCircuit(1)
regression_circuit.compose(feature_map_reg, inplace=True)
regression_circuit.compose(ansatz_reg, inplace=True)

# QMLEstimator con precisión cero realiza una simulación exacta
qml_estimator = QMLEstimator(default_precision=0)

# Definir explícitamente el gradiente evita que EstimatorQNN tenga que crearlo
# de manera automática (en otro caso muestra un mensaje informativo)
parameter_shift_gradient = ParamShiftEstimatorGradient(qml_estimator)

regression_estimator_qnn = EstimatorQNN(
    circuit=regression_circuit,
    estimator=qml_estimator,
    gradient=parameter_shift_gradient,
    input_params=feature_map_reg.parameters,
    weight_params=ansatz_reg.parameters,
)

regression_circuit.draw("mpl")
```

```python
objective_func_vals = []


def callback_graph(weights, objective_value):
    clear_output(wait=True)
    objective_func_vals.append(objective_value)
    fig, ax = plt.subplots(figsize=(7, 4))
    ax.plot(objective_func_vals)
    ax.set_title("Función objetivo durante el entrenamiento")
    ax.set_xlabel("Evaluación")
    ax.set_ylabel("Valor de la función objetivo")
    plt.show()
    plt.close(fig)


regressor = NeuralNetworkRegressor(
    neural_network=regression_estimator_qnn,
    loss="squared_error",
    optimizer=SLSQP(maxiter=20),
    initial_point=np.array([-1.0]),
    callback=callback_graph,
)
```

```python
objective_func_vals = []

regressor.fit(X, y)

regression_score = regressor.score(X, y)
print(f"Coeficiente R² del EstimatorQNN: {regression_score:.3f}")
```

```python
plt.plot(X_, f(X_), "r--", label="Función seno")
plt.plot(X, y, "bo", label="Datos con ruido")

y_estimator = regressor.predict(X_)
plt.plot(X_, y_estimator, "g-", label="EstimatorQNN")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
regressor.weights  # peso aprendido por el modelo
```

### Regresor Cuántico Variacional (Variational Quantum Regresor) `VQR`


De forma análoga a `VQC` para clasificación, `VQR` es una variante de alto nivel de `NeuralNetworkRegressor`. Construye internamente un `EstimatorQNN` a partir del *feature map* y el ansatz. De manera predeterminada utiliza una pérdida cuadrática para comparar las predicciones con los valores objetivo.


```python
vqr = VQR(
    feature_map=feature_map_reg,
    ansatz=ansatz_reg,
    estimator=qml_estimator,
    pass_manager=PassManager(),
    optimizer=SLSQP(maxiter=20),
    initial_point=np.array([-1.0]),
    callback=callback_graph,
)
```

```python
objective_func_vals = []

vqr.fit(X, y)

vqr_score = vqr.score(X, y)
print(f"Coeficiente R² del VQR: {vqr_score:.3f}")
```

```python
plt.plot(X_, f(X_), "r--", label="Función seno")
plt.plot(X, y, "bo", label="Datos con ruido")

y_vqr = vqr.predict(X_)
plt.plot(X_, y_vqr, "g-", label="VQR")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

## 4.2. Ejemplo: Clasificación de señales circulares


Consideremos un sensor que registra dos características, $x_1$ y $x_2$, de cada señal. Las observaciones pertenecen a dos clases que forman regiones aproximadamente circulares: una clase se concentra cerca del centro y la otra forma un anillo exterior.

Utilizaremos un conjunto de datos sintético para que la geometría del problema sea visible y podamos concentrarnos en el funcionamiento del clasificador. Una sola línea recta no puede separar las dos clases, por lo que necesitamos una frontera de decisión no lineal.


```python
from sklearn.datasets import make_circles
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler


# Generar dos clases con forma de círculos concéntricos
circle_features, circle_labels = make_circles(
    n_samples=100,
    noise=0.05,
    factor=0.35,
    random_state=42,
)

# Separar los datos antes de ajustar el escalamiento
train_features_raw, test_features_raw, train_labels, test_labels = train_test_split(
    circle_features,
    circle_labels,
    test_size=0.25,
    random_state=42,
    stratify=circle_labels,
)

# Los ángulos del feature map se mantendrán entre 0 y pi/2
scaler = MinMaxScaler(feature_range=(0, np.pi / 2))
train_features = scaler.fit_transform(train_features_raw)
test_features = scaler.transform(test_features_raw)

fig, ax = plt.subplots(figsize=(6, 5))
ax.scatter(
    train_features_raw[:, 0],
    train_features_raw[:, 1],
    c=train_labels,
    cmap="coolwarm",
    edgecolors="black",
    label="Entrenamiento",
)
ax.scatter(
    test_features_raw[:, 0],
    test_features_raw[:, 1],
    c=test_labels,
    cmap="coolwarm",
    marker="X",
    s=90,
    edgecolors="black",
    label="Prueba",
)
ax.set_xlabel("Característica $x_1$")
ax.set_ylabel("Característica $x_2$")
ax.set_title("Datos para clasificación")
ax.legend()
plt.show()
```

El conjunto contiene 100 muestras equilibradas. Reservamos el 25 % para evaluar el modelo con datos que no se utilizaron durante el entrenamiento. El escalador se ajusta únicamente con el conjunto de entrenamiento para evitar que la información del conjunto de prueba influya en el modelo.

El intervalo $[0,\pi/2]$ se eligió para convertir las dos características en ángulos sin recorrer innecesariamente varios periodos de las compuertas. Como vimos en la sección de *feature maps*, el escalamiento forma parte del diseño del modelo.


### Clasificación con un kernel cuántico

Un *feature map* prepara un estado $\lvert\phi(\vec{x})\rangle$ para cada muestra. El kernel de fidelidad compara dos muestras mediante el traslape de sus estados:

$$
K(\vec{x},\vec{z})=
\left\lvert\left\langle\phi(\vec{x})\middle|\phi(\vec{z})\right\rangle\right\rvert^2
$$

Si dos estados codificados son parecidos, el valor del kernel se acerca a $1$, si son muy diferentes, se acerca a $0$. `QSVC` (Quantum Support Vector Classifier) utiliza esta matriz de similitudes dentro de un clasificador de vectores de soporte.


```python
num_features = train_features.shape[1]

feature_map = zz_feature_map(
    feature_dimension=num_features,
    reps=1,
    entanglement="full",
)

feature_map.decompose().draw(output="mpl", style="clifford", fold=20)
```

El circuito utiliza dos qubits, uno por característica. Las compuertas de fase codifican los valores de entrada y las operaciones entre qubits permiten que el estado dependa de relaciones entre ambas características.

Para mantener el ejemplo rápido y determinista utilizaremos `FidelityStatevectorKernel`, que calcula las fidelidades mediante simulación exacta de vectores de estado. Esta elección no añade ruido de muestreo y no representa una ejecución en hardware real.


```python
quantum_kernel = FidelityStatevectorKernel(feature_map=feature_map)

train_kernel_matrix = quantum_kernel.evaluate(x_vec=train_features)

fig, ax = plt.subplots(figsize=(6, 5))
image = ax.imshow(train_kernel_matrix, cmap="viridis", vmin=0, vmax=1)
ax.set_title("Matriz del kernel cuántico")
ax.set_xlabel("Muestra de entrenamiento")
ax.set_ylabel("Muestra de entrenamiento")
fig.colorbar(image, ax=ax, label="Fidelidad")
plt.show()
```

Ahora entregamos el kernel cuántico a `QSVC`. La interfaz `fit`, `predict` y `score` es la misma que se utiliza habitualmente en `scikit-learn`.


```python
qsvc = QSVC(quantum_kernel=quantum_kernel)

start = time.time()
qsvc.fit(train_features, train_labels)
elapsed = time.time() - start

train_score = qsvc.score(train_features, train_labels)
test_score = qsvc.score(test_features, test_labels)

print(f"Exactitud en entrenamiento: {train_score:.2f}")
print(f"Exactitud en prueba:        {test_score:.2f}")
print(f"Tiempo de entrenamiento:    {elapsed:.2f} segundos")
```

Con la semilla, la partición y las versiones indicadas, el modelo obtiene una exactitud de 1.00 en entrenamiento y de 1.00 en prueba. Este resultado es adecuado para comprobar que el flujo completo funciona, pero debe interpretarse con cuidado: el conjunto fue diseñado para ser pequeño y visual, y una sola partición de 25 muestras de prueba no constituye una evaluación exhaustiva.

Este resultado **no demuestra una ventaja cuántica**. Clasificadores clásicos también pueden resolver fácilmente este conjunto. El objetivo aquí es observar cómo un circuito parametrizado por datos produce un kernel que puede utilizarse en una tarea real de clasificación.


```python
# Evaluar la frontera de decisión en una cuadrícula del espacio original
padding = 0.15
x1_min, x1_max = circle_features[:, 0].min() - padding, circle_features[:, 0].max() + padding
x2_min, x2_max = circle_features[:, 1].min() - padding, circle_features[:, 1].max() + padding

x1_grid, x2_grid = np.meshgrid(
    np.linspace(x1_min, x1_max, 80),
    np.linspace(x2_min, x2_max, 80),
)
grid_raw = np.c_[x1_grid.ravel(), x2_grid.ravel()]
grid_scaled = scaler.transform(grid_raw)
grid_predictions = qsvc.predict(grid_scaled).reshape(x1_grid.shape)

fig, ax = plt.subplots(figsize=(6, 5))
ax.contourf(
    x1_grid,
    x2_grid,
    grid_predictions,
    levels=[-0.5, 0.5, 1.5],
    cmap="coolwarm",
    alpha=0.25,
)
ax.scatter(
    train_features_raw[:, 0],
    train_features_raw[:, 1],
    c=train_labels,
    cmap="coolwarm",
    edgecolors="black",
    label="Entrenamiento",
)
ax.scatter(
    test_features_raw[:, 0],
    test_features_raw[:, 1],
    c=test_labels,
    cmap="coolwarm",
    marker="X",
    s=90,
    edgecolors="black",
    label="Prueba",
)
ax.set_xlabel("Característica $x_1$")
ax.set_ylabel("Característica $x_2$")
ax.set_title("Frontera aprendida por QSVC")
ax.legend()
plt.show()
```

### Otras opciones dentro de Qiskit Machine Learning

También podríamos resolver una clasificación con `VQC`. A diferencia de `QSVC`, un `VQC` combina un *feature map*, un ansatz y un optimizador clásico. Su resultado puede cambiar al modificar:

- La estructura y el número de repeticiones del *feature map* y del ansatz. Más repeticiones añaden flexibilidad, pero también profundidad, parámetros y riesgo de sobreajuste.
- El optimizador. Algunos utilizan gradientes y otros, como COBYLA, solo valores de la función objetivo.
- El punto inicial de los parámetros. Conviene comparar varias inicializaciones de forma controlada.
- El número de características. Algunos mapas sencillos utilizan un qubit por característica, por lo que puede ser necesario seleccionar o reducir características antes de entrenar.

- [**Undergraduate School on Experimental Quantum Information Processing**](https://uwaterloo.ca/institute-for-quantum-computing/outreach/useqip) (USEQIP): es un programa presencial de dos semanas en la universidad de Waterloo (Canadá), sobre el estudio teórico y experimental de la información cuántica dirigido principalmente a estudiantes que están a un año de completar sus estudios de licenciatura.
