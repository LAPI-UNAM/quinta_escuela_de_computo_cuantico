# 1. ¿Qué es un Algoritmo?

Un **algoritmo** es una secuencia finita y bien definida de instrucciones para resolver un problema específico o realizar una tarea computacional. Los algoritmos son fundamentales en todos los campos de la computación y pueden ejecutarse tanto en computadoras clásicas como cuánticas.

En la **computación clásica**, los algoritmos manipulan bits ($0$ o $1$) y ejecutan instrucciones deterministas o probabilísticas mediante compuertas lógicas. Estos algoritmos se ejecutan en dispositivos digitales convencionales, como las CPU y las GPU.

En la **computación cuántica**, los algoritmos operan sobre **qubits**, cuyos estados pueden ser superposiciones de $0$ y $1$. Además, los sistemas de varios qubits pueden presentar **entrelazamiento**, y sus amplitudes pueden combinarse mediante **interferencia**. Un **algoritmo cuántico** es un procedimiento computacional paso a paso en el que cada operación puede realizarse en una computadora cuántica, generalmente modelada mediante circuitos cuánticos.

Las computadoras cuánticas también pueden ejecutar cálculos clásicos. Esto es posible porque cualquier cálculo clásico puede reescribirse como un circuito reversible, y los circuitos cuánticos pueden implementar compuertas reversibles universales [[Lanzagorta, M., 2009](https://books.google.com/books?id=-wkJIuw0YRsC&q=quantum%2520computer%2520equivalent%2520classical%2520computer&pg=PA23)]. Sin embargo, nos interesan especialmente los algoritmos que aprovechan propiedades cuánticas para obtener ventajas computacionales sobre los mejores métodos clásicos conocidos.

<a id="1.1"></a>
## 1.1. ¿Qué hace que un algoritmo cuántico sea diferente?

Los algoritmos cuánticos se distinguen por explotar fenómenos de la mecánica cuántica, como:

* **Superposición**: Permite describir el estado de un qubit como una combinación de los estados de una base dada
* **Entrelazamiento**: Produce estados de varios qubits que no pueden describirse asignando un estado independiente a cada qubit
* **Interferencia**: Permite que las amplitudes de distintas rutas computacionales se refuercen o se cancelen

Al aprovechar estas características, algunos algoritmos cuánticos ofrecen ventajas cuadráticas o exponenciales frente a los mejores algoritmos clásicos conocidos para problemas específicos.

Sin embargo, la computación cuántica no es universalmente mejor para todos los problemas. El objetivo es identificar problemas específicos donde los algoritmos cuánticos ofrecen **ventaja cuántica**.

## 1.2. Ventaja cuántica

La **ventaja cuántica** se alcanza cuando un dispositivo cuántico supera a los mejores métodos clásicos conocidos en una tarea computacional bien definida y de acuerdo con una métrica concreta, como el tiempo de ejecución o la cantidad de recursos empleados. En los primeros trabajos también se utilizó el término *supremacía cuántica*, pero actualmente suele preferirse *ventaja cuántica*.

En octubre de 2019, Google afirmó haber demostrado ventaja cuántica con su procesador *Sycamore* mediante el muestreo aleatorio de circuitos. El experimento cuántico tardó alrededor de 200 segundos y Google estimó que una supercomputadora clásica requeriría aproximadamente 10 000 años [[Arute et al., 2019](https://www.nature.com/articles/s41586-019-1666-5)]. IBM cuestionó esa comparación y mostró que una simulación clásica más eficiente podría completar la tarea en unos 2.5 días, aunque con recursos extraordinarios [[Pednault et al., 2019](https://arxiv.org/abs/1910.09534)]. Este caso ilustra por qué toda afirmación de ventaja depende del mejor método clásico y de la métrica utilizada.

Lograr una ventaja cuántica generalmente requiere:

* Un problema computacional bien definido
* Un algoritmo cuántico capaz de resolverlo
* Un punto de referencia clásico para comparar el rendimiento
* Un argumento teórico (a menudo basado en la teoría de la complejidad) que sostiene que es improbable que los algoritmos clásicos igualen el rendimiento cuántico

Entre los algoritmos con ventajas teóricas y las tareas utilizadas en demostraciones experimentales se encuentran:

* El [Algoritmo de Shor](https://quantum.cloud.ibm.com/learning/en/courses/fundamentals-of-quantum-algorithms/phase-estimation-and-factoring/shor-algorithm) para la factorización de enteros. Su ventaja es algorítmica y todavía requiere una computadora tolerante a fallos para ser útil a gran escala.
* El [Muestreo de Bosones](https://doi.org/10.1145/1993636.1993682), una tarea de muestreo con fotones no interactuantes.
* El [Muestreo Aleatorio de Circuitos Cuánticos](https://www.nature.com/articles/s41586-019-1666-5), utilizado por Google en 2019.

Una parte importante del hardware actual pertenece a la era **[NISQ](https://doi.org/10.22331/q-2018-08-06-79)** (*Noisy Intermediate-Scale Quantum*): dispositivos con una cantidad limitada de qubits y operaciones propensas a errores. Estos equipos ya permiten estudiar tareas especializadas y realizar demostraciones experimentales, pero no implican una ventaja general para cualquier problema.

## 1.4. Ejemplos de algoritmos cuánticos

No existe una clasificación única y universal de los algoritmos cuánticos. Además, un mismo algoritmo puede combinar varias técnicas. Una organización práctica consiste en agruparlos por el modelo de acceso al problema o por su subrutina cuántica principal.

Esta organización es consistente con la revisión académica [*Quantum algorithms: an overview*](https://www.nature.com/articles/npjqi201523) y con el curso [*Fundamentals of Quantum Algorithms* de IBM Quantum](https://quantum.cloud.ibm.com/learning/en/courses/fundamentals-of-quantum-algorithms):

* **Algoritmos de consulta a oráculos, es decir, cajas negras, y búsqueda de estructura oculta**

  * [Algoritmo de Deutsch-Jozsa](https://quantum.cloud.ibm.com/learning/en/courses/fundamentals-of-quantum-algorithms/quantum-query-algorithms/deutsch-jozsa-algorithm): determina con una consulta si una función es constante o balanceada, bajo la promesa de que cumple uno de estos dos casos. Un algoritmo clásico determinista exacto puede requerir un número exponencial de consultas.
  * [Algoritmo de Bernstein-Vazirani](https://quantum.cloud.ibm.com/learning/en/courses/fundamentals-of-quantum-algorithms/quantum-query-algorithms/deutsch-jozsa-algorithm): encuentra una cadena oculta con una sola consulta al oráculo.
  * [Algoritmo de Simon](https://quantum.cloud.ibm.com/learning/en/courses/fundamentals-of-quantum-algorithms/quantum-query-algorithms/simon-algorithm): encuentra una periodicidad oculta y ofrece una ventaja exponencial en el número de consultas frente a algoritmos clásicos probabilísticos.

* **Algoritmos basados en la [Transformada Cuántica de Fourier (QFT) y la estimación de fase (QPE)](https://quantum.cloud.ibm.com/learning/en/courses/fundamentals-of-quantum-algorithms/phase-estimation-and-factoring/introduction)**

  * [Estimación de Fase Cuántica](https://quantum.cloud.ibm.com/learning/courses/fundamentals-of-quantum-algorithms/phase-estimation-and-factoring/phase-estimation-procedure) (QPE): estima la fase asociada a un valor propio de un operador unitario y es una subrutina clave de numerosos algoritmos.
  * [Algoritmo de Shor](https://quantum.cloud.ibm.com/learning/en/courses/fundamentals-of-quantum-algorithms/phase-estimation-and-factoring/shor-algorithm): resuelve la factorización de enteros en tiempo polinomial; no se conoce un algoritmo clásico de tiempo polinomial para este problema.

* **Algoritmos basados en la [amplificación y estimación de amplitud](https://quantum.cloud.ibm.com/docs/en/tutorials/grovers-algorithm)**

  * [Algoritmo de Grover](https://quantum.cloud.ibm.com/docs/en/tutorials/grovers-algorithm): realiza búsqueda no estructurada con una ventaja cuadrática en el número de consultas.
  * [Algoritmo de Conteo Cuántico](https://arxiv.org/abs/quant-ph/9805082): estima el número de soluciones de un problema de búsqueda.

* **Algoritmos basados en caminatas cuánticas**

  * [Problema de distinción de elementos](https://doi.org/10.1137/S0097539705447311): determina si una lista contiene elementos repetidos.
  * [Problema de detección de triángulos](https://doi.org/10.1137/050643684): determina si un grafo contiene un triángulo.

* **[Algoritmos variacionales híbridos cuántico-clásicos](https://quantum.cloud.ibm.com/learning/en/courses/utility-scale-quantum-computing/variational-quantum-algorithms)** (importantes para dispositivos NISQ)

  * Algoritmo Cuántico de Optimización Aproximada (QAOA): se aplica a problemas de optimización combinatoria mediante circuitos parametrizados y optimización clásica.
  * [Solucionador Variacional de Eigenvalores Cuánticos](https://quantum.cloud.ibm.com/learning/en/courses/quantum-diagonalization-algorithms/vqe) (VQE): aproxima valores propios, también llamados eigenvalores, y estados fundamentales, con aplicaciones en química y ciencia de materiales.

En esta sesión, primero exploraremos cómo los circuitos cuánticos pueden implementar **operaciones lógicas clásicas**, comenzando con una versión cuántica del **Sumador Completo**.

## 1.5. ¿Por qué nos interesan las computadoras cuánticas?

Incluso las mejores computadoras clásicas tienen dificultades para resolver, con tiempo y recursos prácticos, algunos problemas científicos y computacionales importantes.

Nos interesa la computación cuántica para:

* Simulación de la naturaleza.  
* Búsqueda y optimización.  
* Procesamiento de datos con estructura (p. ej., *machine learning* cuántico).

**Las computadoras cuánticas están pensadas para trabajar junto con las computadoras clásicas, no para reemplazarlas.**

## 2.1. Sumador completo cuántico

<div align="center"><img width="900px" style="display:block; margin-left: auto; margin-right: auto;" src="https://images.claudiazm.xyz/ecc26/ClassicalVsReversibleAdder.gif"></div>

Un sumador completo cuántico muestra cómo las computadoras cuánticas pueden replicar algoritmos clásicos respetando principios cuánticos como la reversibilidad.

Las computadoras cuánticas pueden implementar lógica clásica dentro de circuitos cuánticos. La diferencia clave es que, mientras que las compuertas clásicas como AND u OR suelen ser irreversibles, las operaciones de un circuito cuántico deben ser reversibles. Esto nos obliga a rediseñar la lógica clásica como lógica reversible, que después puede traducirse a compuertas cuánticas.

Al igual que en la electrónica clásica, donde diseñamos [sumadores binarios](https://en.wikipedia.org/wiki/Adder_%28electronics%29) (sumadores parciales, sumadores completos, sumadores con acarreo en cascada, etc.), los circuitos cuánticos también pueden implementar estas operaciones.

Un **sumador completo** clásico toma tres entradas: dos bits a sumar (`A` y `B`) y un bit de acarreo de entrada (`Cin`). Produce dos salidas: el bit de suma (`S`) y el bit de acarreo de salida (`Cout`).

La lógica clásica se describe mediante:


<div align="center"><img width="800px" style="display:block; margin-left: auto; margin-right: auto;" src="https://images.claudiazm.xyz/ecc26/fullAdderClass.png"></div>

<p style="font-size:10px;font-style:italic;text-align:right"><a href="https://microcontrollerslab.com/cd4008-4-bit-full-adder-ic-pinout-features-example-datasheet/" target="_blank">Imagen obtenida de MicrocontrollersLab</a></p>

Las operaciones de un circuito cuántico deben representarse mediante transformaciones unitarias y, por tanto, reversibles (por lo que deben tener el mismo número de entradas y salidas). Por ello, diseñamos una versión de 4 qubits en la que:

* Entradas: `A`, `B`, `CarryIn` y `aux` (ancilla)
* Salidas: `A`, `B`, `Sum` y `CarryOut`

Esto nos permite implementar lógica clásica de forma reversible utilizando **compuertas CNOT** y **Toffoli (CCX)**.


Una posible implementación en Qiskit se ve así:

````python
from qiskit import QuantumRegister, ClassicalRegister, QuantumCircuit

A = QuantumRegister(1, 'A')
B = QuantumRegister(1, 'B')
CarryIn = QuantumRegister(1, 'CarryIn-Sum')
cero = QuantumRegister(1, 'aux-CarryOut')
adder = QuantumCircuit(A, B, CarryIn, cero)

adder.ccx(0, 1, 3)
adder.cx(0, 1)
adder.ccx(1, 2, 3)
adder.cx(1, 2)
adder.cx(0, 1)

adder.draw('mpl')
````

Notemos que este circuito es completamente reversible:

* Todas las compuertas utilizadas son reversibles (CNOT y Toffoli).
* No se pierde información, pues todas las entradas pueden recuperarse de las salidas si es necesario.

Esta reversibilidad es esencial para la computación cuántica, ya que la mecánica cuántica no permite que se destruya información durante el cálculo.


Ahora podemos inicializar y probar este sumador cuántico completo con entradas específicas.

### Ejemplo 1: Suma `A=1`, `B=0`, `CarryIn=1`

Preparamos el estado inicial:


````python
initialization = QuantumCircuit(4)

initialization.x(0)       # A       = 1
initialization.x(2)       # CarryIn = 1
initialization.barrier()

initialization.draw('mpl')
````

Compongamos el circuito de inicialización con el sumador, midamos y ejecutemos la simulación:


````python
sum = initialization.compose(adder)

sum.barrier()

sum.add_register(ClassicalRegister(2, 'Resultado'))

sum.measure([2, 3], [0, 1])

sum.draw('mpl')
````

````python
from qiskit.providers.basic_provider import BasicSimulator
from qiskit.visualization import plot_histogram

simulator_basic = BasicSimulator()
job_sum = simulator_basic.run(sum)

legend = ["Suma: 1 + 0 + 1"]
plot_histogram([job_sum.result().get_counts()], legend=legend)
````

### Ejemplo 2: Suma `A=1`, `B=1`, `CarryIn=1`

Repetimos el proceso con otro conjunto de entradas:

````python
initialization2 = QuantumCircuit(4)
initialization2.x(0)       # A       = 1
initialization2.x(1)       # B       = 1
initialization2.x(2)       # CarryIn = 1
initialization2.barrier()

sum2 = initialization2.compose(adder)
sum2.barrier()
sum2.add_register(ClassicalRegister(2, 'Result'))
sum2.measure([2, 3], [0, 1])
display(sum2.draw('mpl'))

job_sum2 = simulator_basic.run(sum2, shots=1)   # una sola ejecución es suficiente

legend = ["Suma: 1 + 1 + 1"]
plot_histogram([job_sum2.result().get_counts()], legend=legend)
````

### 2.1.1. Uso de la superposición


También podemos aprovechar la superposición y el entrelazamiento.

Por ejemplo, preparemos un estado de Bell entre `A` y `B`, y un estado superpuesto para `CarryIn`.

El estado de entrada es una superposición con componentes correspondientes a **E1**, **E4**, **E5** y **E8**.

````python
initialization3 = QuantumCircuit(4)
initialization3.h(0)        # A y B en estado de Bell
initialization3.cx(0, 1)
initialization3.h(2)        # estado en superposición: ( |0> + |1> ) / sqrt(2)
initialization3.barrier()

sum3 = initialization3.compose(adder)  # el 'adder' es el mismo
sum3.barrier()
sum3.add_register(ClassicalRegister(2, 'Result'))
sum3.measure([2, 3], [0, 1])
display(sum3.draw('mpl'))

job_sum3 = simulator_basic.run(sum3, shots=4000)   # en este caso sí necesitamos múltiples shots

print(job_sum3.result().get_counts())

legend = ["Conteos del resultado de la suma3"]
plot_histogram([job_sum3.result().get_counts()], legend=legend)
````

<a id="2.1.2"></a>
### 2.1.2. Alternativa: usar la clase `CDKMRippleCarryAdder` de Qiskit

Qiskit incluye varias clases para construir sumadores cuánticos, entre ellas [`CDKMRippleCarryAdder`](https://quantum.cloud.ibm.com/docs/api/qiskit/qiskit.circuit.library.CDKMRippleCarryAdder), [`DraperQFTAdder`](https://quantum.cloud.ibm.com/docs/api/qiskit/qiskit.circuit.library.DraperQFTAdder) y [`VBERippleCarryAdder`](https://quantum.cloud.ibm.com/docs/api/qiskit/qiskit.circuit.library.VBERippleCarryAdder).

A continuación tenemos un ejemplo que usa la clase `CDKMRippleCarryAdder` para sumar $3+5$:


````python
from qiskit.circuit.library.arithmetic.adders import CDKMRippleCarryAdder
from qiskit.primitives import StatevectorSampler

qiskit_adder = CDKMRippleCarryAdder(3, 'full', 'Sumador completo')

addend1 = QuantumRegister(3, 'a1')  # primer número a sumar
addend2 = QuantumRegister(3, 'a2')  # segundo número a sumar
aux = QuantumRegister(2, 'aux')     # acarreo de entrada y de salida
cr = ClassicalRegister(4, 'Result') # para almacenar el resultado en bits clásicos

circ = QuantumCircuit(addend1, addend2, aux, cr)

circ.x([addend1[0], addend1[1]])  # 3 en binario: 011
circ.x([addend2[0], addend2[2]])  # 5 en binario: 101

# El orden esperado por la clase 'CDKMRippleCarryAdder' es: carry_in - 1er operando - 2do operando - carry_out
circ.append(qiskit_adder, [aux[0]] + addend1[0:3] + addend2[0:3] + [aux[1]])

circ.measure(addend2[0:3] + [aux[1]], cr) # esta clase escribe el resultado en el 2do operando y en el carry_out
display(circ.draw('mpl'))



sampler = StatevectorSampler()   # porque 'CDKMRippleCarryAdder' no es una compuerta del conjunto base

job_circ = sampler.run([circ])   # en este caso sí necesitamos múltiples shots

print(job_circ.result()[0].data.Result.get_counts())   # resultado es 8, que en binario es 1000

legend = ["CDKMRippleCarryAdder: 5 + 3"]
plot_histogram([job_circ.result()[0].data.Result.get_counts()], legend=legend)  # notemos el nombre del registro clásico
````

### Reflexión

* Este sumador cuántico completo demuestra cómo la lógica clásica se integra en los circuitos cuánticos al rediseñarla de forma reversible. Las computadoras cuánticas implementan esta lógica mediante compuertas cuánticas y también pueden aprovechar propiedades propias de la mecánica cuántica.
* El circuito es completamente reversible, de acuerdo con la naturaleza unitaria de las operaciones cuánticas.
* La superposición permite transformar varias componentes de una entrada. No obstante, la medición entrega un solo resultado y se necesita obtener la estadística.
* Este ejemplo permite intuir cómo las operaciones clásicas reversibles pueden servir como bloques de construcción para algoritmos cuánticos más avanzados.

## 2.2. Protocolo de teleportación

<div align="center"><img width="700px" style="display:block; margin-left: auto; margin-right: auto;" src="https://images.claudiazm.xyz/ecc26/QuantumTeleportationWikipedia.gif"></div>

<p style="font-size:10px;font-style:italic;text-align:right"><a href="https://en.wikipedia.org/wiki/Superdense_coding" target="_blank">Animación obtenida de Wikipedia: Superdense coding</a></p>

Alicia quiere enviar a Bob un estado cuántico de un qubit:

$$
\lvert \psi
\rangle=\alpha\lvert 0
\rangle+\beta\lvert 1
\rangle
$$

El objetivo es transferir el estado sin que Alicia necesite conocer ni comunicar una descripción clásica de $\alpha$ y $\beta$.

El [teorema de la no clonación](https://quantum.cloud.ibm.com/learning/en/courses/basics-of-quantum-information/entanglement-in-action/quantum-teleportation#the-no-cloning-theorem) establece que no es posible crear una copia perfecta de un estado cuántico arbitrario y desconocido. Una CNOT puede copiar estados de la base computacional si el qubit objetivo comienza en $\lvert 0 \rangle$, pero ninguna operación unitaria puede clonar cualquier superposición desconocida. Por tanto, Alicia no puede generar simplemente una copia de $\lvert \psi \rangle$ y entregársela a Bob.

Sin embargo, mediante dos bits clásicos y un par de qubits entrelazados, Alicia puede transferir el estado $\lvert \psi \rangle$ a Bob. Al finalizar, el qubit de Bob tendrá ese estado y el qubit original de Alicia ya no lo conservará. El par entrelazado también se consume durante el proceso. A esto se le conoce como **teleportación cuántica**.

Para realizar el protocolo, Alicia y Bob reciben un par de qubits entrelazados preparado por un tercero, a quien llamaremos Carlos. Alicia conserva uno de ellos y Bob recibe el otro. Alicia realiza operaciones y mediciones sobre el qubit que desea teleportar y sobre su mitad del par entrelazado. Después envía a Bob los dos resultados mediante un *canal de comunicación clásico*. Con esa información, Bob aplica las correcciones necesarias a su qubit.

Veamos los pasos:


### Paso 1

Comenzamos con un estado cuántico desconocido de un qubit:

$$\lvert \psi \rangle=\alpha\lvert 0 \rangle+\beta\lvert 1 \rangle$$

Alicia no necesita conocer los valores de $\alpha$ y $\beta$ para teleportarlo.

Creamos los qubits y registros clásicos necesarios para llevar a cabo el protocolo:


````python
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister

qr_alicia = QuantumRegister(2, name="alicia")  # dos qubits de Alicia
qr_bob = QuantumRegister(1, name="bob")        # un qubit de Bob
crz = ClassicalRegister(1, name="crz")         # 2 bits clasicos
crx = ClassicalRegister(1, name="crx")         # en 2 diferentes registros

circuito_teleportacion = QuantumCircuit(qr_alicia, qr_bob, crz, crx)

circuito_teleportacion.draw('mpl')
````

Su amigo en común, Carlos, prepara un estado entrelazado entre uno de los qubits de Alicia (el qubit 1) y el qubit de Bob. Este es un **estado de Bell**. Para crear el par, Carlos aplica primero una compuerta de Hadamard al qubit 1, que transforma $\lvert 0 \rangle$ en $\lvert + \rangle$. Después aplica una compuerta $CNOT$ con ese qubit como control y el qubit de Bob como objetivo.

El par entrelazado, formado por los qubits $\lvert q_2q_1 \rangle=\lvert 0_b 0_{a_1} \rangle$, queda como:

$$CNOT_{1,2}\big(I\otimes H\big)\lvert 0_b 0_{a_1} \rangle=\frac{1}{\sqrt2}\big(\lvert 0_b 0_{a_1} \rangle+\lvert 1_b 1_{a_1} \rangle\big)=\lvert e \rangle$$


````python
def crear_estado_bell(circ, a, b):
    circ.h(a)     # poner el qubit a en el estado |+>
    circ.cx(a, b) # CNOT con control el qubit a y objetivo el qubit b
````

````python
crear_estado_bell(circuito_teleportacion, 1, 2) # circuito, qubit a, qubit b

circuito_teleportacion.draw('mpl')
````

Después de esto, el estado de los 3 qubits, incluido el estado que se desea enviar $\lvert \psi \rangle$, queda como:

$$\begin{aligned}
\lvert e \rangle\otimes\lvert \psi \rangle&=\frac{1}{\sqrt2}\Big(\big(\lvert 00 \rangle+\lvert 11 \rangle\big)\otimes \alpha\lvert 0 \rangle + \big(\lvert 00 \rangle+\lvert 11 \rangle\big)\otimes \beta\lvert 1 \rangle\Big)\\
&=\frac{1}{\sqrt2}\big(\alpha\lvert 000 \rangle+\alpha\lvert 110 \rangle+\beta\lvert 001 \rangle+\beta\lvert 111 \rangle\big)
\end{aligned}$$


Se envía a Alicia uno de los qubits entrelazados ($|e_1\rangle$) y el otro a Bob ($|e_2\rangle$).

### Paso 2

Ahora, de acuerdo con el protocolo, Alicia aplica una compuerta $CNOT$ sobre sus dos qubits. Utiliza el qubit que contiene $\lvert \psi \rangle$ como control y su mitad del par entrelazado como objetivo. Después aplica una compuerta de Hadamard al qubit que contenía $\lvert \psi \rangle$. Esto da como resultado el estado:

$$\begin{aligned}
\Big(I\otimes I\otimes H\Big)\Big(I\otimes CNOT_{0,1}\Big)\Big(\lvert e \rangle\otimes\lvert \psi \rangle\Big)&=\Big(I\otimes I\otimes H\Big)\Big(I\otimes CNOT_{0,1}\Big)\frac{1}{\sqrt2}\Big(\alpha\lvert 000 \rangle+\alpha\lvert 110 \rangle+\beta\lvert 001 \rangle+\beta\lvert 111 \rangle\Big)\\
&=\Big(I\otimes I\otimes H\Big)\frac{1}{\sqrt2}\Big(\alpha\lvert 000 \rangle+\alpha\lvert 110 \rangle+\beta\lvert 011 \rangle+\beta\lvert 101 \rangle\Big)\\
&=\frac{1}{2}\Big(\alpha\big(\lvert 000 \rangle+\lvert 001 \rangle+\lvert 110 \rangle+\lvert 111 \rangle\big)+\beta\big(\lvert 010 \rangle-\lvert 011 \rangle+\lvert 100 \rangle-\lvert 101 \rangle\big)\Big)
\end{aligned}$$

Esta expresión puede reescribirse como:

$$\begin{aligned}
=\frac{1}{2} \Big( & ~~~~~~\big(\alpha\lvert 0 \rangle+\beta\lvert 1 \rangle\big)~~\lvert 00 \rangle\\
&+ ~\big(\alpha\lvert 0 \rangle-\beta\lvert 1 \rangle\big)~~\lvert 01 \rangle\\
&+ ~\big(\alpha\lvert 1 \rangle+\beta\lvert 0 \rangle\big)~~\lvert 10 \rangle\\
&+ ~\big(\alpha\lvert 1 \rangle-\beta\lvert 0 \rangle\big)~~\lvert 11 \rangle~~~~~\Big)
\end{aligned}$$

En código queda como sigue:


````python
def compuertas_alicia(circ, psi, a):
    circ.barrier()
    circ.cx(psi, a)
    circ.h(psi)
````

````python
compuertas_alicia(circuito_teleportacion, 0, 1)

circuito_teleportacion.draw('mpl')
````

### Paso 3

Alicia mide sus dos qubits y envía a Bob los dos resultados como bits clásicos. Siempre obtiene uno de los cuatro estados de la base computacional, $\lvert 00 \rangle$, $\lvert 01 \rangle$, $\lvert 10 \rangle$ o $\lvert 11 \rangle$. Para un estado $\lvert \psi \rangle$ normalizado, cada resultado tiene probabilidad $1/4$.

De acuerdo con el resultado de la medición, el qubit de Bob se proyecta en uno de los siguientes estados:

$$\lvert 00 \rangle\rightarrow \big(\alpha\lvert 0 \rangle+\beta\lvert 1 \rangle\big)$$
$$\lvert 01 \rangle\rightarrow \big(\alpha\lvert 0 \rangle-\beta\lvert 1 \rangle\big)$$
$$\lvert 10 \rangle\rightarrow \big(\alpha\lvert 1 \rangle+\beta\lvert 0 \rangle\big)$$
$$\lvert 11 \rangle\rightarrow \big(\alpha\lvert 1 \rangle-\beta\lvert 0 \rangle\big)$$

Agregamos las mediciones a los qubits de Alicia:


````python
def mediciones_y_envio(circ, psi, a):
    circ.barrier()
    circ.measure(psi, 0)
    circ.measure(a, 1)
````

````python
mediciones_y_envio(circuito_teleportacion, 0, 1)

circuito_teleportacion.draw('mpl')
````

### Paso 4

Cuando Bob recibe los dos bits de Alicia, puede recuperar el estado original aplicando las transformaciones apropiadas a su qubit, que inicialmente formaba parte del par de Bell.

Las transformaciones que necesita aplicar son:

| Estado de Bob | Bits recibidos | Compuerta(s) a aplicar |
|:-------------:|:--------------:|:----------------------:|
|$$\big(\alpha\lvert 0 \rangle+\beta\lvert 1 \rangle\big)$$|00|$$I$$|
|$$\big(\alpha\lvert 0 \rangle-\beta\lvert 1 \rangle\big)$$|01|$$Z$$|
|$$\big(\alpha\lvert 1 \rangle+\beta\lvert 0 \rangle\big)$$|10|$$X$$|
|$$\big(\alpha\lvert 1 \rangle-\beta\lvert 0 \rangle\big)$$|11|$$ZX$$|

Después de este paso, Bob habrá reconstruido el estado que Alicia deseaba enviar.

**Nota:** la comunicación que Alicia envía a Bob en esta etapa es clásica. El protocolo completo también consume el par entrelazado, por lo que **no permite transmitir información más rápido que la luz**.


````python
def compuertas_bob(circ, b, crz, crx):
    circ.barrier()
    with circ.if_test((crz, 1)):
        circ.z(b)
    with circ.if_test((crx, 1)):
        circ.x(b)
````

````python
compuertas_bob(circuito_teleportacion, 2, crz, crx)

circuito_teleportacion.draw('mpl')
````

Al final del protocolo, el estado que se encontraba en el qubit de Alicia ha sido transferido al qubit de Bob. El qubit físico de Alicia no viajó hasta Bob.

Comprobémoslo ejecutando y simulando el protocolo completo:


````python
from qiskit.quantum_info import random_statevector
from qiskit.visualization import plot_histogram, plot_bloch_multivector, array_to_latex

# generamos un estado aleatorio (el que enviará Alicia a Bob y al ser aleatorio ella no lo conoce)
psi = random_statevector(2)

# veamos de qué estado se trata
display(psi.draw('latex', prefix="|\\psi\\rangle ="))

plot_bloch_multivector(psi)
````

````python
from qiskit.circuit.library import Initialize

# crear una instrucción de inicialización que coloque al qubit en el estado que generamos aleatoriamente
instruccion_inicializacion = Initialize(psi)
instruccion_inicializacion.label = "init"
````

````python
# El protocolo completo:

# Configuración inicial
qr_alicia = QuantumRegister(2, name="alicia")  # dos qubits de Alicia
qr_bob = QuantumRegister(1, name="bob")        # un qubit de Bob
crz = ClassicalRegister(1, name="crz")         # 2 bits clasicos
crx = ClassicalRegister(1, name="crx")         # en 2 diferentes registros
circuito_teleportacion = QuantumCircuit(qr_alicia, qr_bob, crz, crx)

# Paso 0
# Iniciar el qubit a enviar
circuito_teleportacion.append(instruccion_inicializacion, [qr_alicia[0]])
circuito_teleportacion.barrier()

# Paso 1
# Crear el par entrelazado
crear_estado_bell(circuito_teleportacion, qr_alicia[1], qr_bob[0]) # circuito, qubit a, qubit b

# "Enviar" q1 a Alicia y q2 a Bob

# Paso 2
# Alicia aplica sus compuertas a los dos qubits que posee
compuertas_alicia(circuito_teleportacion, qr_alicia[0], qr_alicia[1])

# Paso 3
# Alicia envía sus bits clásicos a Bob
mediciones_y_envio(circuito_teleportacion, qr_alicia[0], qr_alicia[1])

# Paso 4
# Bob decodifica su qubit
compuertas_bob(circuito_teleportacion, qr_bob[0], crz, crx)

circuito_teleportacion.draw('mpl')
````

Utilizaremos otra forma de realizar la simulación. El paquete **[Qiskit Aer](https://qiskit.github.io/qiskit-aer/)** está especializado en la simulación de circuitos cuánticos en computadoras clásicas.

En este ejemplo usamos `StatevectorSimulator` de Aer porque admite las mediciones a mitad del circuito (*mid-circuit measurements*) y la instrucción `Initialize` empleadas en el protocolo.


Instalación de **Qiskit Aer**:

    pip install qiskit-aer

````python
from qiskit_aer import StatevectorSimulator
 
simulator = StatevectorSimulator()

estado_circ_teleportacion = simulator.run(circuito_teleportacion, shots=1).result().get_statevector()  # indicar shots = 1

plot_bloch_multivector(estado_circ_teleportacion)
````

## 2.3. Protocolo de codificación superdensa


<div align="center"><img width="700px" style="display:block; margin-left: auto; margin-right: auto;" src="https://images.claudiazm.xyz/ecc26/SuperdenseCodingWikipedia.gif"></div>

<p style="font-size:10px;font-style:italic;text-align:right"><a href="https://en.wikipedia.org/wiki/Quantum_teleportation" target="_blank">Animación obtenida de Wikipedia: Quantum teleportation</a></p>

### Diferencia entre la Codificación Superdensa y la Teleportación Cuántica


La teleportación cuántica y la codificación superdensa están estrechamente relacionadas. Para evitar confusiones, aclaremos la diferencia.

La teleportación cuántica transfiere el estado de un qubit, $\lvert \psi \rangle$, mediante dos bits de comunicación clásica y un par de Bell compartido. El estado original deja de estar disponible en la ubicación de Alicia y se reconstruye en el qubit de Bob con ayuda del entrelazamiento.

La **codificación superdensa** permite enviar dos bits clásicos mediante la transmisión de un solo qubit. Para lograrlo, Alicia y Bob también deben compartir previamente un par entrelazado, que se consume durante el protocolo.

| Teleportación |                           Codificación Superdensa                            |
|:-------------:|:----------------------------------------------------------------------------:|
| Transmite el estado de un qubit usando dos bits clásicos y un par entrelazado compartido. | Transmite dos bits clásicos usando un qubit y un par entrelazado compartido. |

Los recursos de ambos protocolos son complementarios: la teleportación usa dos bits clásicos para transferir un estado cuántico, mientras que la codificación superdensa usa un qubit para comunicar dos bits clásicos.


### Paso 1

El proceso comienza con una tercera persona, a quien llamaremos Carlos, que prepara dos qubits en un estado entrelazado. Inicializa el sistema en $\lvert 00 \rangle$ y aplica una compuerta Hadamard ($H$) al primer qubit para crear una superposición. Después aplica una compuerta $CNOT$ ($CX$), utilizando el primer qubit como control y el segundo como objetivo. Así obtiene el par de Bell que mencionamos anteriormente.


Comenzamos con el estado:

$$\lvert 00 \rangle=\lvert 0 \rangle_A\otimes\lvert 0 \rangle_B$$

donde el qubit que se enviará a Alicia está etiquetado con $A$ y el que se enviará a Bob está etiquetado con $B$. Carlos aplica primero una compuerta de Hadamard al primer qubit. Esto crea una superposición y produce el estado:

$$\lvert +0 \rangle=\frac{1}{\sqrt2}\Big(\lvert 00 \rangle+\lvert 10 \rangle\Big)$$

Después, Carlos aplica la compuerta $CNOT$. Esta operación entrelaza ambos qubits: invierte el qubit objetivo cuando el control se encuentra en $\lvert 1 \rangle$. En esta notación, el qubit de control es el de la izquierda.

$$CNOT\frac{1}{\sqrt2}\Big(\lvert 00 \rangle+\lvert 10 \rangle\Big)=\frac{1}{\sqrt2}\Big(\lvert 00 \rangle+\lvert 11 \rangle\Big)$$


````python
def crear_par_de_bell():
    qc = QuantumCircuit(2)
    qc.h(1)
    qc.cx(1, 0)
    return qc
````

### Paso 2

Carlos envía el primer qubit a Alicia y el segundo a Bob. El objetivo del protocolo es que Alicia comunique dos bits clásicos a Bob mediante la transmisión de su único qubit. Antes de enviarlo, aplica una operación determinada por los dos bits del mensaje:

| Mensaje a enviar | Compuerta a aplicar | Estado resultante $\Big(\cdot\frac{1}{\sqrt2}\Big)$ |
|:----------------:|:-------------------:|:--------------------------------------------------:|
|00|$$I$$|$$\lvert 00 \rangle+\lvert 11 \rangle$$|
|01|$$X$$|$$\lvert 10 \rangle+\lvert 01 \rangle$$|
|10|$$Z$$|$$\lvert 00 \rangle-\lvert 11 \rangle$$|
|11|$$ZX$$|$$-\lvert 10 \rangle+\lvert 01 \rangle$$|


Por ejemplo, si Alicia quiere enviar $00$, no modifica su qubit, es decir, aplica la identidad ($I$). Si quiere enviar $01$, aplica la compuerta $X$. Para $10$ aplica $Z$ y para $11$ aplica $ZX$. Después de codificar el mensaje, envía su qubit a Bob para completar el protocolo.


````python
def codificar_mensaje(qc, qubit, msg):
    if len(msg) != 2 or not set(msg).issubset({"0", "1"}): # el mensaje debe ser de solo 0s y 1s
        raise ValueError(f"El mensaje '{msg}' es inválido")
    if msg[1] == "1":
        qc.x(qubit)
    if msg[0] == "1":
        qc.z(qubit)
    return qc
````

### Paso 3

Bob recibe el qubit de Alicia, que en nuestra notación es el qubit de la izquierda, y utiliza también su qubit para decodificar el mensaje. No necesita conocer las amplitudes del estado entrelazado ni saber de antemano qué mensaje eligió Alicia.

Bob aplica una compuerta CNOT con el qubit de la izquierda como control y el de la derecha como objetivo. Después aplica una compuerta Hadamard al qubit de la izquierda y, finalmente, mide ambos qubits. En condiciones ideales, obtiene de forma determinista los dos bits enviados por Alicia.

| Bob recibe $\Big(\cdot\frac{1}{\sqrt2}\Big)$ | Después de la compuerta $CNOT$ | Después de la compuerta $H$ |
|:-------------------------------------------:|:--------------------------------:|:----------------------------:|
|$$\lvert 00 \rangle+\lvert 11 \rangle$$|$$\lvert 00 \rangle+\lvert 10 \rangle$$|$$\lvert 00 \rangle$$|
|$$\lvert 10 \rangle+\lvert 01 \rangle$$|$$\lvert 11 \rangle+\lvert 01 \rangle$$|$$\lvert 01 \rangle$$|
|$$\lvert 00 \rangle-\lvert 11 \rangle$$|$$\lvert 00 \rangle-\lvert 10 \rangle$$|$$\lvert 10 \rangle$$|
|$$-\lvert 10 \rangle+\lvert 01 \rangle$$|$$-\lvert 11 \rangle+\lvert 01 \rangle$$|$$\lvert 11 \rangle$$|


````python
def decodificar_mensaje(qc):
    qc.cx(1, 0)
    qc.h(1)
    return qc
````

````python
# El protocolo completo

# Carlos crea un par entrelazado entre Alicia y Bob
qc = crear_par_de_bell()

# Agregamos una barrera para una separación visual
qc.barrier()

# En este punto, el qubit 1 va a Alicia y el qubit 0 va a Bob

# Luego, Alicia codifica su mensaje en el qubit 1. En este caso, 
# queremos enviar el mensaje '10'. Puedes intentar cambiar este
# valor y ver cómo afecta al circuito
mensaje = '10'
qc = codificar_mensaje(qc, 1, mensaje)
qc.barrier()
# Luego Alicia envía su qubit a Bob (q1)

# Después de recibir el qubit 1, Bob aplica el protocolo de recuperación:
qc = decodificar_mensaje(qc)

# Finalmente, Bob mide sus qubits para leer el mensaje de Alicia
qc.measure_all()

# Mostramos el circuito
qc.draw('mpl')
````

````python
from qiskit.primitives import StatevectorSampler as Sampler

sampler = Sampler()
job_qc = sampler.run([(qc, None, 1000)]) # usamos múltiples shots

counts = job_qc.result()[0].data.meas.get_counts() 

print(counts)

legend = ["Mensaje enviado"]
plot_histogram([counts], legend=legend)
````

## 2.4. Más allá de los 2 Qubits: Redes Cuánticas y Entrelazamiento Multipartito


Los protocolos de **Codificación Superdensa** y **Teleportación Cuántica** que hemos estudiado utilizando pares de Bell son los bloques fundamentales de la información cuántica. Sin embargo, ambos conceptos pueden escalarse de manera muy poderosa a sistemas de 3 o más qubits utilizando estados entrelazados multipartitos, como los estados $GHZ$ y los estados $W$.

### 1. Codificación Superdensa Multipartita

Al usar un estado de más qubits, la capacidad del canal de comunicación escala:

* Con un estado GHZ de $n$ qubits ($\frac{1}{\sqrt{2}}[\lvert 00\dots0 \rangle + \lvert 11\dots1 \rangle]$), es posible realizar codificación distribuida entre múltiples emisores y un solo receptor, permitiendo transmitir hasta $n$ bits clásicos de información mediante operaciones locales.

### 2. Teleportación Multipartita y Redes

En el caso de la teleportación, añadir más qubits permite distribuir un estado cuántico a través de una red:

* Estados $GHZ$: Permiten protocolos de Teleportación Controlada. Alice puede teleportar un qubit a Bob, pero el estado queda "bloqueado" hasta que Carlos (un tercer participante) realiza una medición y autoriza la transferencia. Esto actúa como un sistema de llaves criptográficas cuánticas.
* Estados $W$ ($\frac{1}{\sqrt{3}}[\lvert 100 \rangle + \lvert 010 \rangle + \lvert 001 \rangle]$): Ofrecen una ventaja crítica en telecomunicaciones reales debido a su resistencia al ruido. Si uno de los qubits en un estado $W$ se pierde en la fibra óptica, los qubits restantes conservan parte de su entrelazamiento, permitiendo que la teleportación o la codificación sigan funcionando de manera degradada pero exitosa.

Esta capacidad de escalar ambos protocolos utilizando diferentes tipos de física y estructuras de entrelazamiento es el fundamento teórico detrás del desarrollo del Internet Cuántico y el enrutamiento de información en repetidores cuánticos.

## 3.1. Ejemplo: Sumador Completo Cuántico


Antes de enviar un circuito a una QPU es necesario transpilarlo para ese dispositivo. Cada QPU admite un conjunto concreto de instrucciones y posee una conectividad determinada entre sus qubits. La transpilación traduce el circuito a esas instrucciones y adapta las operaciones de dos qubits a la conectividad disponible.

En Qiskit podemos construir un administrador de pases predefinido con `generate_preset_pass_manager()`. Consulta la guía oficial sobre [transpilación con administradores de pases](https://quantum.cloud.ibm.com/docs/en/guides/transpile-with-pass-managers) para obtener más detalles.


````python
from qiskit.transpiler.preset_passmanagers import generate_preset_pass_manager

pass_manager = generate_preset_pass_manager(
    optimization_level=3,
    backend=backend_menos_ocupado,
)

circuito_transpilado = pass_manager.run(sum3)
circuito_transpilado.draw("mpl", idle_wires=False)
````

Ahora podemos enviar el circuito transpilado mediante `SamplerV2`. El trabajo se coloca en la fila de la QPU y su ejecución puede tardar desde unos minutos hasta varias horas, dependiendo de la demanda y del plan de acceso. El envío es asíncrono: `run()` devuelve un objeto que permite consultar el estado y recuperar el resultado posteriormente.


````python
from qiskit_ibm_runtime import SamplerV2 as Sampler

sampler = Sampler(mode=backend_menos_ocupado)
job = sampler.run([circuito_transpilado], shots=1000)

print(f"Identificador del trabajo: {job.job_id()}")
````

No es necesario mantener el notebook ejecutándose mientras la QPU procesa el trabajo. Conviene guardar su identificador. También podemos consultar los trabajos recientes enviados a una QPU:


````python
service.jobs(
    backend_name=backend_menos_ocupado.name,
    limit=3,
)  # obtener los últimos tres trabajos enviados a esa QPU
````

Con el identificador de un trabajo podemos recuperarlo en otra sesión y consultar su resultado cuando haya terminado. Los trabajos también aparecen en la sección [Workloads de IBM Quantum Platform](https://quantum.cloud.ibm.com/workloads).


````python
job_id = job.job_id()
trabajo_recuperado = service.job(job_id)
resultado = trabajo_recuperado.result()
````

````python
counts_hardware = resultado[0].data.Result.get_counts()
plot_histogram([counts_hardware])
````

### Primitivas de Qiskit Runtime


Qiskit Runtime utiliza las interfaces `SamplerV2` y `EstimatorV2`. `SamplerV2` obtiene muestras de las mediciones de un circuito, mientras que `EstimatorV2` estima valores esperados de observables. Ambas primitivas reciben una o más unidades de trabajo llamadas PUB (*Primitive Unified Bloc*).

Consulta la [guía de migración a las primitivas V2](https://quantum.cloud.ibm.com/docs/en/migration-guides/v2-primitives) y la [documentación de `SamplerV2`](https://quantum.cloud.ibm.com/docs/en/api/qiskit-ibm-runtime/sampler-v2) para conocer sus interfaces vigentes.


````python
from qiskit_ibm_runtime import SamplerV2

samplerv2 = SamplerV2(mode=backend_fake)  # sustituir por una QPU real

# Activar el desacoplamiento dinámico con la secuencia XpXm.
samplerv2.options.dynamical_decoupling.enable = True
samplerv2.options.dynamical_decoupling.sequence_type = "XpXm"

print(
    ">> Secuencia de desacoplamiento dinámico:",
    samplerv2.options.dynamical_decoupling.sequence_type,
)
````

````python
from qiskit_ibm_runtime import SamplerOptions, SamplerV2

options = SamplerOptions()
samplerv2 = SamplerV2(mode=backend_fake, options=options)  # sustituir por una QPU real
job_local = samplerv2.run([tr_lv2], shots=1000)
````

La clase `SamplerOptions` permite configurar, entre otros, los siguientes campos:

* `default_shots`: número predeterminado de *shots* cuando no se especifica en el PUB ni en `run()`. El valor predeterminado del servicio es 4096.
* `dynamical_decoupling`: opciones de desacoplamiento dinámico.
* `execution`: opciones de ejecución.
* `twirling`: opciones de *Pauli twirling*.
* `experimental`: opciones experimentales, que pueden cambiar sin previo aviso.

El número de *shots* también puede indicarse directamente mediante `samplerv2.run(..., shots=1000)`, como en el ejemplo anterior.


El **desacoplamiento dinámico** (*dynamical decoupling*, DD) es una técnica que busca reducir los efectos de la descoherencia durante los periodos inactivos de los qubits. Para ello se insertan secuencias de pulsos que ayudan a cancelar parcialmente ciertas interacciones no deseadas con el entorno. Su utilidad depende del circuito, la QPU y su calibración.


El **Pauli twirling** inserta de forma aleatoria compuertas de Pauli alrededor de determinadas operaciones, sin cambiar el cálculo ideal. Al promediar varias realizaciones aleatorias, algunos errores coherentes se transforman en ruido con una estructura más sencilla de tratar. Esta técnica aumenta el número de circuitos que deben ejecutarse y no elimina todo el ruido.


La mitigación de errores de `EstimatorV2` se configura mediante sus opciones de resiliencia. Consulta la guía oficial para [configurar la mitigación de errores en Qiskit Runtime](https://quantum.cloud.ibm.com/docs/en/guides/configure-error-mitigation).


````python
from qiskit_ibm_runtime import EstimatorV2 as Estimator

estimator = Estimator(mode=backend_fake)  # sustituir por una QPU real
options = estimator.options

# Activar Pauli twirling y la mitigación de errores de lectura.
options.twirling.enable_gates = True
options.resilience.measure_mitigation = True

print(f">>> Pauli twirling activado: {options.twirling.enable_gates}")
print(f">>> Mitigación de lectura activada: {options.resilience.measure_mitigation}")
````

La **extrapolación a ruido cero** (*Zero-Noise Extrapolation*, ZNE) es una técnica de mitigación que ejecuta variantes del cálculo con distintos factores de ruido y extrapola los resultados hacia el límite ideal de ruido cero. En Qiskit Runtime, el escalamiento se realiza mediante transformaciones controladas de las compuertas del circuito.

ZNE puede reducir el sesgo de una estimación, pero requiere más ejecuciones y no garantiza que el resultado sea exacto. Los factores de ruido pueden configurarse explícitamente, como en el siguiente ejemplo.


````python
estimator = Estimator(mode=backend_seleccionado)

estimator.options.resilience_level = 2
estimator.options.resilience.zne_mitigation = True
estimator.options.resilience.zne.noise_factors = [1, 3, 5]
````
