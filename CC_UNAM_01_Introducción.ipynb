# 1. Introducción

<a id="1.1"></a>
## 1.1. Computación cuántica, ¿cómo la uso?

La computación cuántica es una tecnología emergente que aprovecha fenómenos de la mecánica cuántica (**superposición, interferencia y entrelazamiento**) para abordar ciertos problemas difíciles para las computadoras clásicas. Los dispositivos que realizan estos cálculos se denominan hardware cuántico.

Al igual que en la computación clásica, podemos crear **algoritmos cuánticos** para estas nuevas computadoras. Una forma común de implementarlos es mediante circuitos cuánticos, que se construyen utilizando **compuertas cuánticas**.

## 1.2. ¿En dónde estamos actualmente?

Las computadoras cuánticas actuales todavía tienen capacidades limitadas y son sensibles al ruido. Aunque el número de qubits ha crecido, aún no existe una computadora cuántica de propósito general capaz de ejecutar algoritmos grandes con tolerancia a fallos.

Hay dos factores limitantes en los problemas que pueden resolver:

1. **Cantidad de qubits**: Un **qubit** es un *bit cuántico*. Si no tenemos suficientes qubits, no podemos representar ni procesar problemas grandes.
2. **Tasa de error**: Las operaciones y mediciones no son perfectas. Estos errores se acumulan conforme el circuito crece y pueden hacer que el resultado deje de ser confiable.

Un ejemplo que permite dimensionar el reto es la factorización de **RSA-2048**, un número de 2048 bits (aproximadamente 617 dígitos). La conocida cifra de **20 millones de qubits físicos ruidosos para factorizarlo en unas ocho horas** proviene de una estimación presentada en 2019 y [publicada en 2021](https://quantum-journal.org/papers/q-2021-04-15-433/). El cálculo suponía, entre otras condiciones, una tasa de error físico de 0.1 %, ciclos de corrección de errores de 1 μs y una arquitectura basada en códigos de superficie.

Las estimaciones han disminuido conforme mejoran los algoritmos, los códigos de corrección de errores y las arquitecturas. En 2026, una prepublicación sobre la [arquitectura Pinnacle](https://arxiv.org/abs/2602.11457) estimó que RSA-2048 podría factorizarse con **menos de 100 000 qubits físicos** mediante un tipo de corrección de errores denominado códigos qLDPC, bajo supuestos de error y velocidad similares. Otra [estimación de 2026 basada en una arquitectura heterogénea](https://arxiv.org/abs/2604.06319) calculó **381 000 qubits físicos y 9.2 días** con una disposición de qubits en rejilla ya demostrada experimentalmente, o **439 000 qubits y 4.9 días** al añadir un acelerador especializado. Estas cifras son proyecciones teóricas condicionadas a cada arquitectura, no capacidades alcanzadas por el hardware actual.

Como referencia del hardware disponible, a comienzos de 2026 IBM reportaba [2 344 qubits distribuidos entre 16 unidades de procesamiento cuántico (QPU)](https://www.ibm.com/quantum/blog/whats-new-q1-2026), no en una sola computadora tolerante a fallos. Además, la cantidad de qubits no basta para medir la capacidad de un sistema: también importan su **calidad, conectividad, velocidad y la profundidad de los circuitos que puede ejecutar**.

Mientras se desarrolla la computación tolerante a fallos, también se investigan posibles aplicaciones útiles con hardware ruidoso. Un ejemplo es el trabajo liderado por IBM [*Evidence for the utility of quantum computing before fault tolerance*](https://www.nature.com/articles/s41586-023-06096-3), publicado en 2023. Más recientemente, [*Quantum error correction below the surface code threshold*](https://www.nature.com/articles/s41586-024-08449-y), publicado en *Nature* en 2025, mostró que al aumentar el tamaño del código de superficie disminuía el error lógico, un avance importante hacia computadoras cuánticas tolerantes a fallos.

Empresas, universidades y gobiernos continúan invirtiendo en hardware, algoritmos y corrección de errores. Al mismo tiempo, cualquier persona puede crear y ejecutar circuitos cuánticos para aprender a utilizar esta tecnología emergente.

# 2. Frameworks para Computación Cuántica

Actualmente existen múltiples formas de trabajar con computación cuántica: desde simular sistemas cuánticos en computadoras clásicas hasta acceder a computadoras cuánticas reales (**hardware cuántico**) a través de la nube.

Para programar una computadora cuántica necesitamos un lenguaje de programación. La mayoría de las herramientas actuales extienden lenguajes ya existentes mediante frameworks, bibliotecas o SDK. A continuación veremos algunos de ellos.

<a id="2.1"></a>
## 2.1. Qiskit

Desarrollado por **IBM Quantum**, Qiskit es un SDK de código abierto para computación cuántica que permite escribir, simular y ejecutar programas cuánticos en hardware real y simuladores. Corre bajo el lenguaje de programación Python.

- Página principal: https://www.ibm.com/quantum/qiskit
- Tutoriales: https://quantum.cloud.ibm.com/learning/en
- API: https://quantum.cloud.ibm.com/docs/en/api/qiskit
- Repositorio: https://github.com/Qiskit
- Instalación
```
    pip install qiskit
    pip install qiskit[visualization]

    pip install qiskit-ibm-runtime
```

<a id="2.2"></a>
## 2.2. Cirq

Creado por **Google Quantum AI**, Cirq es un framework de Python para crear, editar y ejecutar circuitos cuánticos en simuladores cuánticos y hardware cuántico.

- Página principal: https://quantumai.google/cirq
- Tutoriales: https://quantumai.google/cirq/start/start
- API: https://quantumai.google/reference/python/cirq/all_symbols
- Repositorio: https://github.com/quantumlib/Cirq
- Instalación
```
    pip install cirq
```

<a id="2.3"></a>
## 2.3. PennyLane

Un framework de **Xanadu** que se centra en la computación cuántica híbrida clásica-cuántica y el Machine Learning Cuántico.

- Página principal: https://pennylane.ai/
- Tutoriales: https://pennylane.ai/demonstrations/
- API: https://docs.pennylane.ai/en/stable/
- Repositorio: https://github.com/PennyLaneAI/pennylane
- Instalación
```
    pip install pennylane
```

<a id="2.4"></a>
## 2.4. QuTiP

QuTiP (Quantum Toolbox in Python), creado por J.R. Johansson y P.D. Nation, está más enfocado en la simulación de sistemas cuánticos abiertos. QuTiP es ampliamente utilizado para estudios en computación cuántica y óptica cuántica.

- Página principal: https://qutip.org/
- Tutoriales: https://qutip.org/qutip-tutorials/
- API: https://qutip.readthedocs.io/en/stable/apidoc/apidoc.html
- Repositorio: https://github.com/qutip/qutip
- Instalación
```
    pip install qutip
```

<a id="2.5"></a>
## 2.5. ProjectQ

Un framework de código abierto para computación cuántica en Python creado por investigadores de la **ETH Zúrich**, permite implementar y ejecutar programas cuánticos en simuladores y hardware real de terceros (IBM, AWS Braket, Azure Quantum, IonQ).

- Página principal: https://www.projectq.ch/
- Tutoriales: https://projectq.readthedocs.io/en/latest/tutorials.html
- API: https://projectq.readthedocs.io/en/latest/projectq.html
- Repositorio: https://github.com/ProjectQ-Framework/ProjectQ
- Instalación
```
    pip install projectq
```

<a id="2.6"></a>
## 2.6. Q#

Creado por **Microsoft**, Q# es un lenguaje de programación específico para la computación cuántica. Forma parte del Quantum Development Kit (QDK), que proporciona herramientas para escribir, simular y ejecutar programas cuánticos en distintos entornos, así como para integrar Q# con Python.

- Página principal: https://learn.microsoft.com/en-us/azure/quantum/qsharp-overview
- Tutoriales: https://learn.microsoft.com/en-us/azure/quantum/qsharp-quickstart
- API: https://learn.microsoft.com/en-us/qsharp/api/qsharp-lang/
- Repositorio: https://github.com/microsoft/qdk
- Instalación para Python y Jupyter
```
    python -m pip install "qdk[jupyter]" ipykernel
```
**Más detalles para la instalación:** https://learn.microsoft.com/en-us/azure/quantum/install-overview-qdk

# 3. Plataformas con acceso a Hardware Cuántico

Varias plataformas ofrecen acceso a hardware cuántico a través de la nube, lo que permite a estudiantes, investigadores, desarrolladores y empresas ejecutar experimentos en computadoras cuánticas reales. A continuación se presentan algunas de ellas.

<a id="3.1"></a>
## 3.1. IBM Quantum Platform

IBM Quantum Platform permite acceder a computadoras cuánticas de IBM a través de la nube. Mediante Qiskit y Qiskit Runtime, los usuarios pueden crear y ejecutar circuitos en hardware real o simuladores.

Ofrece planes gratuitos y de pago para utilizar procesadores cuánticos superconductores.

- Página principal: https://quantum.cloud.ibm.com/

<a id="3.2"></a>
## 3.2. Amazon Braket

Amazon Braket es un servicio administrado de AWS para construir, probar y ejecutar algoritmos cuánticos. Proporciona una interfaz unificada para acceder a computadoras cuánticas y simuladores de distintos proveedores.

Entre el hardware disponible en 2026 se encuentran sistemas de AQT e IonQ (iones atrapados), IQM y Rigetti (circuitos superconductores), y QuEra (arreglos de átomos neutros). La oferta puede cambiar con el tiempo.

- Página principal: https://aws.amazon.com/braket/

<a id="3.3"></a>
## 3.3. Azure Quantum

Azure Quantum es la plataforma de Microsoft para computación cuántica en la nube. Proporciona acceso a hardware cuántico y simuladores de distintos proveedores desde un mismo entorno.

Entre sus proveedores actuales se encuentran IonQ y Quantinuum (iones atrapados), Pasqal (átomos neutros) y Rigetti (circuitos superconductores). Los dispositivos disponibles dependen de la región y del plan contratado.

- Página principal: https://azure.microsoft.com/en-us/solutions/quantum-computing/

<a id="3.4"></a>
## 3.4. Google Quantum AI

Google Quantum AI desarrolla procesadores y algoritmos cuánticos. Su Quantum Computing Service permite ejecutar programas escritos con Cirq en procesadores superconductores de Google; sin embargo, el acceso continúa restringido a grupos aprobados y normalmente requiere colaborar con un patrocinador de Google.

- Página principal: https://quantumai.google/
- Información de acceso: https://quantumai.google/cirq/google/access

<a id="3.5"></a>
## 3.5. Rigetti Quantum Cloud Services (QCS)

Rigetti Quantum Cloud Services (QCS) proporciona acceso directo a las computadoras cuánticas de Rigetti y un entorno JupyterLab con herramientas para desarrollar y ejecutar programas cuánticos. Sus sistemas también están disponibles mediante otros servicios en la nube.

Su hardware cuántico está basado en circuitos superconductores.

- Página principal: https://docs.rigetti.com/qcs/

<a id="3.6"></a>
## 3.6. IonQ

IonQ ofrece acceso directo a sus computadoras cuánticas y simuladores mediante IonQ Quantum Cloud. Sus sistemas, basados en iones atrapados, también están disponibles a través de servicios como Amazon Braket, Microsoft Azure Quantum y Google Cloud Marketplace.

- Página principal: https://www.ionq.com/quantum-cloud

<a id="3.7"></a>
## 3.7. D-Wave Leap

D-Wave Leap proporciona acceso en la nube a computadoras de recocido cuántico (*quantum annealing*) y a solucionadores híbridos cuántico-clásicos. Estos sistemas están orientados principalmente a problemas de optimización y muestreo.

- Página principal: https://cloud.dwavesys.com/leap/

** No se puede acceder desde México, a menos que se use una VPN.

## 3.8. Quandela Cloud

## 4.3. Instalación local

La guía interactiva de la subsección anterior explica paso a paso la instalación en Windows, macOS y Linux, tanto de forma local como remota. Incluye instrucciones para distintos administradores de entornos y para ejecutar los notebooks desde Jupyter, Visual Studio Code o PyCharm.

Como referencia adicional, se puede consultar la [guía oficial de instalación de Qiskit](https://quantum.cloud.ibm.com/docs/en/guides/install-qiskit).

<div align="center"><img width="200px" style="display:block; margin-left: auto; margin-right: auto;" src="https://images.claudiazm.xyz/ecc26/qr-qiskit.png">

### Instalación recomendada con uv

Para una instalación rápida y reproducible se recomienda [uv](https://docs.astral.sh/uv/). Después de instalarlo, abre una terminal en la carpeta del repositorio y ejecuta:

```shell
uv python install 3.13
uv venv --python 3.13 .venv
```

Activa el entorno en Windows PowerShell:

```powershell
.\.venv\Scripts\Activate.ps1
```

O en macOS y Linux:

```shell
source .venv/bin/activate
```

Instala Qiskit con sus dependencias de visualización, Jupyter Notebook, JupyterLab y el kernel para notebooks:

```shell
uv pip install "qiskit[visualization]" jupyter
```

Comprueba las versiones e inicia Jupyter Notebook:

```shell
python --version
python -c "import qiskit; print(qiskit.__version__)"
python -m jupyter notebook
```

Jupyter Notebook se utiliza como opción principal porque su interfaz resulta más directa para trabajar con un notebook a la vez. Como alternativa, puedes ejecutar `python -m jupyter lab`: JupyterLab es conveniente cuando quieres abrir varios notebooks, archivos y terminales en pestañas dentro de un mismo espacio de trabajo.

La guía interactiva contiene también las instrucciones completas para Conda, Anaconda, virtualenv, Pipenv, Poetry, Docker y una instalación sin entorno virtual.

# 5. Qubit

Así como el **bit** es la unidad fundamental de información en una computadora clásica, el **qubit** es la unidad fundamental de información cuántica.

Una definición más precisa de **qubit** sería:

<div style="color: #3c763d; background-color: #dff0d8; border: 1px solid #d6e9c6; border-radius: 4px; padding: 15px; margin-bottom: 20px;">
Un qubit puede realizarse físicamente mediante un sistema cuántico de dos niveles (dos estados). Su estado puro se representa con un vector unitario en un espacio de Hilbert bidimensional sobre los números complejos ($\mathbb{C}^2$).

Un [espacio de Hilbert](https://en.wikipedia.org/wiki/Hilbert_space) es un espacio vectorial equipado con un producto interno, que permite definir longitudes y ortogonalidad. Para un qubit se utiliza $\mathbb{C}^2$: sus vectores tienen dos componentes complejas. Estas componentes son las amplitudes asociadas a los dos estados de la base computacional. Puede consultarse una introducción más amplia en [IBM Quantum Learning](https://quantum.cloud.ibm.com/learning/en/courses/basics-of-quantum-information).

<details>
<summary><strong>Recordatorio: Números complejos</strong></summary>

Los números complejos amplían a los números reales mediante la **unidad imaginaria** $i$, definida por

$$i^2=-1.$$

Todo número complejo puede escribirse como

$$z=a+bi,$$

donde $a,b\in\mathbb{R}$. El número $a$ es la parte real y $b$ es el coeficiente de la parte imaginaria. Por ejemplo, para $z=3+2i$, la parte real es $3$ y la parte imaginaria es $2$. Los números reales también son complejos: basta tomar $b=0$.

Una forma útil de visualizarlos es como puntos $(a,b)$ en un plano. El **conjugado** de $z$ es $z^*=a-bi$ y su módulo es

$$|z|=\sqrt{a^2+b^2}, \qquad |z|^2=z^*z=a^2+b^2.$$

Esta última expresión será importante porque las probabilidades de medición de un qubit se calculan con los módulos al cuadrado de sus amplitudes complejas.

</details>

## 5.1. Representación de los qubits: Matemáticamente

Para representar los estados de los qubits se utiliza la **[notación de Dirac](https://en.wikipedia.org/wiki/Bra%E2%80%93ket_notation)**, también conocida como notación *bra-ket*.

El estado $\lvert 0 \rangle$ se pronuncia “**ket cero**” y el estado $\lvert 1 \rangle$, “**ket uno**”. Estos dos estados son ortogonales entre sí y forman la **base computacional** $\{\lvert 0 \rangle,\lvert 1 \rangle\}$, que genera el espacio de Hilbert bidimensional $\mathbb{C}^2$ del qubit.

La notación de Dirac es una forma compacta de escribir **vectores**. Para indicar la amplitud asociada a cada uno de los dos estados de la base necesitamos un vector con dos componentes. Por ello, el estado puro más general de un qubit se escribe como

$$\lvert \psi \rangle=\begin{bmatrix}  \alpha \\ \beta \end{bmatrix}$$

donde $\alpha,\beta\in\mathbb{C}$ son amplitudes complejas y el ket $\lvert \psi \rangle$ representa el vector de estado.

En el caso de los elementos de la base computacional $\{\lvert 0 \rangle, \lvert 1 \rangle\}$, representan a los *vectores de estado* siguientes:

$$\lvert 0 \rangle=\begin{bmatrix}  1 \\ 0 \end{bmatrix} ~~~~~~~~~~~~~~~~~~~~ \lvert 1 \rangle=\begin{bmatrix}  0 \\ 1 \end{bmatrix}$$

Un qubit es una superposición de los estados de la base; es decir, puede expresarse como una [combinación lineal](https://en.wikipedia.org/wiki/Linear_combination) de $\lvert 0 \rangle$ y $\lvert 1 \rangle$:

$$\lvert \psi \rangle=\alpha\lvert 0 \rangle+\beta\lvert 1 \rangle$$

donde $\alpha$ y $\beta$ son [amplitudes de probabilidad](https://en.wikipedia.org/wiki/Probability_amplitude). Al medir este qubit en la base computacional, la probabilidad de obtener $0$ es $|\alpha|^2$ y la de obtener $1$ es $|\beta|^2$. Como estas son las únicas posibilidades, sus probabilidades deben sumar uno:

$$|\alpha|^2 + |\beta|^2 = 1$$

**Nota**: Esta condición se llama **normalización** y hace que la longitud del vector de estado sea igual a $1$. La [documentación de IBM Quantum sobre mediciones](https://quantum.cloud.ibm.com/docs/en/guides/measure-qubits) muestra esta misma relación entre amplitudes y probabilidades.

<div style="color: #31708f; background-color: #d9edf7; border: 1px solid #bce8f1; border-radius: 4px; padding: 15px; margin-bottom: 20px;">
<strong>Ejemplo</strong>:

$$\lvert \psi \rangle=\frac{\sqrt{3}}{2}\lvert 0 \rangle+\frac{1}{2}\lvert 1 \rangle=\begin{bmatrix}  \dfrac{\sqrt{3}}{2} \\\\ \dfrac{1}{2} \end{bmatrix}$$

La probabilidad de observar el estado $\lvert 0 \rangle$ es de $\left|\dfrac{\sqrt{3}}{2}\right|^2=\dfrac{3}{4}$, que es $75\%$.
    
La probabilidad de observar el estado $\lvert 1 \rangle$ es de $\left|\dfrac{1}{2}\right|^2=\dfrac{1}{4}$, que es $25\%$.
    
La suma de las probabilidades para cada caso: $\dfrac{3}{4}+\dfrac{1}{4}=1$, que es $100\%$.

<details>
<summary><strong>Recordatorio: Base de un espacio vectorial</strong></summary>

Un **espacio vectorial** es un conjunto de objetos llamados vectores que pueden sumarse entre sí y multiplicarse por escalares, respetando ciertas reglas. Por ejemplo, las flechas del plano forman el espacio $\mathbb{R}^2$.

Una **combinación lineal** de los vectores $\vec v_1,\ldots,\vec v_n$ es una expresión de la forma

$$c_1\vec v_1+\cdots+c_n\vec v_n,$$

donde los $c_i$ son escalares. Un conjunto $B$ es una **base** de un espacio vectorial $V$ cuando cumple dos condiciones: sus vectores son linealmente independientes (ninguno puede construirse combinando los demás) y generan todo $V$. Por eso, cada vector de $V$ tiene una representación única como combinación lineal de los elementos de $B$.

Por ejemplo, $(1,0)$ y $(0,1)$ forman la base usual de $\mathbb{R}^2$: cualquier vector $(x,y)$ se escribe como $x(1,0)+y(0,1)$. De forma análoga, $\lvert 0 \rangle$ y $\lvert 1 \rangle$ forman la base computacional de $\mathbb{C}^2$.

</details>

### Otras bases para los qubits

La base más común es la base computacional, formada por $\{\lvert 0 \rangle,\lvert 1 \rangle\}$.

Cualquier par de vectores linealmente independientes de $\mathbb{C}^2$ forma una base matemática. Sin embargo, las bases utilizadas para describir mediciones cuánticas se eligen **ortonormales**: sus vectores son ortogonales y tienen longitud uno.

Otra base muy común es $\{\lvert + \rangle,\lvert - \rangle\}$, cuyos vectores se definen como

$$\lvert + \rangle=\begin{bmatrix}  \frac{1}{\sqrt{2}} \\ \frac{1}{\sqrt{2}} \end{bmatrix} ~~~~~~~~~~~~~~~~~~~~ \lvert - \rangle=\begin{bmatrix}  \frac{1}{\sqrt{2}} \\ -\frac{1}{\sqrt{2}} \end{bmatrix}$$

Estos estados también pueden escribirse en términos de la base computacional:

$$\lvert + \rangle=\dfrac{1}{\sqrt2}\Big(\lvert 0 \rangle+\lvert 1 \rangle\Big)$$

$$\lvert - \rangle=\dfrac{1}{\sqrt2}\Big(\lvert 0 \rangle-\lvert 1 \rangle\Big)$$

En estos cuatro vectores los coeficientes son reales. Esto no contradice que el espacio sea complejo: todo número real también pertenece a $\mathbb{C}$, pues puede escribirse como $z=a+0i$.

Por ello, estos cuatro vectores pueden representarse fácilmente en un plano:

## 5.2. Representación de los qubits: Esfera de Bloch

¿Qué ocurre cuando las amplitudes de un qubit no son solamente números reales?

En general son números complejos, por lo que pueden tener una parte real y una parte imaginaria: $z=a+bi$.

El estado de un qubit se escribe como

$$\lvert \psi \rangle=\begin{bmatrix}  \alpha \\ \beta \end{bmatrix}$$

Como $\alpha,\beta\in\mathbb{C}$ y cada número complejo se describe mediante dos números reales, inicialmente se necesitan cuatro números reales:

$$\alpha=a+bi$$

$$\beta=c+di$$

donde $a,b,c,d \in \mathbb{R}$.

A primera vista hay cuatro grados de libertad. La condición de normalización $|\alpha|^2+|\beta|^2=1$ elimina uno, de modo que los vectores normalizados forman una superficie tridimensional, dejándonos con 3 grados de libertad.

Entonces, utilizando una *transformación* de coordenadas conveniente, se puede eliminar ese grado de libertad, es decir, lograr la conversión de $\mathbb{R}^4\rightarrow\mathbb{R}^3$. Una elección posible son las [coordenadas de Hopf](https://en.wikipedia.org/wiki/3-sphere#Hopf_coordinates), que manda el espacio de $4$ dimensiones a la superficie de una esfera unitaria en $3$ dimensiones.

Además, dos vectores que solo difieren por una **fase global** representan el mismo estado físico. Al eliminar ese parámetro queda un estado descrito por dos ángulos reales, $\theta$ y $\varphi$, que determinan un punto sobre la superficie de la **esfera de Bloch**. Esta representación se desarrolla con más detalle en [IBM Quantum Learning](https://quantum.cloud.ibm.com/learning/en/courses/general-formulation-of-quantum-information/density-matrices/bloch-sphere).

<div style="color: #3c763d; background-color: #dff0d8; border: 1px solid #d6e9c6; border-radius: 4px; padding: 15px; margin-bottom: 20px;">
La esfera de Bloch es una representación geométrica de los estados de un sistema cuántico de dos niveles (un qubit).

<img  width="400" style="display:block; margin-left: auto; margin-right: auto;" src="https://images.claudiazm.xyz/ecc26/BlochSphereLoop.gif"/>

Una vez elegida la fase global de modo que $\alpha$ sea real y no negativa, cualquier estado puro de un qubit puede escribirse como

$$\begin{aligned}
\lvert \psi \rangle&=\cos\left(\frac{\theta}{2}\right) \, \lvert 0 \rangle+e^{i\varphi}\sin\left(\frac{\theta}{2}\right) \, \lvert 1 \rangle\\\\\\
&=\cos\left(\frac{\theta}{2}\right)\begin{bmatrix}  1 \\ 0 \end{bmatrix} + e^{i\varphi}\sin\left(\frac{\theta}{2}\right)\begin{bmatrix}  0 \\ 1 \end{bmatrix}\\\\
&=\begin{bmatrix}  \cos\left(\frac{\theta}{2}\right) \\ e^{i\varphi}\sin\left(\frac{\theta}{2}\right) \end{bmatrix}
\end{aligned}$$

donde 

$$0\leq\theta\leq\pi$$

$$0\leq\varphi < 2\pi$$

Por lo tanto,

$$\alpha=\cos\left(\frac{\theta}{2}\right)$$

$$\beta=e^{i\varphi}\sin\left(\frac{\theta}{2}\right)$$

Con esta elección, $\alpha\in\mathbb{R}$ y $\beta\in\mathbb{C}$.

**Nota**: Los parámetros $\theta$ y $\varphi$ corresponden, respectivamente, al ángulo polar medido desde el eje $z$ y al ángulo azimutal medido en el plano $xy$. Las coordenadas cartesianas del punto son $(x,y,z)=(\sin\theta\cos\varphi,\sin\theta\sin\varphi,\cos\theta)$, con radio $r=1$.

En la esfera de Bloch, los [puntos antipodales](https://en.wikipedia.org/wiki/Antipodal_point) representan estados mutuamente ortogonales. Por convención, el polo norte y el polo sur corresponden a $\lvert 0 \rangle$ y $\lvert 1 \rangle$, respectivamente. En una realización física basada en el espín de un electrón podrían asociarse, por ejemplo, con $\lvert \uparrow \rangle$ y $\lvert \downarrow \rangle$; la asociación concreta depende del sistema físico.

Así, los vectores de Bloch de $\lvert 0 \rangle$ y $\lvert 1 \rangle$ son antiparalelos en la representación tridimensional, mientras que sus vectores de estado son ortogonales en el espacio de Hilbert.

La distinción puede resultar confusa al principio: la esfera de Bloch es una representación geométrica, no el espacio de Hilbert mismo.

### Otras bases en la esfera de Bloch

<img width="400" style="display:block; margin-left: auto; margin-right: auto;" src="https://images.claudiazm.xyz/ecc26/BlochSphereBases.gif"/>

Los estados de la base $\{\lvert 0 \rangle,\lvert 1 \rangle\}$ son puntos antipodales sobre el eje $z$, mientras que los de $\{\lvert + \rangle,\lvert - \rangle\}$ se encuentran sobre el eje $x$. También existe una base asociada con el eje $y$, formada por $\{\lvert +i \rangle,\lvert -i \rangle\}$. En términos de la base computacional,

$$\lvert +i \rangle=\dfrac{1}{\sqrt2}\Big(\lvert 0 \rangle+i\lvert 1 \rangle\Big)=\begin{bmatrix}  \frac{1}{\sqrt{2}} \\ \frac{i}{\sqrt{2}} \end{bmatrix}$$

$$\lvert -i \rangle=\dfrac{1}{\sqrt2}\Big(\lvert 0 \rangle-i\lvert 1 \rangle\Big)=\begin{bmatrix}  \frac{1}{\sqrt{2}} \\ -\frac{i}{\sqrt{2}} \end{bmatrix}$$

**Nota**: Estas tres bases ortonormales son las más comunes, pero no son las únicas. Para describir una base de medición de un qubit se puede elegir cualquier par de estados ortogonales y normalizados; en la esfera de Bloch aparecerán como puntos antipodales.

<div style="color: #31708f; background-color: #d9edf7; border: 1px solid #bce8f1; border-radius: 4px; padding: 15px; margin-bottom: 20px;">
<strong>Ejemplo</strong>: Comprobemos que los elementos de la base $\{\lvert +i \rangle,\lvert -i \rangle\}$ son ortogonales.
    
    
Dos vectores son ortogonales cuando su producto interno es igual a cero. En notación de Dirac este producto se escribe $\langle \phi \mid \psi \rangle$. El bra $\langle \phi \rvert$ es el complejo conjugado transpuesto (también llamado **daga**) del ket $\lvert \phi \rangle$. Entonces,

    
$$\langle +i \mid -i \rangle=\begin{bmatrix}  \frac{1}{\sqrt{2}} \\ \frac{i}{\sqrt{2}} \end{bmatrix}^{\dagger} \; \begin{bmatrix}  \frac{1}{\sqrt{2}} \\ \frac{-i}{\sqrt{2}} \end{bmatrix}=\begin{bmatrix}  \frac{1}{\sqrt{2}} & \frac{-i}{\sqrt{2}} \end{bmatrix}\cdot\begin{bmatrix}  \frac{1}{\sqrt{2}} \\ \frac{-i}{\sqrt{2}} \end{bmatrix}=\left(\frac{1}{\sqrt{2}}\right)\left(\frac{1}{\sqrt{2}}\right)+\left(\frac{-i}{\sqrt{2}}\right)\left(\frac{-i}{\sqrt{2}}\right)=\frac{1}{2}-\frac{1}{2}=0$$

# 6. Circuito cuántico

Para realizar operaciones con los qubits utilizamos un **circuito cuántico**.

<div style="color: #3c763d; background-color: #dff0d8; border: 1px solid #d6e9c6; border-radius: 4px; padding: 15px; margin-bottom: 20px;">
Un circuito cuántico es un modelo de computación cuántica, similar a un circuito clásico, en el que una computación se describe como una secuencia de compuertas cuánticas, mediciones, inicializaciones de qubits en estados conocidos y, posiblemente, otras operaciones.

Las condiciones mínimas que debe satisfacer una implementación física para permitir la computación cuántica se conocen como los [criterios de DiVincenzo](https://en.wikipedia.org/wiki/DiVincenzo%27s_criteria). Entre ellas se encuentran la capacidad de inicializar los qubits, aplicar un conjunto universal de compuertas y medir su estado.

Estos circuitos se representan de manera que el eje horizontal indica el avance del tiempo, desde el lado izquierdo hacia el derecho. Las líneas horizontales sencillas representan qubits y las líneas dobles representan bits clásicos. Los elementos conectados por estas líneas corresponden a operaciones, como compuertas y mediciones.

<div align="center"><img width="450px" style="display:block; margin-left: auto; margin-right: auto;" src="https://images.claudiazm.xyz/ecc26/circuit-qiskit.png">

Veamos cómo incluir en un circuito uno de los qubits que creamos anteriormente:

````python
from qiskit import QuantumCircuit

circuit = QuantumCircuit(qubit1) # crear un circuito cuántico con un qubit

circuit.draw('mpl')              # mostramos la representación gráfica del circuito
````

En el siguiente ejemplo se define un circuito cuántico con 3 qubits. No es necesario crearlos previamente con la clase `QuantumRegister`; basta con indicar cuántos qubits debe tener el circuito. También se definen 2 bits clásicos, representados mediante líneas dobles, y se aplican 3 compuertas. Más adelante veremos qué efecto tiene cada una.

````python
circ = QuantumCircuit(3, 2) # circuito con 3 qubits y 2 bits clásicos

circ.h(0)           # aplicar compuerta H al qubit 0
circ.cx(0, 1)       # aplicar compuerta CNOT a los qubits 0 y 1
circ.cx(0, 2)       # aplicar compuerta CNOT a los qubits 0 y 2
circ.measure(0, 1)  # medir el qubit 0 en el bit clásico 1
circ.measure(2, 0)  # medir el qubit 2 en el bit clásico 0

circ.draw('mpl')    # mostrar el circuito
````

Como se mencionó, el tiempo avanza de izquierda a derecha. Por lo tanto, primero se aplica la compuerta $H$, después la compuerta $CNOT_{0,1}$ y, por último, la compuerta $CNOT_{0,2}$. Al final se realizan las mediciones.

Un punto importante es que, de manera predeterminada, *todos los qubits de un circuito de Qiskit comienzan en el estado $\lvert 0 \rangle$*. La capacidad de inicializar los qubits en un estado conocido forma parte de los criterios de DiVincenzo.

# 7. Compuerta cuántica

Una vez que tenemos qubits en un circuito cuántico, necesitamos manipularlos para implementar algoritmos. Esta manipulación se realiza mediante **compuertas cuánticas**.

<div style="color: #3c763d; background-color: #dff0d8; border: 1px solid #d6e9c6; border-radius: 4px; padding: 15px; margin-bottom: 20px;">
Las compuertas cuánticas son operadores unitarios y se representan mediante matrices unitarias con respecto a una base. Usualmente se utiliza la base computacional.

Las compuertas cuánticas son los componentes básicos de los circuitos cuánticos, de manera análoga a las compuertas lógicas clásicas en los circuitos digitales convencionales. A diferencia de muchas compuertas lógicas clásicas, las compuertas cuánticas son **reversibles**.

<div align="center"><img width="400px" style="display:block; margin-left: auto; margin-right: auto;" src="https://images.claudiazm.xyz/ecc26/UnitaryReversibility.gif">

<details>
<summary><strong>Recordatorio: Matriz unitaria</strong></summary>

Una **matriz** es un arreglo rectangular de números que, entre otros usos, permite representar una transformación de vectores. Una matriz compleja cuadrada $U$ es **unitaria** cuando su inversa es igual a su transpuesta conjugada, también llamada [adjunto hermitiano](https://en.wikipedia.org/wiki/Hermitian_adjoint). Para obtener el adjunto se conjuga cada elemento, es decir, se cambia $i$ por $-i$, y después se intercambian las filas por las columnas. Se denota por $U^{\dagger}$.

La condición de unitariedad es

$$U^{\dagger}U=UU^{\dagger}=I,$$

por lo que $U^{-1}=U^{\dagger}$. Aquí, $I$ es la matriz identidad, que deja sin cambios a cualquier vector. Otra forma de entender esta condición es que las filas y las columnas de $U$ forman conjuntos ortonormales.

Las matrices unitarias conservan productos internos y normas. Por ello, si un estado cuántico tiene longitud uno antes de aplicar $U$, también la tendrá después. Así se conserva la suma de las probabilidades. Además, toda operación unitaria es reversible, pues puede deshacerse aplicando $U^{\dagger}$.

Por ejemplo, la compuerta $X$ es unitaria y satisface $X^{\dagger}X=X^2=I$.

</details>

<a id="7.1"></a>
## 7.1. Representación matemática de las compuertas

Una compuerta cuántica que actúa sobre $n$ qubits está representada por una matriz unitaria de $2^n\times 2^n$. El estado conjunto de esos qubits se representa mediante un vector unitario con $2^n$ componentes.

Las compuertas cuánticas más comunes actúan sobre uno o dos qubits, al igual que muchas compuertas lógicas clásicas actúan sobre uno o dos bits.

La acción de una compuerta sobre un estado cuántico se obtiene multiplicando la matriz $U$, que representa la compuerta, por el vector $\lvert \psi_1 \rangle$, que representa el estado. El resultado es un nuevo estado cuántico $\lvert \psi_2 \rangle$:

$$U\lvert \psi_1 \rangle=\lvert \psi_2 \rangle$$

Cuando tenemos un qubit, $n=1$, su vector de estado tiene dos componentes y una compuerta que actúa sobre él se representa mediante una matriz de $2\times 2$. El estado resultante se obtiene como sigue:

$$U\lvert \psi_1 \rangle=\begin{bmatrix}  u_{11} & u_{12} \\ u_{21} & u_{22} \end{bmatrix}\begin{bmatrix}  \alpha \\ \beta \end{bmatrix}=\begin{bmatrix}  \alpha \, u_{11} + \beta \, u_{12} \\ \alpha \, u_{21} + \beta \, u_{22} \end{bmatrix}=\begin{bmatrix}  \alpha' \\ \beta' \end{bmatrix}=\lvert \psi_2 \rangle$$

## 7.2. Compuertas de un solo qubit

<a id="7.2.1"></a>
### 7.2.1. Compuertas de Pauli

Las **compuertas de Pauli** corresponden a las tres [matrices de Pauli](https://en.wikipedia.org/wiki/Pauli_matrices) ($\sigma_x,\sigma_y,\sigma_z$) y actúan sobre un solo qubit.

Las compuertas $X$, $Y$ y $Z$ equivalen, respectivamente, a rotaciones de $\pi$ radianes ($180^{\circ}$) alrededor de los ejes $x$, $y$ y $z$ de la *esfera de Bloch*.

### Compuerta $X$

Esta compuerta es equivalente a la compuerta de negación ($NOT$) en las computadoras clásicas.

También se conoce como *bit-flip*, ya que transforma los estados de la base como sigue:

$$X\lvert 0 \rangle = \lvert 1 \rangle$$

$$X\lvert 1 \rangle = \lvert 0 \rangle$$

Su representación matricial es:

$$X=NOT=\begin{bmatrix}  0 & 1 \\ 1 & 0 \end{bmatrix}$$

<div style="color: #31708f; background-color: #d9edf7; border: 1px solid #bce8f1; border-radius: 4px; padding: 15px; margin-bottom: 20px;">
<strong>Ejemplo</strong>: Efecto de la compuerta $X$ sobre el estado $\lvert 1 \rangle$
    
$$X\lvert 1 \rangle=\begin{bmatrix}  0 & 1 \\ 1 & 0 \end{bmatrix}\begin{bmatrix}  0 \\ 1 \end{bmatrix}=\begin{bmatrix}  1 \\ 0 \end{bmatrix}=\lvert 0 \rangle$$

En Qiskit aplicamos la compuerta $X$ con el método `x()`:

````python
circ1 = QuantumCircuit(1) # circuito con 1 qubit

circ1.x(0)                # aplicar compuerta X al (único) qubit 0

circ1.draw('mpl')         # mostrar el circuito
````

### Obtener el vector de estado con Qiskit

Con la clase `Statevector` de Qiskit se puede obtener el estado de un circuito en el punto que se desee:

````python
from qiskit.quantum_info import Statevector
from qiskit.visualization import array_to_latex


circ1 = QuantumCircuit(1)    # circuito con 1 qubit

psi_0 = Statevector(circ1)   # estado justo después de crear el circuito

circ1.x(0)                   # aplicar compuerta X al (único) qubit 0

psi_1 = Statevector(circ1)   # estado después de aplicar la compuerta X

display(circ1.draw('mpl'))   # mostrar el circuito

print("👉 Estado inicial:")
print("➤ en forma de vector:")
display(array_to_latex(psi_0.data))
print("➤ en forma de ket:")
display(psi_0.draw('latex'))
print()
print("👉 Estado después de X:")
print("➤ en forma de vector:")
display(array_to_latex(psi_1.data))
print("➤ en forma de ket:")
display(psi_1.draw('latex'))
````

### Visualización en la esfera de Bloch con Qiskit

Con la función `plot_bloch_multivector()` podemos visualizar el estado de un qubit a partir de su vector de estado.

````python
from qiskit.visualization import plot_bloch_multivector


print("🔵 Estado inicial:")
display(plot_bloch_multivector(psi_0))

print("🔵 Estado después de X:")
display(plot_bloch_multivector(psi_1))
````

### Compuerta $Y$

Esta compuerta transforma los estados de la base como sigue:

$$Y\lvert 0 \rangle=i\lvert 1 \rangle$$
$$Y\lvert 1 \rangle=-i\lvert 0 \rangle$$

Su representación matricial es:

$$Y=\begin{bmatrix}  0 & -i \\ i & 0 \end{bmatrix}$$

<div style="color: #31708f; background-color: #d9edf7; border: 1px solid #bce8f1; border-radius: 4px; padding: 15px; margin-bottom: 20px;">
<strong>Ejemplo</strong>: Efecto de la compuerta $Y$ sobre el estado $\lvert + \rangle$

Con notación de Dirac:
    
$$Y\lvert + \rangle=Y \, \left(\dfrac{1}{\sqrt2}\Big(\lvert 0 \rangle+\lvert 1 \rangle\Big)\right)=\dfrac{1}{\sqrt2}\Big(Y\lvert 0 \rangle+Y\lvert 1 \rangle\Big)=\dfrac{1}{\sqrt2}\Big(i\lvert 1 \rangle-i\lvert 0 \rangle\Big)=\dfrac{-i}{\sqrt2}\Big(\lvert 0 \rangle-\lvert 1 \rangle\Big)=-i\lvert - \rangle$$

Con notación matricial:
    
$$Y\lvert + \rangle=\begin{bmatrix}  0 & -i \\ i & 0 \end{bmatrix}\begin{bmatrix}  \frac{1}{\sqrt{2}} \\ \frac{1}{\sqrt{2}} \end{bmatrix}=\begin{bmatrix}  \frac{-i}{\sqrt{2}} \\ \frac{i}{\sqrt{2}} \end{bmatrix}=-i\begin{bmatrix}  \frac{1}{\sqrt{2}} \\ \frac{-1}{\sqrt{2}} \end{bmatrix}=-i\lvert - \rangle$$

En Qiskit aplicamos la compuerta $Y$ con el método `y()`:

````python
circ2 = QuantumCircuit(1)    # circuito con 1 qubit

psi_0 = Statevector(circ2)   # estado justo después de crear el circuito

circ2.y(0)                   # aplicar compuerta Y al (único) qubit 0

psi_1 = Statevector(circ2)   # estado después de aplicar la compuerta Y

display(circ2.draw('mpl'))   # mostrar el circuito

print("👉 Estado inicial:")
print("➤ en forma de vector:")
display(array_to_latex(psi_0))
print("➤ en forma de ket:")
display(psi_0.draw('latex'))
print()
print("👉 Estado después de Y:")
print("➤ en forma de vector:")
display(array_to_latex(psi_1))
print("➤ en forma de ket:")
display(psi_1.draw('latex'))
````

````python
print("🔵 Estado inicial:")
display(plot_bloch_multivector(psi_0))

print("🔵 Estado después de Y:")
display(plot_bloch_multivector(psi_1))
````

### 7.2.2. Rotaciones con otros ángulos

Las compuertas de Pauli representan rotaciones de $\pi$ radianes, pero también podemos rotar el estado cuántico otros ángulos. Para ello se utilizan las compuertas de rotación alrededor de los ejes $x$, $y$ y $z$, definidas como sigue:

$$R_x(\phi)=\begin{bmatrix}  \cos\left(\frac{\phi}{2}\right) & -i\sin\left(\frac{\phi}{2}\right) \\ -i\sin\left(\frac{\phi}{2}\right) & \cos\left(\frac{\phi}{2}\right) \end{bmatrix}=\cos\left(\frac{\phi}{2}\right) \, I-i\sin\left(\frac{\phi}{2}\right) \, X$$

$$R_y(\phi)=\begin{bmatrix}  \cos\left(\frac{\phi}{2}\right) & -\sin\left(\frac{\phi}{2}\right) \\ \sin\left(\frac{\phi}{2}\right) & \cos\left(\frac{\phi}{2}\right) \end{bmatrix}=\cos\left(\frac{\phi}{2}\right) \, I-i\sin\left(\frac{\phi}{2}\right) \, Y$$

$$R_z(\phi)=\begin{bmatrix}  e^{-i\frac{\phi}{2}} & 0 \\ 0 & e^{i\frac{\phi}{2}} \end{bmatrix}=\cos\left(\frac{\phi}{2}\right) \, I-i\sin\left(\frac{\phi}{2}\right) \, Z$$

En Qiskit existen los métodos `rx()`, `ry()` y `rz()` para lograr las rotaciones con diferentes ángulos.

En el siguiente código rotaremos el estado cuántico primero $\pi/2$ radianes alrededor del eje $y$, lo que lo dejará sobre el eje $+x$. Después lo rotaremos $\pi/4$ radianes alrededor del eje $z$ y, finalmente, $\pi$ radianes alrededor del eje $x$. En forma matemática, expresamos estos pasos como:

$$\lvert \psi_3 \rangle=R_x\left(\pi\right) \, R_z\left(\frac{\pi}{4}\right) \, R_y\left(\frac{\pi}{2}\right) \; \lvert 0 \rangle$$

Notemos que las operaciones actúan de derecha a izquierda. Primero se aplica la matriz que está más cerca del estado $\lvert 0 \rangle$, después la siguiente matriz hacia la izquierda y, por último, la que aparece al inicio de la expresión.

Los pasos son entonces:

$\lvert \psi_0 \rangle=\lvert 0 \rangle ~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~ =\lvert 0 \rangle$

$\lvert \psi_1 \rangle= R_y\left(\frac{\pi}{2}\right) \; \lvert \psi_0 \rangle ~~~~~~~~~~~~~~ = R_y\left(\frac{\pi}{2}\right) \; \lvert 0 \rangle$

$\lvert \psi_2 \rangle= R_z\left(\frac{\pi}{4}\right) \; \lvert \psi_1 \rangle ~~~~~~~~~~~~~~ = R_z\left(\frac{\pi}{4}\right) \, R_y\left(\frac{\pi}{2}\right) \; \lvert 0 \rangle$

$\lvert \psi_3 \rangle=R_x\left(\pi\right) \; \lvert \psi_2 \rangle ~~~~~~~~~~~~~~~~ = R_x\left(\pi\right) \, R_z\left(\frac{\pi}{4}\right) \, R_y\left(\frac{\pi}{2}\right) \; \lvert 0 \rangle$

````python
import numpy as np


circ4 = QuantumCircuit(1)    # circuito con 1 qubit

psi_0 = Statevector(circ4)   # estado justo después de crear el circuito

circ4.ry(np.pi/2, 0)         # rotar pi/2 alrededor del eje y

psi_1 = Statevector(circ4)   # estado después de la primera rotación

circ4.rz(np.pi/4, 0)         # rotar pi/4 alrededor del eje z

psi_2 = Statevector(circ4)   # estado después de la segunda rotación

circ4.rx(np.pi, 0)           # rotar pi alrededor del eje x

psi_3 = Statevector(circ4)   # estado después de la tercera rotación

display(circ4.draw('mpl'))   # mostrar el circuito

print("👉 Estado inicial:")
print("➤ en forma de vector:")
display(array_to_latex(psi_0))
print("➤ en forma de ket:")
display(psi_0.draw('latex'))
print()
print("👉 Estado después rotación de pi/2 alrededor del eje y:")
print("➤ en forma de vector:")
display(array_to_latex(psi_1))
print("➤ en forma de ket:")
display(psi_1.draw('latex'))
print()
print("👉 Estado después rotación de pi/4 alrededor del eje z:")
print("➤ en forma de vector:")
display(array_to_latex(psi_2))
print("➤ en forma de ket:")
display(psi_2.draw('latex'))
print()
print("👉 Estado después rotación de pi alrededor del eje x:")
print("➤ en forma de vector:")
display(array_to_latex(psi_3))
print("➤ en forma de ket:")
display(psi_3.draw('latex'))

print()
print("🔵 Estado inicial:")
display(plot_bloch_multivector(psi_0))

print("🔵 Estado después rotación de pi/2 alrededor del eje y:")
display(plot_bloch_multivector(psi_1))

print("🔵 Estado después rotación de pi/4 alrededor del eje z:")
display(plot_bloch_multivector(psi_2))

print("🔵 Estado después rotación de pi alrededor del eje x:")
display(plot_bloch_multivector(psi_3))
````

### 7.2.3. Compuerta Hadamard

Cuando actúa sobre un estado de la base computacional $\{\lvert 0 \rangle,\lvert 1 \rangle\}$, la compuerta Hadamard crea una superposición y transforma los estados como sigue:

$$H\lvert 0 \rangle=\begin{bmatrix}  \frac{1}{\sqrt{2}} \\ \frac{1}{\sqrt{2}} \end{bmatrix}=\lvert + \rangle$$

$$H\lvert 1 \rangle=\begin{bmatrix}  \frac{1}{\sqrt{2}} \\ -\frac{1}{\sqrt{2}} \end{bmatrix}=\lvert - \rangle$$

La compuerta Hadamard también puede interpretarse como una rotación de $\pi$ radianes alrededor del eje $\frac{\left(\hat{x}+\hat{z}\right)}{\sqrt{2}}$, situado en el plano $xz$ a $45^{\circ}$ de los ejes $x$ y $z$.

<p style="font-size:10px;font-style:italic;text-align:right"><a href="https://physics.stackexchange.com/questions/313959/visual-interpretation-on-the-bloch-sphere-when-hadamard-gate-is-applied-twice" target="_blank">Imagen obtenida de Physics StackExchange</a></p>

Su representación matricial es:

$$H=\frac{1}{\sqrt{2}} \begin{bmatrix} 1 & 1 \\ 1 & -1\end{bmatrix} = \begin{bmatrix} \frac{1}{\sqrt{2}} & \frac{1}{\sqrt{2}} \\ \frac{1}{\sqrt{2}} & -\frac{1}{\sqrt{2}}\end{bmatrix}$$

<div style="color: #31708f; background-color: #d9edf7; border: 1px solid #bce8f1; border-radius: 4px; padding: 15px; margin-bottom: 20px;">
<strong>Ejemplo</strong>: Efecto de la compuerta $H$ sobre el estado $\lvert - \rangle$

Con notación de Dirac:
    
$$\begin{aligned}
H\lvert - \rangle&=H \, \left(\dfrac{1}{\sqrt2}\Big(\lvert 0 \rangle-\lvert 1 \rangle\Big)\right)=\dfrac{1}{\sqrt2}\Big(H\lvert 0 \rangle-H\lvert 1 \rangle\Big)\\
&=\dfrac{1}{\sqrt2}\left[\left(\dfrac{1}{\sqrt2}\Big(\lvert 0 \rangle+\lvert 1 \rangle\Big)\right)-\left(\dfrac{1}{\sqrt2}\Big(\lvert 0 \rangle-\lvert 1 \rangle\Big)\right)\right]\\
&=\dfrac{1}{2}\Big(\lvert 0 \rangle+\lvert 1 \rangle-\lvert 0 \rangle+\lvert 1 \rangle\Big)=\dfrac{1}{2}\Big(2\lvert 1 \rangle\Big)=\lvert 1 \rangle
\end{aligned}$$
    
Con notación matricial:
    
$$H\lvert - \rangle=\begin{bmatrix} \frac{1}{\sqrt{2}} & \frac{1}{\sqrt{2}} \\ \frac{1}{\sqrt{2}} & -\frac{1}{\sqrt{2}}\end{bmatrix}\begin{bmatrix}  \frac{1}{\sqrt{2}} \\ -\frac{1}{\sqrt{2}} \end{bmatrix}=\frac{1}{2}\begin{bmatrix}  1-1 \\ 1+1 \end{bmatrix}=\frac{1}{2}\begin{bmatrix}  0 \\ 2 \end{bmatrix}=\begin{bmatrix}  0 \\ 1 \end{bmatrix}=\lvert 1 \rangle$$

En Qiskit aplicamos la compuerta $H$ con el método `h()`:

````python
circ5 = QuantumCircuit(1)    # circuito con 1 qubit

psi_0 = Statevector(circ5)   # estado justo después de crear el circuito

circ5.h(0)                   # apliquemos H al estado |0>, quedamos en |+>

psi_1 = Statevector(circ5)   # estado justo después de aplicar H

circ5.z(0)                   # aplicar compuerta Z al (único) qubit 0, quedamos en el estado |->

psi_2 = Statevector(circ5)   # estado después de aplicar la compuerta Z

circ5.h(0)                   # aplicar compuerta H de nuevo, quedamos en el estado |1>

psi_3 = Statevector(circ5)   # estado después de aplicar la segunda compuerta H

display(circ5.draw('mpl'))   # mostrar el circuito

print("👉 Estado inicial:")
print("➤ en forma de vector:")
display(array_to_latex(psi_0))
print("➤ en forma de ket:")
display(psi_0.draw('latex'))
print()
print("👉 Estado después de primera H:")
print("➤ en forma de vector:")
display(array_to_latex(psi_1))
print("➤ en forma de ket:")
display(psi_1.draw('latex'))
print()
print("👉 Estado después de Z:")
print("➤ en forma de vector:")
display(array_to_latex(psi_2))
print("➤ en forma de ket:")
display(psi_2.draw('latex'))
print()
print("👉 Estado después de segunda H:")
print("➤ en forma de vector:")
display(array_to_latex(psi_3))
print("➤ en forma de ket:")
display(psi_3.draw('latex'))

print()
print("🔵 Estado inicial:")
display(plot_bloch_multivector(psi_0))

print("🔵 Estado después de primera H:")
display(plot_bloch_multivector(psi_1))

print("🔵 Estado después de Z:")
display(plot_bloch_multivector(psi_2))

print("🔵 Estado después de segunda H:")
display(plot_bloch_multivector(psi_3))
````

### 7.2.4. Compuertas de cambio de fase

Las compuertas de cambio de fase forman una familia de compuertas de un solo qubit que transforman los estados de la base como sigue:

$$P(\phi) \, \lvert 0 \rangle = \lvert 0 \rangle$$

$$P(\phi) \, \lvert 1 \rangle = e^{i\phi}\lvert 1 \rangle$$

La probabilidad de medir $\lvert 0 \rangle$ o $\lvert 1 \rangle$ no cambia después de aplicar esta compuerta. Sin embargo, sí cambia la fase relativa del estado cuántico. En la esfera de Bloch, esto equivale a una rotación de $\phi$ radianes alrededor del eje $z$, por lo que el estado recorre un círculo horizontal o línea de latitud. La compuerta de cambio de fase está representada por la matriz:

$$P(\phi)=\begin{bmatrix} 1 & 0 \\ 0 & e^{i\phi}\end{bmatrix}$$

donde $\phi$ es el cambio de fase y tiene periodo $2\pi$.

Notemos que la compuerta $P(\phi)$ no es [hermitiana](https://en.wikipedia.org/wiki/Hermitian_matrix), excepto cuando $\phi=n\pi$ con $n\in\mathbb{Z}$.

Algunos valores particulares de $\phi$ corresponden a compuertas con nombre propio.

**Compuertas $S$ y $S^{\dagger}$**

La compuerta $S$ produce una rotación de $\pi/2$ radianes alrededor del eje $z$. Por lo tanto,

$$S=\sqrt{Z}=P\left(\frac{\pi}{2}\right)$$

La compuerta $S^{\dagger}$ es el adjunto y la inversa de la compuerta $S$. Por lo tanto,

$$S^{\dagger}=P\left(-\frac{\pi}{2}\right)$$

Sus matrices son:

$$S=\begin{bmatrix} 1 & 0 \\ 0 & i\end{bmatrix} ~~~~~~~~~~~~~~~~~~~~~~~~~~~ S^{\dagger}=\begin{bmatrix} 1 & 0 \\ 0 & -i\end{bmatrix}$$

**Compuertas $T$ y $T^{\dagger}$**

La compuerta $T$ produce una rotación de $\pi/4$ radianes alrededor del eje $z$. Por lo tanto,

$$T=\sqrt{S}=\sqrt[4]{Z}=P\left(\frac{\pi}{4}\right)$$

$$T^{\dagger}=P\left(-\frac{\pi}{4}\right)$$

Sus matrices son:

$$T=\begin{bmatrix} 1 & 0 \\ 0 & e^{i\frac{\pi}{4}}\end{bmatrix} ~~~~~~~~~~~~~~~~~~~~~~~~~~~ T^{\dagger}=\begin{bmatrix} 1 & 0 \\ 0 & e^{-i\frac{\pi}{4}}\end{bmatrix}$$

En Qiskit tenemos a nuestra disposición métodos para aplicar las compuertas de cambio de fase: `p()`, `s()`, `sdg()`, `t()`, `tdg()`.

Veamos el efecto de estas compuertas. Como representan rotaciones alrededor del eje $z$, partiremos del estado $\lvert + \rangle$ para que las rotaciones sean visibles.

````python
circ6 = QuantumCircuit(1)    # circuito con 1 qubit

circ6.h(0)                   # nos movemos al estado |+>

psi_0 = Statevector(circ6)   # estado después de preparar |+>

circ6.s(0)                   # aplicar compuerta S, rotando pi/2

psi_1 = Statevector(circ6)   # estado después de aplicar S

circ6.tdg(0)                 # aplicar compuerta T†, rotando -pi/4

psi_2 = Statevector(circ6)   # estado después de aplicar T†

circ6.p(2*np.pi/3, 0)        # aplicar compuerta P, rotando 2pi/3

psi_3 = Statevector(circ6)   # estado después de aplicar P

display(circ6.draw('mpl'))   # mostrar el circuito

print("👉 Estado inicial |+>:")
print("➤ en forma de vector:")
display(array_to_latex(psi_0))
print("➤ en forma de ket:")
display(psi_0.draw('latex'))
print()
print("👉 Estado después de S:")
print("➤ en forma de vector:")
display(array_to_latex(psi_1))
print("➤ en forma de ket:")
display(psi_1.draw('latex'))
print()
print("👉 Estado después de T†:")
print("➤ en forma de vector:")
display(array_to_latex(psi_2))
print("➤ en forma de ket:")
display(psi_2.draw('latex'))
print()
print("👉 Estado después de P(2*pi/3):")
print("➤ en forma de vector:")
display(array_to_latex(psi_3))
print("➤ en forma de ket:")
display(psi_3.draw('latex'))

print()
print("🔵 Estado inicial |+>:")
display(plot_bloch_multivector(psi_0))

print("🔵 Estado después de S:")
display(plot_bloch_multivector(psi_1))

print("🔵 Estado después de T†:")
display(plot_bloch_multivector(psi_2))

print("🔵 Estado después de P(2*pi/3):")
display(plot_bloch_multivector(psi_3))
````

### 7.2.5. Compuerta más general de un qubit

La compuerta general de un solo qubit se conoce como $U$ y está definida por:

$$U(\theta, \phi,\lambda)=\begin{bmatrix} \cos\left(\frac{\theta}{2}\right) & -e^{i\lambda}\sin\left(\frac{\theta}{2}\right) \\ e^{i\phi}\sin\left(\frac{\theta}{2}\right) & e^{i\left(\phi+\lambda\right)}\cos\left(\frac{\theta}{2}\right)\end{bmatrix}$$

En referencias anteriores también puede aparecer con el nombre $U3$.

Algunos ejemplos de equivalencias son:

$$U\left(\theta, -\frac{\pi}{2}, \frac{\pi}{2}\right)=R_x(\theta)$$

$$U\left(\theta, 0, 0\right)=R_y(\theta)$$

En Qiskit podemos usar el método `u()` para aplicar esta compuerta:

````python
circ_u = QuantumCircuit(1)               # circuito con 1 qubit

psi_0 = Statevector(circ_u)              # estado inicial |0>

circ_u.u(np.pi/3, np.pi/2, np.pi/5, 0)   # aplicar compuerta U al qubit 0

psi_1 = Statevector(circ_u)              # estado después de aplicar la compuerta U

display(circ_u.draw('mpl'))              # mostrar el circuito

print("👉 Estado inicial:")
print("➤ en forma de vector:")
display(array_to_latex(psi_0))
print("➤ en forma de ket:")
display(psi_0.draw('latex'))
print()
print("👉 Estado después de U(pi/3, pi/2, pi/5):")
print("➤ en forma de vector:")
display(array_to_latex(psi_1))
print("➤ en forma de ket:")
display(psi_1.draw('latex'))

print()
print("🔵 Estado inicial:")
display(plot_bloch_multivector(psi_0))

print("🔵 Estado después de U(pi/3, pi/2, pi/5):")
display(plot_bloch_multivector(psi_1))
````
