# 1. ¿Qué significa que una aplicación sea real?

Una aplicación puede ser real aunque todavía no genere una ventaja económica o científica frente a las mejores computadoras clásicas. Por ejemplo, un problema industrial puede formularse y ejecutarse en hardware cuántico real, pero con una instancia pequeña, simplificada o apoyada por una computadora clásica.

Para interpretar correctamente los ejemplos de este notebook conviene distinguir los siguientes niveles:

| Nivel | ¿Qué se demostró? | ¿Qué no implica? |
|---|---|---|
| **Propuesta o estimación de recursos** | Existe un algoritmo y se calculó qué hardware necesitaría. | Que el cálculo ya pueda ejecutarse. |
| **Simulación clásica** | El algoritmo cuántico se probó con un simulador convencional. | Que se haya utilizado una computadora cuántica. |
| **Demostración en hardware** | Un circuito relacionado con la aplicación se ejecutó en una QPU real. | Que el tamaño del problema sea útil o que exista ventaja cuántica. |
| **Piloto** | El método se integró en un flujo real o utilizó datos reales. | Que esté listo para producción o sea mejor que todas las alternativas clásicas. |
| **Ventaja cuántica práctica** | La solución cuántica ofrece una mejora relevante y verificable frente al mejor método clásico conocido. | Que esa ventaja se extienda a cualquier problema. |

En particular, **cuántico**, **híbrido cuántico-clásico** y **cuántico-inspirado** no significan lo mismo. Un método cuántico-inspirado se ejecuta completamente en hardware clásico, aunque utilice ideas matemáticas provenientes de la teoría cuántica.

<div style="color: #31708f; background-color: #d9edf7; border: 1px solid #bce8f1; border-radius: 4px; padding: 15px; margin-bottom: 20px;">
<b>Idea principal:</b> que un experimento use datos reales o hardware cuántico real no basta para afirmar que existe una ventaja cuántica práctica. Siempre hay que preguntar qué parte se ejecutó en la QPU, cuál fue el tamaño del problema y contra qué métodos clásicos se comparó.

# 2. ¿Qué puede hacerse actualmente?

Las computadoras cuánticas actuales permiten investigar algoritmos, ejecutar circuitos de cientos o miles de compuertas según el dispositivo, estudiar modelos físicos y probar flujos híbridos. También permiten aprender a programar hardware real mediante servicios en la nube.

**Actualmente sí es posible:**

- ejecutar demostraciones de algoritmos en procesadores cuánticos reales;
- estudiar moléculas pequeñas, espacios activos reducidos y modelos de materiales;
- resolver instancias acotadas de optimización y probar sistemas híbridos;
- entrenar o evaluar modelos pequeños de aprendizaje automático cuántico;
- simular dinámicas cuánticas y medir observables de interés científico;
- estimar los recursos necesarios para futuras computadoras tolerantes a fallos.

**Actualmente no es posible, de manera general:**

- reemplazar una computadora personal, un centro de datos o una GPU;
- acelerar cualquier programa por el solo hecho de ejecutarlo en una QPU;
- resolver a escala industrial todos los problemas de rutas, horarios o portafolios;
- simular con precisión arbitraria proteínas, fármacos o materiales completos;
- romper RSA-2048 o la criptografía usada cotidianamente en Internet;
- superar sistemáticamente al aprendizaje automático clásico con datos clásicos.

Los dispositivos ruidosos tienen límites importantes de profundidad, fidelidad y conectividad. Un análisis de 2025 estudia formalmente algunas de estas [limitaciones de los dispositivos cuánticos ruidosos](https://www.nature.com/articles/s41534-025-01136-4). Por ello, las computadoras cuánticas actuales se utilizan normalmente como **aceleradores especializados dentro de flujos híbridos**, no como sustitutos completos de las computadoras clásicas.

<div style="color: #3c763d; background-color: #dff0d8; border: 1px solid #d6e9c6; border-radius: 4px; padding: 15px; margin-bottom: 20px;">
<b>Una expectativa razonable:</b> hoy el valor principal está en aprender, experimentar, crear métodos y determinar en qué problemas podría aparecer una ventaja útil. Algunos experimentos ya son científicamente interesantes, pero todavía no existe una ventaja práctica general para las aplicaciones descritas a continuación.
</div>

# 3. Optimización

La optimización consiste en elegir una buena solución entre muchas posibilidades, normalmente bajo restricciones. Aparece en rutas, logística, horarios, manufactura, redes, energía y finanzas. Parte de estos problemas puede expresarse como un modelo de Ising o como un problema QUBO, formatos compatibles con recocido cuántico y con algunos algoritmos de compuertas.

### Tráfico urbano con Volkswagen y D-Wave

<div align="center"><img width="500px" style="display:block; margin-left: auto; margin-right: auto;" alt="Autobús del piloto de optimización de tráfico de Volkswagen" src="https://uploads.vw-mms.de/system/production/images/vwn/030/981/images/e6350e21896169a87eefac4e28468b0d69897661/DB2019AL02384_web_1600.jpg?1649155636"></div>

En 2017, investigadores de Volkswagen formularon la distribución de **418 vehículos** sobre rutas alternativas de Pekín y ejecutaron el problema en un recocedor D-Wave. El [artículo científico](https://doi.org/10.3389/fict.2017.00029) documenta la formulación y los experimentos. En 2019, Volkswagen llevó la idea a un piloto operativo durante Web Summit en Lisboa: nueve autobuses utilizaron un sistema híbrido para recomendar rutas casi en tiempo real. El [comunicado oficial de Volkswagen](https://www.volkswagen-group.com/en/press-releases/volkswagen-optimizes-traffic-flow-with-quantum-computers-16995) describe el proyecto.

Este caso es importante porque la QPU formó parte de un servicio con vehículos reales. Sin embargo, fue un piloto híbrido y no una demostración de que el procesador cuántico superara a los mejores optimizadores clásicos. Un trabajo posterior de 2025 continuó investigando el problema mediante [QUBO iterativos para tráfico](https://www.nature.com/articles/s41598-025-04568-2).

### Asignación de turnos de enfermería

<div align="center"><img width="500px" style="display:block; margin-left: auto; margin-right: auto;" alt="Flujo para formular la asignación de turnos de enfermería en un recocedor cuántico" src="https://media.springernature.com/full/springer-static/image/art%3A10.1038%2Fs41598-019-49172-3/MediaObjects/41598_2019_49172_Fig1_HTML.png"></div>

El problema de asignar turnos debe satisfacer necesidades de personal y restricciones laborales. En 2019 se ejecutaron instancias de este problema en un D-Wave 2000Q y se compararon diferentes formulaciones. El [artículo en Scientific Reports](https://www.nature.com/articles/s41598-019-49172-3) es una demostración en hardware de un problema realista, aunque sus instancias y restricciones siguen siendo mucho menores que las de un hospital completo.

Una introducción accesible a estos ejemplos puede consultarse en el artículo [*En las Fronteras de la Información Cuántica: Aprendizaje Automático y Optimización Cuántica*](https://quantum2025.org/es/news-link/en-las-fronteras-de-la-informacion-cuantica-aprendizaje-automatico-y-optimizacion-cuantica/), publicado por el Año Internacional de la Ciencia y la Tecnología Cuánticas.

**Estado actual:** existen formulaciones, demostraciones en QPU y pilotos híbridos con datos reales. No se ha establecido una ventaja práctica general frente a los mejores solvers clásicos para logística, tráfico o planificación industrial.

<a id="4"></a>
# 4. Química

Las propiedades químicas dependen de la estructura electrónica de las moléculas, que es cuántica. Esta correspondencia hace que la química sea una de las aplicaciones más naturales de la computación cuántica. Una meta frecuente es estimar energías de estado base, curvas de disociación y energías de reacción.

### Energías moleculares con VQE

<div align="center"><img width="500px" style="display:block; margin-left: auto; margin-right: auto;" alt="Química cuántica en un procesador superconductivo" src="https://media.springernature.com/full/springer-static/image/art%3A10.1038%2Fnature23879/MediaObjects/41586_2017_Article_BFnature23879_Fig1_HTML.jpg"></div>

En 2017, un equipo de IBM utilizó un procesador superconductivo para estimar energías de estado base de moléculas como hidruro de litio, $\mathrm{LiH}$, e hidruro de berilio, $\mathrm{BeH_2}$. El experimento introdujo una versión de VQE adaptada al hardware y se publicó en [*Nature*](https://www.nature.com/articles/nature23879). Fue una demostración fundamental, pero utilizó moléculas pequeñas, bases reducidas y hasta seis qubits.

En 2024 se publicó una implementación experimental de VQE con un ansatz de *coupled cluster* optimizado, circuitos de hasta **12 qubits** y técnicas de mitigación de errores. El trabajo obtuvo energías de alta precisión para varios modelos moleculares y puede consultarse en [*Nature Physics*](https://www.nature.com/articles/s41567-024-02530-z).

### Ruptura del enlace del nitrógeno

<div align="center"><img width="400px" style="display:block; margin-left: auto; margin-right: auto;" alt="Flujo del cálculo cuántico de la curva de disociación del nitrógeno" src="https://media.springernature.com/full/springer-static/image/art%3A10.1038%2Fs41534-024-00952-4/MediaObjects/41534_2024_952_Fig1_HTML.png"></div>

La disociación de $\mathrm{N_2}$ es difícil para métodos de referencia única porque aparecen correlaciones electrónicas fuertes al separar los átomos. Un experimento publicado en 2025 calculó su curva de disociación en hardware superconductivo mediante un espacio contextual de cinco qubits. Los resultados capturaron mejor la ruptura del enlace que los métodos de referencia única comparados, aunque el problema completo había sido reducido a un espacio activo pequeño. Véase el [artículo en *npj Quantum Information*](https://www.nature.com/articles/s41534-024-00952-4).

**Estado actual:** las QPU ya reproducen propiedades de modelos moleculares pequeños y permiten probar nuevos algoritmos. Para moléculas relevantes suele ser necesario congelar orbitales, reducir el espacio activo y combinar la QPU con cálculos clásicos. Esto todavía no reemplaza a los paquetes clásicos de química computacional.

<a id="5"></a>
# 5. Materiales

El diseño de materiales requiere comprender fenómenos colectivos, electrones fuertemente correlacionados, magnetismo, fases topológicas y respuesta a campos externos. Algunos de estos problemas son difíciles para métodos clásicos incluso cuando se utiliza un modelo efectivo.

### Funciones de respuesta de una onda de densidad de carga

<div align="center"><img width="400px" style="display:block; margin-left: auto; margin-right: auto;" alt="Método de respuesta lineal para obtener funciones de correlación" src="https://media.springernature.com/full/springer-static/image/art%3A10.1038%2Fs41467-024-47729-z/MediaObjects/41467_2024_47729_Fig1_HTML.png"></div>

En 2024 se ejecutó en el procesador superconductivo `ibm_auckland` un método de respuesta lineal para obtener funciones de Green y espectros relacionados con mediciones experimentales. La demostración estudió un modelo de material con onda de densidad de carga y el modelo de Hubbard unidimensional. El [artículo en *Nature Communications*](https://www.nature.com/articles/s41467-024-47729-z) muestra una conexión concreta entre cantidades calculadas en la QPU y técnicas como espectroscopía fotoelectrónica.

### Moléculas y materiales con arreglos de átomos neutros

<div align="center"><img width="500px" style="display:block; margin-left: auto; margin-right: auto;" alt="Simulación programable de moléculas y materiales con átomos neutros" src="https://media.springernature.com/full/springer-static/image/art%3A10.1038%2Fs41567-024-02738-z/MediaObjects/41567_2024_2738_Fig1_HTML.png"></div>

Procesadores analógicos de átomos neutros pueden implementar directamente ciertos Hamiltonianos de espines. En 2025, investigadores de Harvard, QuEra y otras instituciones publicaron simulaciones programables de modelos de moléculas y materiales en arreglos reconfigurables. El [artículo en *Nature Physics*](https://www.nature.com/articles/s41567-024-02738-z) incluyó dinámica de sistemas fuertemente correlacionados y estimaciones de propiedades asociadas con estructuras moleculares.

También se desarrollan algoritmos que reducen recursos para materiales específicos. Por ejemplo, un trabajo de 2024 redujo considerablemente la profundidad estimada para modelos como el óxido de vanadio y estroncio, $\mathrm{SrVO_3}$, pero se trató de diseño algorítmico y estimación de recursos, no de una simulación completa en hardware. Véase [*Towards near-term quantum simulation of materials*](https://www.nature.com/articles/s41467-023-43479-6).

**Estado actual:** se estudian modelos efectivos y observables relevantes, pero todavía no se diseña de extremo a extremo un material industrial superior utilizando solamente una computadora cuántica. Las demostraciones reducen el sistema físico a los grados de libertad que el hardware puede representar.

<a id="6"></a>
# 6. Fármacos

El descubrimiento de fármacos reúne varios problemas: generación y filtrado de moléculas, cálculo de afinidades, química de enlaces, dinámica molecular y optimización de propiedades. Por ello, los proyectos actuales suelen ser híbridos y utilizan la QPU solamente en una parte del flujo.

### Activación de profármacos e interacción con KRAS

<div align="center"><img width="500px" style="display:block; margin-left: auto; margin-right: auto;" alt="Tubería híbrida de computación cuántica para descubrimiento de fármacos" src="https://media.springernature.com/full/springer-static/image/art%3A10.1038%2Fs41598-024-67897-8/MediaObjects/41598_2024_67897_Fig1_HTML.png"></div>

En 2024 se presentó una tubería híbrida para dos problemas concretos: calcular el perfil de energía libre para activar un profármaco basado en $\beta$-lapachona y modelar la interacción covalente entre Sotorasib y la proteína KRAS(G12C). Los autores ejecutaron circuitos de dos qubits en un dispositivo superconductivo y combinaron sus resultados con dinámica molecular y química clásica. El [artículo en *Scientific Reports*](https://www.nature.com/articles/s41598-024-67897-8) muestra que el flujo puede aplicarse a casos biomédicos reales, pero también aclara que la precisión y los recursos cuánticos deben mejorar.

### Generación y validación experimental de inhibidores de KRAS

<div align="center"><img width="400px" style="display:block; margin-left: auto; margin-right: auto;" alt="Flujo híbrido para generar y validar inhibidores de KRAS" src="https://media.springernature.com/full/springer-static/image/art%3A10.1038%2Fs41587-024-02526-3/MediaObjects/41587_2024_2526_Fig1_HTML.png"></div>

Un trabajo publicado en 2025 combinó una red LSTM clásica con una máquina de Born cuántica para proponer pequeñas moléculas dirigidas a KRAS. Se sintetizaron y probaron **15 candidatos**; dos mostraron potencial para desarrollo posterior. El estudio incluyó muestras producidas con hardware cuántico y validación experimental mediante ensayos bioquímicos y celulares. Véase [*Quantum-computing-enhanced algorithm unveils potential KRAS inhibitors*](https://www.nature.com/articles/s41587-024-02526-3), publicado en *Nature Biotechnology*.

Este es un resultado tangible, pero no demuestra que la parte cuántica fuera indispensable ni que el flujo supere de forma general a los mejores modelos generativos clásicos. Tampoco significa que se haya producido un medicamento aprobado: un compuesto inicial debe superar muchas etapas preclínicas y clínicas.

**Estado actual:** ya existen flujos híbridos aplicados a objetivos farmacológicos reales e incluso candidatos sintetizados y probados. Aún no existe un fármaco aprobado cuyo descubrimiento dependa de una ventaja cuántica comprobada.

# 7. Criptografía

La relación más conocida entre computación cuántica y criptografía es el algoritmo de Shor. Una computadora cuántica tolerante a fallos suficientemente grande podría factorizar enteros y calcular logaritmos discretos eficientemente, lo que afectaría RSA y los esquemas de curvas elípticas.

### Lo que se ha factorizado en hardware

<div align="center"><img width="500px" style="display:block; margin-left: auto; margin-right: auto;" alt="Circuito compilado del algoritmo de Shor para factorizar 21" src="https://media.springernature.com/full/springer-static/image/art%3A10.1038%2Fs41598-021-95973-w/MediaObjects/41598_2021_95973_Fig1_HTML.png"></div>

En 2021 se ejecutó una versión compilada de Shor en procesadores de IBM para factorizar **21** con cinco qubits. El [artículo en *Scientific Reports*](https://www.nature.com/articles/s41598-021-95973-w) verificó además entrelazamiento entre los registros. La compilación aprovecha información específica del número y reduce mucho el circuito, por lo que no equivale a una implementación escalable capaz de atacar claves reales.

RSA-2048 sigue fuera del alcance del hardware actual. Las estimaciones dependen de la arquitectura y mejoran con nuevos algoritmos y códigos, pero requieren computación cuántica con corrección de errores a una escala que todavía no existe. Como referencia histórica, una estimación publicada en 2021 calculó [20 millones de qubits físicos y unas ocho horas](https://quantum-journal.org/papers/q-2021-04-15-433/) bajo supuestos específicos.

### Una aplicación que sí está ocurriendo: migrar la criptografía

<div align="center"><img width="400px" style="display:block; margin-left: auto; margin-right: auto;" alt="Ilustración de criptografía poscuántica de NIST" src="https://www.nist.gov/sites/default/files/styles/social/public/images/2022/04/05/Crypto-rev1.png?itok=n-O8Vqo3"></div>

La consecuencia práctica actual es defensiva. En 2024, NIST publicó los primeros estándares de criptografía poscuántica para proteger información frente a computadoras clásicas y futuras computadoras cuánticas. NIST explica el riesgo, incluido el escenario de guardar datos cifrados hoy para descifrarlos en el futuro, en [*What Is Post-Quantum Cryptography?*](https://www.nist.gov/cybersecurity-and-privacy/what-post-quantum-cryptography). En 2025 seleccionó además [HQC como algoritmo de respaldo](https://www.nist.gov/news-events/news/2025/03/nist-selects-hqc-fifth-algorithm-post-quantum-encryption).

La criptografía poscuántica se ejecuta en computadoras clásicas. Se llama así porque está diseñada para resistir ataques cuánticos, no porque requiera una QPU.

**Estado actual:** Shor se ha demostrado solamente en números pequeños y con circuitos muy simplificados. Las QPU actuales no rompen la criptografía moderna, pero el riesgo futuro ya motivó estándares y migraciones reales hacia criptografía poscuántica.

<a id="8"></a>
# 8. Aprendizaje automático

El aprendizaje automático cuántico, o **QML**, investiga modelos que utilizan circuitos cuánticos para representar datos, calcular kernels, generar muestras o aprender propiedades de sistemas cuánticos. Cuando los datos son clásicos hay un costo adicional: deben codificarse en un estado cuántico.

### Clasificación de eventos del LHC

<div align="center"><img width="500px" style="display:block; margin-left: auto; margin-right: auto;" alt="Máquinas de soporte vectorial cuánticas para clasificación de eventos del Higgs" src="https://quantum.cern/sites/default/files/inline-images/Quantum%20support%20vector%20machines.png"></div>

CERN, Fermilab y otras instituciones probaron un clasificador variacional para distinguir señal y fondo en dos análisis del bosón de Higgs. El modelo se ejecutó con **10 qubits** en hardware de IBM y trabajó con muestras pequeñas de eventos realistas. Su desempeño fue comparable al simulador y a clasificadores clásicos en esas instancias, no superior. El [registro y artículo en CERN](https://cds.cern.ch/record/2748292) documenta el experimento y la publicación revisada por pares.

### Clasificación de imágenes médicas

<div align="center"><img width="500px" style="display:block; margin-left: auto; margin-right: auto;" alt="Flujo de clasificación de MedMNIST en hardware cuántico real" src="https://media.springernature.com/full/springer-static/image/art%3A10.1038%2Fs41598-026-35605-3/MediaObjects/41598_2026_35605_Fig1_HTML.png"></div>

En 2026 se evaluaron modelos cuánticos con conjuntos MedMNIST en un procesador IBM de 127 qubits. El trabajo estudió compilación, supresión y mitigación de errores en hardware real para clasificación de imágenes médicas. El [artículo en *Scientific Reports*](https://www.nature.com/articles/s41598-026-35605-3) constituye un punto de referencia experimental, no un sistema clínico ni una superioridad demostrada frente al aprendizaje profundo clásico.

Los modelos de QML también enfrentan dificultades de entrenamiento, ruido, escalabilidad y comparación justa. Una discusión reciente puede consultarse en [*Pitfalls and prospects of quantum machine learning*](https://www.nature.com/articles/s43588-025-00914-6).

**Estado actual:** QML puede ejecutarse en hardware real y se prueba con datos científicos y médicos. Para datos clásicos, los ejemplos suelen ser pequeños y no han mostrado una ventaja general sobre GPU, redes neuronales, árboles o métodos de kernel clásicos. Una posible ventaja podría aparecer antes al aprender directamente de datos cuánticos.

<a id="9"></a>
# 9. Simulación

Una computadora cuántica puede representar directamente la evolución de otro sistema cuántico. Esta fue una de las motivaciones originales propuestas por Richard Feynman y abarca química, materia condensada, física nuclear, altas energías y sistemas abiertos.

### Dinámica de un modelo de Ising con 127 qubits

<div align="center"><img width="500px" style="display:block; margin-left: auto; margin-right: auto;" alt="Circuitos de evolución temporal del modelo de Ising en 127 qubits" src="https://media.springernature.com/full/springer-static/image/art%3A10.1038%2Fs41586-023-06096-3/MediaObjects/41586_2023_6096_Fig1_HTML.png"></div>

En 2023, IBM ejecutó la evolución temporal de un modelo de Ising bidimensional con **127 qubits**, hasta 60 capas de compuertas de dos qubits y 2 880 CNOT. El trabajo utilizó mitigación de errores y comparó los resultados con métodos clásicos aproximados. El [artículo en *Nature*](https://www.nature.com/articles/s41586-023-06096-3) presentó evidencia de utilidad antes de la tolerancia a fallos, no una aceleración probada para un problema comercial. Después de su publicación, mejores simulaciones clásicas reprodujeron partes importantes del experimento, lo que ilustra que la referencia clásica también avanza.

### Sistemas cuánticos abiertos

<div align="center"><img width="600px" style="display:block; margin-left: auto; margin-right: auto;" alt="Simulación de larga duración de un sistema cuántico abierto" src="https://media.springernature.com/full/springer-static/image/art%3A10.1038%2Fs41534-025-00964-8/MediaObjects/41534_2025_964_Fig1_HTML.png"></div>

Los sistemas reales interactúan con su ambiente. En 2025 se demostró una simulación de larga duración de un sistema abierto con circuitos de hasta **dos mil compuertas entrelazantes**, aprovechando y mitigando características del ruido del procesador. Véase el [artículo en *npj Quantum Information*](https://www.nature.com/articles/s41534-025-00964-8).

### Teorías de norma en física de altas energías

<div align="center"><img width="500px" style="display:block; margin-left: auto; margin-right: auto;" alt="Simulación de electrodinámica cuántica bidimensional con qudits" src="https://media.springernature.com/full/springer-static/image/art%3A10.1038%2Fs41567-025-02797-w/MediaObjects/41567_2025_2797_Fig1_HTML.png"></div>

En 2025, un procesador de iones atrapados basado en qudits simuló una teoría de norma en dos dimensiones con materia y campos electromagnéticos. Esta es una demostración de investigación sobre física fundamental. El resultado se publicó como [*Simulating two-dimensional lattice gauge theories on a qudit quantum computer*](https://www.nature.com/articles/s41567-025-02797-w).

**Estado actual:** la simulación cuántica es una de las áreas con experimentos de mayor escala y valor científico. Aun así, demostrar que un circuito es difícil de simular clásicamente no basta: el resultado debe ser preciso, verificable y responder una pregunta científica que no pueda resolverse mejor por otros métodos.

# 10. Conclusiones

Las aplicaciones cuánticas reales de hoy son principalmente **experimentos científicos, demostraciones en hardware y pilotos híbridos**. Esto no las hace irrelevantes: permiten descubrir qué algoritmos toleran el ruido, cómo dividir el trabajo entre CPU, GPU y QPU, y qué recursos serán necesarios para problemas más grandes.

| Área | Lo que ya existe | Lo que falta |
|---|---|---|
| Optimización | Instancias en QPU y pilotos de tráfico, horarios y logística. | Ventaja consistente a escala industrial. |
| Química | Energías y curvas de moléculas en espacios reducidos. | Superar a los mejores métodos clásicos en moléculas de interés. |
| Materiales | Modelos de correlación, dinámica y funciones de respuesta. | Predicción o diseño industrial con ventaja comprobada. |
| Fármacos | Flujos híbridos y candidatos con validación experimental inicial. | Un beneficio atribuible a la QPU durante el desarrollo de un fármaco. |
| Criptografía | Demostraciones pequeñas de Shor y migración poscuántica. | Una QPU tolerante a fallos capaz de atacar claves reales. |
| Aprendizaje automático | Clasificadores y generadores pequeños en hardware. | Ventaja general con datos clásicos y a gran escala. |
| Simulación | Experimentos de muchos cuerpos y modelos de física fundamental. | Precisión y escala para responder problemas inaccesibles clásicamente. |

La pregunta útil no es solamente **«¿se usó una computadora cuántica?»**, sino **«¿qué problema se resolvió, con qué aproximaciones, contra qué referencia clásica y qué parte del resultado dependió realmente de la QPU?»**
