# 1. Introducción


Los circuitos estudiados hasta ahora, como sumadores, teleportación y codificación superdensa, no son ejercicios aislados. Representan dos líneas de investigación distintas:

* La **aritmética reversible** proporciona bloques con los que se construyen algoritmos cuánticos mayores.
* La **comunicación cuántica** busca conectar sistemas cuánticos para distribuir estados, entrelazamiento y claves, así como para realizar tareas entre nodos separados.

Para la primera parte se presentan tres ejemplos de aritmética, para mostrar su función de servir como puente hacia algoritmos más grandes. La segunda parte, que habla de las redes cuánticas, reúnen física, ingeniería, teoría de la información y ciencias de la computación mostrando cómo existe una agenda experimental muy activa.


<div style="color:#31708f; background-color:#d9edf7; border:1px solid #bce8f1; border-radius:4px; padding:15px; margin-bottom:20px;">
<b>Cómo interpretar los ejemplos:</b> un experimento puede ser científicamente importante sin constituir todavía un producto comercial o Internet cuántico completo. En cada caso conviene preguntar qué se conectó, a qué distancia, qué recurso cuántico se distribuyó y qué limitaciones permanecen.
</div>


<a id="2"></a>
# 2. Aritmética reversible


Una computadora clásica puede borrar bits y descartar resultados intermedios. Las operaciones unitarias de un circuito cuántico, en cambio, son reversibles. Para implementar una función clásica se conserva información suficiente para reconstruir la entrada:

$$
\lvert x\rangle\lvert 0\rangle \longmapsto \lvert x\rangle\lvert f(x)\rangle
$$

Esto introduce **qubits auxiliares**, resultados temporales y la necesidad de hacer *uncomputation*: ejecutar de forma inversa parte del circuito para devolver los auxiliares a $\lvert 0\rangle$ sin borrar el resultado útil.

En circuitos grandes no basta con preguntar si una suma funciona. También importan el número de qubits, la cantidad de compuertas, la profundidad, el número de compuertas Toffoli y, en un esquema tolerante a fallos, el costo de [estados mágicos](https://pennylane.ai/glossary/what-are-magic-states) y compuertas $T$. Por eso la aritmética reversible sigue siendo un área activa de diseño y optimización.


| Operación aritmética | ¿Dónde aparece? |
|---|---|
| Suma y resta | Contadores, comparaciones, multiplicación y aritmética modular |
| Comparación | Oráculos, búsqueda y optimización |
| Multiplicación modular | Algoritmo de Shor |
| Rotaciones controladas por datos | Simulación, estimación de amplitud y métodos numéricos |


<a id="2.1"></a>
## 2.1. Redes cuánticas para aritmética elemental


### Vedral, Barenco y Ekert: una biblioteca aritmética reversible

En 1996, Vedral, Barenco y Ekert presentaron [redes cuánticas para suma, comparación y aritmética modular](https://arxiv.org/abs/quant-ph/9511018). Su trabajo mostró cómo ensamblar operaciones elementales hasta obtener la exponenciación modular necesaria en el algoritmo de Shor.

La aportación importante no fue afirmar que sumar en una computadora cuántica sea mejor que hacerlo clásicamente. Fue demostrar que una función aritmética compleja puede descomponerse en compuertas reversibles y estimar los recursos de esa construcción. Esta forma de diseñar por módulos sigue siendo esencial en compiladores y estimaciones de algoritmos tolerantes a fallos.

La siguiente imagen representa un sumador completo reversible. Los bits de entrada no desaparecen y las compuertas controladas producen la suma y el acarreo.

## 2.2. Sumadores con acarreo en cascada


### El sumador de Cuccaro, Draper, Kutin y Moulton

El artículo [*A New Quantum Ripple-Carry Addition Circuit*](https://arxiv.org/abs/quant-ph/0410184), publicado como prepublicación en 2004, introdujo un sumador lineal en el número de bits que necesita solamente un qubit auxiliar. El acarreo avanza mediante bloques de mayoría y después se recorre el circuito en sentido inverso para producir la suma y limpiar información temporal.

Este diseño es importante porque ejemplifica una regla práctica: dos circuitos que calculan la misma función pueden tener costos cuánticos muy distintos. La clase [`CDKMRippleCarryAdder`](https://quantum.cloud.ibm.com/docs/en/api/qiskit/qiskit.circuit.library.CDKMRippleCarryAdder) de Qiskit implementa esta familia de circuitos.

Los sumadores con acarreo en cascada son fáciles de entender y usan pocos auxiliares, aunque su profundidad crece linealmente. Cuando la profundidad importa más que el número de qubits pueden preferirse otros diseños, como sumadores basados en la transformada cuántica de Fourier o estructuras de acarreo anticipado.


<a id="2.3"></a>
## 2.3. Aritmética modular y estimaciones para RSA-2048


### De los bloques aritméticos al algoritmo de Shor

El [algoritmo de Shor](https://doi.org/10.1137/S0097539795293172) reduce la factorización a un problema de búsqueda de periodo. La parte cuántica necesita evaluar de manera coherente una exponenciación modular, que a su vez contiene multiplicaciones, sumas, comparaciones y operaciones de limpieza reversible.

Si $n = \lceil \log_2 N \rceil$ es el número de bits del entero, Shor requiere un número de operaciones polinomial en $n$. Con aritmética elemental, suele resumirse como $O(n^3)$. En contraste, el mejor algoritmo clásico general conocido, la criba general del cuerpo de números, tiene una complejidad subexponencial en $N$, pero superpolinomial en $n$, aproximadamente $\exp\left(O\left(n^{1/3}(\log n)^{2/3}\right)\right)$.

La conclusión importante es que **sí existe una ganancia algorítmica muy grande**. A medida que aumenta el tamaño del número, el tiempo del mejor método clásico conocido crece mucho más rápido que el de Shor. Por eso, para enteros suficientemente grandes, una computadora cuántica tolerante a fallos podría factorizar en un tiempo razonable números que serían imprácticos para una computadora clásica. A esta diferencia se le suele llamar una ventaja superpolinomial y, de manera menos precisa, una ventaja exponencial.

La [explicación de IBM Quantum sobre el algoritmo de Shor](https://quantum.cloud.ibm.com/learning/en/modules/computer-science/shors-algorithm) presenta de forma accesible por qué la búsqueda cuántica del orden cambia esta escala. Sin embargo, la ventaja es asintótica: aparece al considerar problemas grandes y no implica que los dispositivos cuánticos actuales sean más rápidos para factorizar. Además, la comparación entre algoritmos no incluye el gran costo físico de construir qubits lógicos y aplicar corrección de errores.

En 2021, Gidney y Ekerå publicaron una [estimación detallada para factorizar RSA-2048](https://doi.org/10.22331/q-2021-04-15-433). El resultado de 20 millones de qubits físicos y unas ocho horas no fue una ejecución real, sino una estimación bajo supuestos concretos de corrección de errores y hardware. Gran parte de la mejora frente a estimaciones anteriores provino de rediseñar la aritmética y organizar mejor los recursos tolerantes a fallos.

El ejemplo muestra por qué estudiar sumadores no es un fin aislado: una reducción pequeña en un bloque que se repite millones de veces puede cambiar de manera considerable el costo de un algoritmo completo.

# 3. Comunicación cuántica


<div align="center"><img width="900px" style="display:block; margin-left: auto; margin-right: auto;" src="https://images.claudiazm.xyz/ecc26/QuantumNetworkAnatomy.gif"></div>

La comunicación cuántica utiliza sistemas físicos cuánticos, normalmente fotones, para distribuir estados o establecer correlaciones entre lugares separados. Sus objetivos incluyen distribución de claves, teleportación de estados, conexión de sensores, acceso remoto a procesadores y cómputo cuántico distribuido.

**Una red cuántica no pretende reemplazar el Internet clásico**. Necesita comunicación clásica para coordinar intentos, anunciar detecciones, comparar bases, corregir errores y decidir qué operaciones deben realizarse. El canal cuántico agrega capacidades que no pueden reproducirse simplemente copiando bits clásicos.


<a id="3.1"></a>
## 3.1. ¿Qué se comunica?


En una red cuántica pueden circular o establecerse distintos recursos:

| Recurso o tarea | Descripción | Ejemplo |
|---|---|---|
| Qubits viajeros | Un sistema cuántico se transmite directamente, a menudo mediante un fotón | Enviar un qubit fotónico por fibra |
| Entrelazamiento | Dos nodos comparten un estado que no puede describirse como estados independientes | Crear un par de Bell remoto |
| Teleportación | Un estado se transfiere consumiendo entrelazamiento y comunicación clásica | Transferir el estado de una memoria a otra |
| Claves cuánticas | Se distribuyen bits secretos cuya seguridad puede detectar intercepción | QKD entre usuarios |
| Operaciones remotas | Una compuerta o cálculo se realiza entre qubits alojados en módulos distintos | Cómputo cuántico distribuido |

No todos los sistemas llamados _red cuántica_ ofrecen las mismas funciones. Una red de [QKD](https://en.wikipedia.org/wiki/Quantum_key_distribution) puede distribuir claves sin disponer de memorias cuánticas o entrelazamiento de extremo a extremo. Una Internet cuántica general requeriría además almacenar, transformar y encaminar estados cuánticos entre nodos heterogéneos.


<a id="3.2"></a>
## 3.2. El entrelazamiento como recurso


El entrelazamiento puede entenderse como un recurso que se crea, distribuye, almacena y consume. Un par de Bell compartido proporciona un **e-bit**, una unidad de entrelazamiento útil para protocolos como teleportación y codificación superdensa.

En una red, los fotones pueden transportar entrelazamiento entre memorias estacionarias. Una detección anunciada, o *heralding*, informa que el intento tuvo éxito. Mediante **intercambio de entrelazamiento**, dos enlaces cortos pueden convertirse en uno más largo sin que los nodos extremos hayan interactuado directamente. Si los pares tienen ruido, la purificación o destilación intenta obtener menos pares, pero de mayor calidad.

El entrelazamiento no permite enviar mensajes más rápido que la luz. Los resultados locales parecen aleatorios y se necesita comunicación clásica para completar protocolos como la teleportación. Su utilidad consiste en habilitar correlaciones y tareas distribuidas que no están disponibles con recursos exclusivamente clásicos.


<a id="3.3"></a>
## 3.3. ¿Qué necesita una red cuántica?


Una red cuántica funcional combina varios componentes:

1. **Nodos:** memorias, sensores o procesadores capaces de controlar qubits locales
2. **Interfaces fotónicas:** convierten información entre qubits estacionarios y fotones viajeros
3. **Canales:** fibra óptica o enlaces de espacio libre
4. **Fuentes y detectores:** generan y registran fotones individuales o pares entrelazados
5. **Memorias cuánticas:** conservan estados mientras otros enlaces tienen éxito
6. **Repetidores cuánticos:** extienden la distancia mediante intercambio, purificación y corrección de errores, sin copiar qubits desconocidos
7. **Control clásico:** sincroniza nodos, anuncia resultados y administra recursos
8. **Protocolos y estándares:** permiten que tecnologías diferentes trabajen juntas

La pérdida de fotones crece con la distancia y no puede solucionarse con amplificadores clásicos, porque amplificar implicaría copiar estados desconocidos. Este es uno de los motivos por los que construir repetidores y memorias útiles es mucho más difícil que tender una fibra adicional.

# 4. Hitos experimentales


Los siguientes experimentos no forman una sola red que haya crecido de manera continua. Son hitos complementarios que resolvieron problemas distintos: transferir estados, cerrar lagunas experimentales, aumentar la distancia, añadir memorias, conectar más nodos y ejecutar operaciones distribuidas.


<a id="4.1"></a>
## 4.1. Teleportación cuántica experimental

### Bouwmeester y colaboradores, 1997


<div align="center"><a href="https://quantum.cloud.ibm.com/learning/en/courses/basics-of-quantum-information/entanglement-in-action/quantum-teleportation" target="_blank"><img width="560px" style="display:block; margin-left:auto; margin-right:auto;" alt="Circuito conceptual de teleportación cuántica" src="https://quantum.cloud.ibm.com/learning/images/courses/basics-of-quantum-information/entanglement-in-action/teleportation.svg"></a></div>

<p style="font-size:10px; font-style:italic; text-align:right;"><a href="https://quantum.cloud.ibm.com/learning/en/courses/basics-of-quantum-information/entanglement-in-action/quantum-teleportation" target="_blank">Imagen: curso de IBM Quantum sobre teleportación</a></p>

El protocolo teórico de Bennett y colaboradores apareció en 1993. Cuatro años después, Bouwmeester y colaboradores reportaron la [primera demostración experimental de teleportación cuántica](https://doi.org/10.1038/37539) con fotones.

El experimento no transportó materia. Transfirió el estado de polarización de un fotón utilizando un par entrelazado, una medición conjunta y comunicación clásica. Fue una demostración probabilística y de laboratorio, pero estableció que la teleportación podía implementarse físicamente y no era solo una identidad matemática.

**Por qué importa:** la teleportación permite mover estados entre memorias y realizar operaciones remotas sin enviar directamente el qubit que almacenaba el estado original.

<a id="4.2"></a>
## 4.2. Entrelazamiento remoto y prueba de Bell sin lagunas

### Hensen y colaboradores, 2015

<div align="center"><a href="https://doi.org/10.1038/nature15759" target="_blank"><img width="720px" style="display:block; margin-left:auto; margin-right:auto;" alt="Configuración del experimento de Bell entre nodos separados" src="https://media.springernature.com/full/springer-static/image/art%3A10.1038%2Fnature15759/MediaObjects/41586_2015_Article_BFnature15759_Fig1_HTML.jpg"></a></div>

<p style="font-size:10px; font-style:italic; text-align:right;"><a href="https://doi.org/10.1038/nature15759" target="_blank">Figura 1 del artículo de Hensen y colaboradores, Nature</a></p>

Hensen y colaboradores entrelazaron espines electrónicos asociados con centros nitrógeno-vacante en diamante, alojados en laboratorios separados por 1.3 km. El [experimento de 2015](https://doi.org/10.1038/nature15759) realizó una prueba de Bell sin las principales lagunas de localidad y detección.

Este trabajo fue importante tanto para los fundamentos de la mecánica cuántica como para las redes. Mostró entrelazamiento anunciado entre qubits de materia remotos, una capacidad necesaria para conectar memorias. Además, las pruebas de Bell sin lagunas sustentan protocolos independientes del dispositivo, en los que la seguridad o la aleatoriedad pueden certificarse a partir de las correlaciones observadas.

**Qué no implica:** no cumplir una desigualdad de Bell no permite comunicación instantánea. Para reconocer y utilizar los eventos exitosos sigue siendo necesario intercambiar información clásica.

## 4.3. Entrelazamiento distribuido por satélite

### El satélite Micius, 2017

<div align="center"><a href="https://www.bcas.cas.cn/sr/micro/201903/t20190301_205881.html" target="_blank"><img width="720px" style="display:block; margin-left:auto; margin-right:auto;" alt="Esquema de comunicación cuántica mediante el satélite Micius" src="https://www.bcas.cas.cn/sr/micro/201903/W020190301389088378437.jpg"></a></div>

<p style="font-size:10px; font-style:italic; text-align:right;"><a href="https://www.bcas.cas.cn/sr/micro/201903/t20190301_205881.html" target="_blank">Imagen: Bulletin of the Chinese Academy of Sciences</a></p>

Yin y colaboradores distribuyeron pares de fotones entrelazados desde el satélite Micius hacia estaciones terrestres separadas por hasta 1 203 km. El [artículo publicado en Science](https://doi.org/10.1126/science.aan3211) mostró que los enlaces espaciales pueden superar algunas limitaciones de pérdida que aparecen al intentar cubrir distancias continentales únicamente con fibra.

**Por qué importa:** los satélites pueden conectar regiones muy distantes y complementar futuras redes terrestres. El experimento no creó por sí solo una Internet cuántica global: dependía de ventanas de visibilidad, seguimiento óptico preciso, condiciones atmosféricas y una infraestructura especializada.

<a id="4.4"></a>
## 4.4. Red integrada de comunicación cuántica de 4 600 km

### Fibra, enlaces satelitales y QKD, 2021

<div align="center"><a href="https://doi.org/10.1038/s41586-020-03093-8" target="_blank"><img width="760px" style="display:block; margin-left:auto; margin-right:auto;" alt="Arquitectura de una red integrada de comunicación cuántica por fibra y satélite" src="https://media.springernature.com/full/springer-static/image/art%3A10.1038%2Fs41586-020-03093-8/MediaObjects/41586_2020_3093_Fig1_HTML.png"></a></div>

<p style="font-size:10px; font-style:italic; text-align:right;"><a href="https://doi.org/10.1038/s41586-020-03093-8" target="_blank">Figura 1 del artículo de Chen y colaboradores, Nature</a></p>

Chen y colaboradores integraron más de 700 enlaces de QKD por fibra con dos enlaces de alta velocidad entre satélite y tierra. La [red descrita en Nature](https://doi.org/10.1038/s41586-020-03093-8) conectó a más de 150 usuarios y alcanzó una distancia total de 4 600 km.

Este ejemplo demuestra que la comunicación cuántica puede integrarse con infraestructura operativa y atender numerosos usuarios. Sin embargo, la red utilizó **relevos confiables**: las claves se recuperaban y volvían a distribuir en nodos intermedios. Por ello es un gran hito de QKD, pero no equivale todavía a una red de repetidores que mantenga entrelazamiento de extremo a extremo.

<a id="4.5"></a>
## 4.5. Una red cuántica de tres nodos

### Pompili y colaboradores, 2021

<div align="center"><a href="https://qutech.nl/lab/hanson-lab/research-highlights/multinode-quantum-network/" target="_blank"><img width="760px" style="display:block; margin-left:auto; margin-right:auto;" alt="Representación de una red cuántica de tres nodos" src="https://qutech.nl/wp-content/uploads/2021/04/cover4.png?t=1617901730"></a></div>

<p style="font-size:10px; font-style:italic; text-align:right;"><a href="https://qutech.nl/lab/hanson-lab/research-highlights/multinode-quantum-network/" target="_blank">Ilustración de Matteo Pompili, QuTech y TU Delft</a></p>

Pompili y colaboradores construyeron una red con tres nodos, llamados Alice, Bob y Charlie, basados en centros nitrógeno-vacante en diamante. El nodo central disponía de una memoria adicional, lo que permitió conservar entrelazamiento con un extremo mientras se generaba el enlace con el otro. El trabajo se publicó como [*Realization of a Multinode Quantum Network of Remote Solid-State Qubits*](https://doi.org/10.1126/science.abg1919).

La red realizó distribución de entrelazamiento entre nodos no conectados directamente y una demostración de teleportación. Esto fue un paso importante desde los enlaces punto a punto hacia protocolos en los que un nodo intermedio administra recursos cuánticos.

**Limitación:** era una red experimental pequeña, con tasas y fidelidades aún lejos de las necesarias para un servicio general.

<a id="4.6"></a>
## 4.6. Memorias nanofotónicas en una red de telecomunicaciones

### Knaut y colaboradores, 2024

<div align="center"><a href="https://doi.org/10.1038/s41586-024-07252-z" target="_blank"><img width="760px" style="display:block; margin-left:auto; margin-right:auto;" alt="Red de dos nodos con memorias cuánticas nanofotónicas" src="https://media.springernature.com/full/springer-static/image/art%3A10.1038%2Fs41586-024-07252-z/MediaObjects/41586_2024_7252_Fig1_HTML.png"></a></div>

<p style="font-size:10px; font-style:italic; text-align:right;"><a href="https://doi.org/10.1038/s41586-024-07252-z" target="_blank">Figura 1 del artículo de Knaut y colaboradores, Nature</a></p>

En el trabajo, [*Entanglement of Nanophotonic Quantum Memory Nodes in a Telecom Network*](https://doi.org/10.1038/s41586-024-07252-z), se conectaron dos nodos de memoria basados en centros de vacancia de silicio en diamante. Se demostró entrelazamiento mediante carretes con 40 km de fibra de baja pérdida y mediante un circuito de 35 km de fibra de telecomunicaciones desplegada en el área urbana de Boston.

Los nodos combinan qubits de distinta naturaleza. Los **qubits fotónicos**, codificados en intervalos temporales, viajan por la fibra y actúan como portadores. En cada nodo, un **centro de vacancia de silicio** (SiV) dentro de una cavidad nanofotónica de diamante proporciona qubits de materia: el espín electrónico funciona como interfaz de comunicación y un espín nuclear cercano funciona como memoria de mayor duración. Un centro SiV es un defecto controlado de la red cristalina, asociado con un átomo de silicio y vacancias vecinas; no es simplemente un hueco macroscópico en el diamante.

Una dificultad central es que las transiciones ópticas convenientes para una memoria no siempre coinciden con las longitudes de onda de baja pérdida utilizadas en telecomunicaciones. El experimento empleó conversión de frecuencia y detección anunciada para generar entrelazamiento remoto, además de transferirlo a espines nucleares con mayor tiempo de almacenamiento.

**Por qué importa:** reúne memorias, nanofotónica, conversión a banda de telecomunicaciones y fibra instalada fuera de un montaje puramente local. Son precisamente las interfaces que una red útil debe hacer compatibles.

<a id="4.7"></a>
## 4.7. Cómputo cuántico distribuido mediante un enlace óptico

### Main y colaboradores, 2025

<div align="center"><a href="https://doi.org/10.1038/s41586-024-08404-x" target="_blank"><img width="760px" style="display:block; margin-left:auto; margin-right:auto;" alt="Arquitectura de cómputo cuántico distribuido con módulos conectados ópticamente" src="https://media.springernature.com/full/springer-static/image/art%3A10.1038%2Fs41586-024-08404-x/MediaObjects/41586_2024_8404_Fig1_HTML.png"></a></div>

<p style="font-size:10px; font-style:italic; text-align:right;"><a href="https://doi.org/10.1038/s41586-024-08404-x" target="_blank">Figura 1 del artículo de Main y colaboradores, Nature</a></p>

Main y colaboradores conectaron dos módulos de iones atrapados mediante fotones y utilizaron entrelazamiento remoto para teleportar una compuerta controlada entre ellos. Después ejecutaron una instancia distribuida del algoritmo de Grover. El resultado se publicó en [*Distributed Quantum Computing across an Optical Network Link*](https://doi.org/10.1038/s41586-024-08404-x).

Este experimento muestra otra función posible de una red: no solo comunicar usuarios, sino unir procesadores pequeños para ejecutar un circuito que atraviesa varios módulos. La modularidad podría ayudar a escalar computadoras cuánticas cuando fabricar un único procesador enorme resulte impráctico.

**Limitación:** se trató de una demostración con dos módulos y una escala pequeña. Convertir esta idea en una arquitectura grande requiere mejorar mucho la velocidad, la fidelidad y la administración de enlaces.

<a id="5"></a>
# 5. Juegos no locales


En un juego no local, dos participantes reciben preguntas separadas y deben producir respuestas sin comunicarse durante la ronda. Antes de comenzar pueden acordar una estrategia y compartir recursos. En el **juego CHSH**, cualquier estrategia clásica gana como máximo el 75 % de las rondas. Si los participantes comparten entrelazamiento y eligen mediciones adecuadas, pueden alcanzar

$$
\cos^2\left(\frac{\pi}{8}\right) \approx 0.8536
$$

La ventaja no aparece porque un participante envíe información al otro, sino porque sus respuestas presentan correlaciones que no admiten una explicación local clásica. Las [desigualdades CHSH](https://doi.org/10.1103/PhysRevLett.23.880) convierten esta idea en una prueba cuantitativa.

Estos juegos son relevantes para:

* estudiar la potencia y los límites del entrelazamiento
* diseñar pruebas de Bell
* certificar aleatoriedad
* construir protocolos criptográficos independientes del dispositivo
* verificar dispositivos o cálculos sin confiar completamente en su implementación interna

Su papel en este notebook es conceptual: explican por qué experimentos como el de Hensen son más que una demostración de correlaciones y cómo una red puede certificar propiedades de sus nodos a partir de estadísticas observables.

Para llevar la idea a un experimento, el tutorial oficial [Demostración de la desigualdad CHSH con Qiskit](https://quantum.cloud.ibm.com/docs/tutorials/chsh-inequality) muestra cómo construir los circuitos, ejecutarlos y comparar el resultado con el límite clásico.


<a id="6"></a>
# 6. Hacia el Internet cuántico


La expresión **Internet Cuántico** describe una red capaz de distribuir y procesar información cuántica entre nodos. Una visión influyente fue presentada por Kimble en 2008 y posteriormente organizada por Wehner, Elkouss y Hanson en etapas de capacidad creciente.

Una posible evolución incluye:

1. redes de preparación y medición, como muchas implementaciones de QKD
2. distribución de entrelazamiento entre nodos cercanos
3. memorias y repetidores que extiendan el entrelazamiento
4. teleportación confiable de estados y compuertas
5. cómputo, sensado y verificación distribuidos a gran escala

Los experimentos anteriores cubren piezas de ese rompecabezas, pero todavía no existe un Internet cuántico general, global y tolerante a fallos. Las tasas de generación de entrelazamiento son limitadas, las memorias pierden coherencia, la conversión entre tecnologías añade ruido y los repetidores cuánticos completos continúan en desarrollo.

También se necesitan protocolos. El [RFC 9340](https://www.rfc-editor.org/rfc/rfc9340.html), publicado por el Quantum Internet Research Group, presenta principios arquitectónicos para un Internet cuántico y ayuda a separar la visión de largo plazo de una colección de enlaces experimentales.


| Lo que ya se ha demostrado | Lo que todavía falta |
|---|---|
| QKD en redes metropolitanas, fibra de larga distancia y enlaces satelitales | Entrelazamiento global de extremo a extremo sin depender de nodos confiables |
| Entrelazamiento anunciado entre memorias remotas | Tasas altas y fidelidades suficientes durante operación continua |
| Redes experimentales de tres nodos | Enrutamiento entre muchos nodos y tolerancia a fallos |
| Conversión entre emisores cuánticos y banda de telecomunicaciones | Interfaces eficientes y estandarizadas entre plataformas heterogéneas |
| Teleportación de estados y compuertas remotas | Cómputo distribuido grande con corrección de errores |
| Pruebas de Bell sin las principales lagunas | Dispositivos independientes del dispositivo prácticos y ampliamente desplegados |

# 7. Conclusiones


La aritmética reversible muestra cómo una operación familiar se transforma en un bloque compatible con la evolución cuántica. Su relevancia aparece cuando esos bloques se repiten dentro de algoritmos como el de Shor y dominan el costo total.

La comunicación cuántica muestra con mayor claridad que el entrelazamiento es un recurso operativo. Puede distribuirse entre memorias, consumirse para teleportar estados o compuertas y certificarse mediante pruebas de Bell y juegos no locales.

Todavía estamos en una etapa temprana: los nodos son pocos, las tasas son bajas y muchas demostraciones dependen de condiciones experimentales exigentes. Al mismo tiempo, los ejemplos revisados son investigaciones reproducibles, publicadas y cada vez más integradas con fibra desplegada, satélites y procesadores modulares. Estudiar computación cuántica permite participar en la construcción de las capas algorítmicas, físicas y de software que esas redes necesitarán.
